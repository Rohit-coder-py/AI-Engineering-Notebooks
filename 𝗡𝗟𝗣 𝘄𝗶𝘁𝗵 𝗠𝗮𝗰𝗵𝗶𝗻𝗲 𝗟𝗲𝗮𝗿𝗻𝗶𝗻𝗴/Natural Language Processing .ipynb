# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">Step 2 — Text Cleaning</span>

<br>

### <span style="color:#8E44AD;">1. Converting Everything to Lowercase</span>

Convert text to lowercase **as per the use case**.

---

### <span style="color:#E67E22;">2. Remove Punctuation</span>

Remove symbols like `. , ! ? @ # $ % ^ & * ( )` etc.

> In traditional ML-based NLP, punctuation is often removed because it may not add much value.

However, in **Deep Learning-based NLP**, especially with models like **BERT, GPT, and T5**, punctuation can carry meaning, so we usually **keep it during preprocessing**.

---

### <span style="color:#27AE60;">3. Remove Numbers</span>

Remove numbers **depending on the use case**.

---

### <span style="color:#2980B9;">4. Remove URLs / Links</span>

Remove links from the text. This is very common when working with **social media data**.

---

### <span style="color:#C0392B;">5. Remove HTML Tags</span>

When performing **web scraping**, remove HTML tags from the collected text.

---

### <span style="color:#7D3C98;">6. Remove Emojis & Special Characters</span>

Remove emojis and unwanted special characters **depending on the use case**.

---

### <span style="color:#D35400;">7. Remove Stopwords</span>

Remove common words like **"is", "the", "was", "and", "in"**, etc.

These words often don't help in understanding the meaning of the text.

---

### <span style="color:#16A085;">8. Optional — Correct Spelling</span>

Correct spelling using Python libraries or other tools.

<strong style="color:#2980B9;">Python Libraries:</strong>

- <strong>TextBlob</strong>
- <strong>pyspellchecker</strong>
- <strong>SymSpell</strong>

<strong style="color:#8E44AD;">Other Ways:</strong>

- Language models such as **GPT**
- Dictionary-based correction
- Custom spelling dictionaries

> **Text Cleaning = Removing or modifying unwanted parts of text before NLP processing.**
