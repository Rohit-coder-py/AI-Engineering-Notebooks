<strong style="color:#2471A3;">Why do we need Feature Extraction?</strong>

Text is not numbers.

But ML models only understand <strong style="color:#C0392B;">numbers</strong> — not feelings, not words, not emojis.

So, we need to convert words into numbers in a smart way. That's called <strong style="color:#8E44AD;">Feature Extraction</strong>.

</div>

### <span style="color:#27AE60;">ML vs Deep Learning-based NLP</span>

In <strong style="color:#2980B9;">Machine Learning-based NLP</strong>, we mostly use traditional text vectorization techniques like <strong>Bag of Words (BoW)</strong> and <strong>TF-IDF</strong> to convert text into numerical format.

On the other hand, <strong style="color:#8E44AD;">Deep Learning-based NLP</strong> relies on more advanced techniques called <strong>embeddings</strong> like <strong>Word2Vec, GloVe, and FastText</strong>, which convert words into dense vectors that understand relationships and context.

For example, they can learn that the words <strong style="color:#C0392B;">"king"</strong> and <strong style="color:#C0392B;">"queen"</strong> are semantically connected instead of treating them as totally unrelated.

## <span style="color:#8E44AD; font-weight:800;">Corpus</span>

A <strong>corpus</strong> is just a collection of text that we use in NLP.

It can be a bunch of sentences, poems, reviews, or any written content. Think of it like all the text your model will read and learn from.

> **Example:** If you're analyzing 100 movie reviews, then those 100 reviews are called your **corpus**.

---

## <span style="color:#2980B9; font-weight:800;">Document</span>

A <strong>document</strong> in NLP is just a single piece of text inside your dataset.

It can be one sentence, one paragraph, or even a full article — it depends on your project.

> **Example:** If your corpus has 100 movie reviews, then each review is called **one document**.

---

## <span style="color:#27AE60; font-weight:800;">Vocabulary</span>

A <strong>vocabulary</strong> in NLP means the list of all unique words present in your corpus.

It's like a dictionary your model uses to understand and convert words into numbers.

> **Example:** If your dataset has the words `"I love pizza"` and `"I love pasta"`, then the vocabulary will be:

```python
["I", "love", "pizza", "pasta"]

## <span style="color:#8E44AD; font-weight:800;">2. Bag of Words </span>



<div style="

<strong style="color:#8E44AD;">Bag of Words</strong> is a technique used to convert <strong>text into numbers</strong> so that machine learning models can understand it.

It creates a list of all the <strong style="color:#2980B9;">unique words</strong> in your dataset (called <strong>vocabulary</strong>), and then for each sentence, it counts how many times each word appears.

</div>

### <span style="color:#E67E22;">Key Idea</span>

It doesn't care about <strong>grammar</strong> or the <strong>order of words</strong> — only the <strong style="color:#C0392B;">word frequency</strong> matters.

> **Bag = Jumbled, unordered collection of words.**

`````python
#code implementation of BOW (bag of words) 

from sklearn.feature_extraction.text import CountVectorizer
CountVectorizer(stop_words='english')

documents = [
    "I love machine learning",
    "Machine learning is powerful",
    "I love Python",
    "Python is powerful"
]


vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)


print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(X.toarray())

bow = X.toarray()

bow.shape
`````

## <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">What does CountVectorizer do?</span>

<br>

<div style="
background-color:#EBF5FB;
border-left:5px solid #3498DB;
padding:14px 18px;
margin:10px 0;
border-radius:5px;
">

<strong style="color:#2471A3;">CountVectorizer</strong> converts text into a <strong>numerical matrix</strong> based on how many times each word appears.

</div>

### <span style="color:#27AE60; font-weight:700;">What it does</span>

- <strong>Lowercase</strong>
- <strong>Tokenize</strong>
- <strong>Build vocabulary</strong>
- <strong>Count word occurrences</strong>
- <strong>Convert text → numerical matrix</strong>

### <span style="color:#C0392B; font-weight:700;">What it does NOT do</span>

- Stopword removal <strong>by default</strong>
- Stemming
- Lemmatization
- Semantic understanding

---

## <span style="color:#8E44AD; font-weight:700;">Parameters we can use</span>

```python
x = CountVectorizer(
    stop_words='english',
    max_features=1000,
    min_df=2,
    max_df=0.9,
    ngram_range=(1, 2),
    binary=True
)
````

<br>

| Parameter              | Purpose                                               |
| ---------------------- | ----------------------------------------------------- |
| `stop_words='english'` | Removes English stopwords                             |
| `max_features=1000`    | Keeps at most 1000 features                           |
| `min_df=2`             | Ignores words appearing in fewer than 2 documents     |
| `max_df=0.9`           | Ignores words appearing in more than 90% of documents |
| `ngram_range=(1, 2)`   | Uses **unigrams + bigrams**                           |
| `binary=True`          | Uses **0/1 presence** instead of word counts          |

## <span style="color:#2C3E50; font-weight:800; border-bottom:2px solid #3498DB; padding-bottom:4px;">Bag of Words (BoW) — Pros & Cons</span>

<br>

### <span style="color:#27AE60; font-weight:800;">Pros</span>

1. <strong style="color:#2980B9;">Easy to Implement</strong>
2. <strong style="color:#8E44AD;">Fixed Size</strong>
3. <strong style="color:#D35400;">Works with ML</strong>
4. <strong style="color:#16A085;">Fast & Efficient</strong>

### <span style="color:#C0392B; font-weight:800;">Cons</span>

1. <strong style="color:#E67E22;">Sparse Matrix</strong>
2. <strong style="color:#8E44AD;">OOV (Out of Vocabulary)</strong>
3. <strong style="color:#2980B9;">No Semantic Meaning</strong> — Better than OHE
4. <strong style="color:#C0392B;">Out of Order</strong>

<mark><b><strong> What is N-grams?

N-gram is a sequence of N consecutive words (or tokens) taken from a sentence.

<b>For example:

<strong style="color:#8E44AD;">"I love machine learning"</strong>


| N     | Name    | N-grams                                      |
| ----- | ------- | -------------------------------------------- |
| **1** | Unigram | `I`, `love`, `machine`, `learning`           |
| **2** | Bigram  | `I love`, `love machine`, `machine learning` |
| **3** | Trigram | `I love machine`, `love machine learning`    |
| **4** | 4-gram  | `I love machine learning`                    |


> **n-Gram** can be called as **Bag of N-Grams**.

### ✅ Pros

1. **Semantic Meaning**
   - A little bit

### ❌ Cons

1. **Dimensions Increase**
2. **Out of Vocabulary (OOV)**

`````python
# Code implementation of BOW (Bag of Words)

from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love machine learning",
    "Machine learning is powerful",
    "I love Python",
    "Python is powerful"
]

vectorizer = CountVectorizer(
    stop_words='english',
    ngram_range=(2, 2)
)

X = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(X.toarray())

bow = X.toarray()

print("\nShape:")
print(bow.shape)
`````

## <span style="color:#27AE60; font-weight:800;">3. TF-IDF</span>


**TF = <span style="color:#2980B9;">Term Frequency</span>**

**IDF = <span style="color:#8E44AD;">Inverse Document Frequency</span>**

**TF = <span style="color:#2980B9;">Term Frequency</span>**

<strong style="color:#2471A3;">TF (Term Frequency)</strong> tells us <strong>how frequently a word appears in a particular document.</strong>

> <strong>TF = Number of times a term appears in a document / Total number of terms in that document</strong>

### <span style="color:#D35400;">Example</span>

<strong>Document:</strong>  
`"I love machine learning and I love Python"`

The word <strong style="color:#8E44AD;">"love"</strong> appears <strong>2 times</strong> and there are <strong>7 words</strong>.

<b style="color:#27AE60;">TF(love) = 2 / 7 = 0.286</b>

<br>

<b><mark>Term Frequency Formula :</mark></b>

<div align="center">

$$
\boxed{
TF =
\frac{\text{No. of Occurrence of Term in Document}}
{\text{Total No. of Terms in Document}}
}
$$

</div>

### <span style="color:#8E44AD; font-weight:800;">IDF — Inverse Document Frequency</span>

<strong style="color:#6C3483;">IDF (Inverse Document Frequency)</strong> tells us <strong>how rare or important a word is across all documents.</strong>

> A word appearing in <strong style="color:#C0392B;">many documents</strong> → <strong style="color:#E67E22;">low IDF</strong>
>
> A word appearing in <strong style="color:#27AE60;">few documents</strong> → <strong style="color:#2980B9;">high IDF</strong>

### <span style="color:#2980B9;">Common Formula</span>

<strong>IDF(term) = log(Total number of documents / Number of documents containing the term)</strong>

### <span style="color:#D35400;">Example</span>

Suppose we have <strong>4 documents</strong> and the word <strong>"Python"</strong> appears in <strong>1 document</strong>:

<b style="color:#8E44AD;">IDF(Python) = log(4 / 1) = log(4) ≈ 1.386</b>

But <strong>"learning"</strong> appears in <strong>3 documents</strong>:

<b style="color:#2980B9;">IDF(learning) = log(4 / 3) ≈ 0.288</b>

> <strong style="color:#27AE60;">Rare word → High IDF → More important</strong>
>
> <strong style="color:#C0392B;">Common word → Low IDF → Less important</strong>

<br>

<b><mark>Inverse Document Frequency Formula :</mark></b>

<div align="center">

$$
\boxed{
IDF =
\log\left(
\frac{\text{Total No. of Documents in Corpus}}
{\text{No. of Documents Containing the Term}}
\right)
}
$$

</div>

### <span style="color:#8E44AD; font-weight:800;">TF-IDF — Together</span>

<strong style="color:#6C3483;">TF-IDF combines TF and IDF to calculate how important a word is in a particular document.</strong>

### <span style="color:#D35400;">Example</span>

<strong style="color:#2980B9;">TF(love) = 0.25</strong>

<strong style="color:#8E44AD;">IDF(love) = 1.386</strong>

Therefore:

$$
\boxed{
TF\text{-}IDF(love) = TF \times IDF
}
$$

$$
TF\text{-}IDF(love) = 0.25 \times 1.386 = 0.3465
$$

> <strong style="color:#2980B9;">TF</strong> → How frequently does the word appear in this document?

> <strong style="color:#8E44AD;">IDF</strong> → How rare is the word across all documents?

> <strong style="color:#27AE60;">TF-IDF</strong> → Overall importance of the word for that document.

### <span style="color:#D35400;">In one line</span>

<strong>A word gets a high TF-IDF score when it appears frequently in a document but is relatively rare across the entire collection of documents.</strong>

<strong style="color:#C0392B;">Common words across many documents → Low TF-IDF</strong>

<strong style="color:#27AE60;">Rare words in a particular document → High TF-IDF</strong>

## <span style="color:#2C3E50; font-weight:800; border-bottom:2px solid #3498DB; padding-bottom:4px;">Basic Idea of TF-IDF</span>

> <strong style="color:#2980B9;">TF</strong> → How frequently does the word appear in <strong>this document?</strong>

> <strong style="color:#8E44AD;">IDF</strong> → How rare is the word across <strong>all documents?</strong>

> <strong style="color:#27AE60;">TF-IDF</strong> → How important is this word to <strong>this document?</strong>

<br>

<div align="center">

<strong style="color:#27AE60;">TF-IDF = TF × IDF</strong>

</div>

`````python
#code implementation of TF-IDF


from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I enjoy learning Python",
    "Python is easy and powerful",
    "I enjoy machine learning",
    "Machine learning is interesting and powerful"
]


vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)



print("Vocabulary:\n \n",vectorizer.get_feature_names_out(),"\n")

print("TF-IDF Matrix:\n\n",X.toarray())

#  (0, 2)	0.6136667440107333
# (row, column)    TF-IDF value
`````

## <span style="color:#2C3E50; font-weight:800; border-bottom:2px solid #3498DB; padding-bottom:4px;">TF-IDF — Pros & Cons</span>

> <strong style="color:#8E44AD;">TF-IDF (Term Frequency–Inverse Document Frequency)</strong> assigns a higher weight to words that are <strong>important in a document but less common across the entire corpus.</strong>

### <span style="color:#27AE60; font-weight:800;">Pros</span>

1. <strong style="color:#2980B9;">Information Retrieval</strong>
   - Useful for <strong>search engines</strong> to find relevant documents.
   - Helps identify <strong>important words</strong> in a document.

### <span style="color:#C0392B; font-weight:800;">Cons</span>

1. <strong style="color:#E67E22;">Sparsity</strong>
   - Produces vectors with <strong>many zero values</strong>.

2. <strong style="color:#8E44AD;">OOV (Out of Vocabulary)</strong>
   - Cannot properly represent <strong>new/unseen words</strong> that are not in the vocabulary.

3. <strong style="color:#2980B9;">High Dimensions</strong>
   - The vector size can become very large when the corpus has a <strong>large vocabulary</strong>.
