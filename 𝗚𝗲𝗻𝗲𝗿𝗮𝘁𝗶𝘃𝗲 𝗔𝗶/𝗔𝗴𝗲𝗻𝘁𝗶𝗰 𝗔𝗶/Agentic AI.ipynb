# Generative AI vs Agentic AI

> Understanding the difference between **Generative AI** and **Agentic AI** is the first step towards understanding modern AI systems.

---

# 1. Generative AI

### Definition

<div style="background:#E8F1FF; border-left:6px solid #2563EB; padding:16px; border-radius:8px;">

**Generative AI** is a branch of Artificial Intelligence that focuses on **generating new content** based on the input or instructions provided by the user.

It uses trained Machine Learning and Deep Learning models to generate content such as **text, images, audio, video, and code.**

</div>

### How Does It Work?

1. User provides an input (prompt).
2. The input is processed by a Generative AI model.
3. The model understands the context and predicts an appropriate output.
4. The model generates new content.

### Architecture

```text
User Input
    |
    v
Generative AI Model (LLM)
    |
    v
Generated Output
```

### Examples

- **ChatGPT** – Generates text and answers questions.
- **DALL·E** – Generates images.
- **GitHub Copilot** – Generates code.
- **Gemini** – Generates text, images and other content.

### Key Characteristics

- **Content generation**
- **Prompt-based interaction**
- Responds to user instructions.
- Primarily focuses on generating outputs.
- Usually does not independently execute multi-step tasks.

---

# 2. Agentic AI

### Definition

<div style="background:#F3E8FF; border-left:6px solid #9333EA; padding:16px; border-radius:8px;">

**Agentic AI** is an approach to Artificial Intelligence in which AI systems can autonomously work towards achieving a specific goal by **reasoning, planning, making decisions, using tools, and executing multiple steps.**

Unlike traditional Generative AI applications, Agentic AI systems can take actions and adapt their workflow based on the results.

</div>

### How Does It Work?

1. User provides a goal.
2. The AI agent understands the objective.
3. It breaks the objective into smaller tasks.
4. It decides which actions or tools are required.
5. It executes those actions.
6. It evaluates the results and determines the next step.
7. It continues until the goal is achieved or human intervention is required.

### Architecture

```text
       User Goal
           |
           v
      AI Agent (LLM)
           |
           v
     Planning & Reasoning
           |
           v
      Decision Making
           |
           v
       Tool Selection
           |
           v
      Tool Execution
           |
           v
     Observe Results
           |
           v
      Evaluate Results
           |
           v
     Next Action / Finish
```

### Examples

- **Research Agent** – Searches multiple websites and prepares a research report.
- **Coding Agent** – Analyzes code, fixes bugs and executes tests.
- **Personal AI Assistant** – Manages tasks, searches information and performs actions.
- **Customer Support Agent** – Retrieves customer information and performs permitted operations.

### Key Characteristics

- **Goal-oriented behaviour**
- **Autonomous decision-making**
- Planning and reasoning.
- Tool calling and execution.
- Multi-step task execution.
- Feedback and iterative improvement.
- Can incorporate human approval (**Human-in-the-Loop**).

# Difference Between LangChain and LangGraph

> Both **LangChain** and **LangGraph** are frameworks used to build LLM-powered applications, but they handle workflows differently.

## 1. LangChain

<div style="background:#E8F1FF; padding:14px; border-left:6px solid #2563EB; border-radius:8px;">

**LangChain** is a framework used to develop applications powered by LLMs by connecting models, prompts, tools, retrievers, and other components.

It is commonly used to build chains and straightforward AI workflows.

</div>

## 2. LangGraph

<div style="background:#F3E8FF; padding:14px; border-left:6px solid #9333EA; border-radius:8px;">

**LangGraph** is a framework for building stateful, multi-step AI agent workflows using graphs consisting of nodes and edges.

It supports conditional execution, loops, parallel execution, and human-in-the-loop interactions.

</div>

## 3. Key Differences

| Basis | LangChain | LangGraph |
|---|---|---|
| Main Purpose | Build LLM applications | Build complex AI workflows and agents |
| Structure | Chains and components | Nodes and edges |
| Workflow | Often linear | Cyclic and flexible |
| State Management | Available through integrations | Built around shared workflow state |
| Conditional Logic | Supported | Core workflow capability |
| Loops | Less straightforward | Naturally supported |
| Human-in-the-Loop | Possible | Built-in workflow support |
| Best For | Simple LLM applications | Complex, stateful agents |

# The Agent Spectrum

> AI agents can have different levels of capabilities, ranging from simple text generation to fully autonomous systems.

## Levels of AI Agents

| **Level** | **What It Does** | **Example** |
|---|---|---|
| **Level 0 – Pure Generation** | LLM answers a prompt, no tools. | ChatGPT writing a poem. |
| **Level 1 – Tool-Using LLM** | LLM calls a tool once and returns an answer. | LLM checking weather via a tool. |
| **Level 2 – ReAct Agent** | LLM loops: Think → Call Tool → Observe → Think → Call Tool, until done. | Agent checking weather and converting it to Fahrenheit using two tools. |
| **Level 3 – Stateful Agent** | Remembers across turns, has persistent memory and can pause for humans. | Customer support bot remembering past tickets. |
| **Level 4 – Multi-Agent System** | Multiple agents coordinate to achieve a goal, each with its own role. | Planner Agent + Researcher Agent + Writer Agent. |
| **Level 5 – Autonomous Agent** | Long-running, self-correcting, self-improving and plans across days/weeks. | Devin-like coding agent that ships PRs. |

## Understanding the Levels

<div style="background:#E8F1FF; padding:13px; border-left:5px solid #2563EB; border-radius:8px;">

**Level 0–1:** Basic generation and tool usage.

</div>

<div style="background:#F3E8FF; padding:13px; border-left:5px solid #9333EA; border-radius:8px;">

**Level 2–3:** Tool-calling loops, reasoning and persistent state.

</div>

<div style="background:#E8F8EF; padding:13px; border-left:5px solid #16A34A; border-radius:8px;">

**Level 4–5:** Multiple cooperating agents and greater autonomy.

</div>

> **Note:** This is not an officially established or universally accepted agent spectrum. It is a conceptual classification created to understand the different types and capabilities of AI agents.

# 2. Parallel Workflow

### Definition

<div style="background:linear-gradient(135deg,#0C4A6E,#164E63); padding:17px; border-left:7px solid #22D3EE; border-radius:10px; color:#FFFFFF;">

A **Parallel Workflow** is a workflow in which multiple independent tasks are executed **simultaneously (at the same time)** rather than sequentially.

Each task works independently, and their outputs can be combined after completion.

</div>

### Example: Tech News Channel

Imagine you run a tech news channel and a big tech update drops. You need to perform three tasks:

1. **Task 1:** Write a technical news article.
2. **Task 2:** Create a YouTube thumbnail.
3. **Task 3:** Write a catchy tweet for X.

**Sequential Execution:**

```text
Article (30 min)
      |
      v
Thumbnail (30 min)
      |
      v
Tweet (30 min)
      |
      v
Total = 90 Minutes
```

**Parallel Execution:**

```text
          Big Tech Update
                 |
       +---------+---------+
       |         |         |
       v         v         v
    Article  Thumbnail   Tweet
       |         |         |
       +---------+---------+
                 |
                 v
           Final Output
```

All three tasks run simultaneously instead of waiting for each other.

### Key Characteristics

- Multiple tasks execute simultaneously.
- Tasks work independently.
- Saves execution time.
- Outputs can be combined after completion.
- Best suited for independent tasks.

<div style="background:linear-gradient(135deg,#14532D,#166534); padding:15px; border:2px solid #4ADE80; border-radius:9px; color:#FFFFFF;">

**Key Takeaway:**

Sequential = One task after another.

Parallel = Multiple independent tasks at the same time.

</div>

## 2. Parallel Workflow – State Management

### How Does State Work in Parallel Workflow?

<div style="background:linear-gradient(135deg,#164E63,#0F766E); padding:17px; border-left:7px solid #22D3EE; border-radius:10px; color:#FFFFFF;">

In a **Parallel Workflow**, multiple nodes can write to the shared **State simultaneously** when the graph splits.

LangGraph handles these updates safely by allowing each node to write to its own key without overriding or breaking the other node's data.

</div>

### Example

Suppose two nodes are working simultaneously:

- **Node A:** Writes its output to `article_text`.
- **Node B:** Writes its output to `tweet_text`.

Both update different keys in the shared State.

### State Example

```python
class State(TypedDict):
    topic: str
    article_text: str
    tweet_text: str
```

### Visual Representation

```text
             Shared State
                  |
            Graph Splits
             /       \
            v         v
         Node A     Node B
            |         |
            v         v
      article_text  tweet_text
            \         /
             \       /
              v     v
          Updated State
```

> **Key Takeaway:** Multiple nodes can update different keys in the shared State simultaneously without overwriting each other's data.

# 2. Parallel Workflow – Nodes & Edges

## 1. Nodes in Parallel Workflow

<div style="background:linear-gradient(135deg,#172554,#164E63); padding:16px; border-left:6px solid #22D3EE; border-radius:9px; color:white;">

In a Parallel Workflow, we define **separate, specialized functions for each task** that can run independently.

These nodes are ultimately just **Python functions**, each responsible for performing a specific task.

</div>

**Example:**
- Article Node → Writes a news article.
- Thumbnail Node → Creates a thumbnail.
- Tweet Node → Generates a tweet.

---

## 2. Edges in Parallel Workflow

In a Parallel Workflow, edges change from a simple **A → B** pathway into two specialized structural movements:

- **Fan-Out**
- **Fan-In**

### A. Fan-Out (The Splitter)

<div style="background:#E8F1FF; padding:15px; border-left:6px solid #2563EB; border-radius:9px;">

**Fan-Out** occurs when a single node connects to **multiple downstream nodes**.

Instead of creating a chain, one source node connects to multiple target nodes.

LangGraph detects this and executes those target nodes **concurrently (simultaneously)**.

</div>

**Example:**

```text
             START
               |
       +-------+-------+
       |       |       |
       v       v       v
    Article Thumbnail Tweet
      Node     Node    Node
```

### B. Fan-In (The Merger)

<div style="background:#F3E8FF; padding:15px; border-left:6px solid #9333EA; border-radius:9px;">

**Fan-In** occurs when multiple nodes connect to a single downstream node.

It brings together the results of parallel tasks so that a subsequent node can process them.

</div>

**Example:**

```text
    Article     Thumbnail     Tweet
      Node         Node        Node
        |            |           |
        +------------+-----------+
                     |
                     v
                Final Node
```

## Quick Revision

| Concept | Meaning |
|---|---|
| **Nodes** | Individual Python functions performing tasks. |
| **Edges** | Connections that define workflow execution. |
| **Fan-Out** | One node → Multiple nodes. |
| **Fan-In** | Multiple nodes → One node. |

> **Remember:** Fan-Out = Split tasks. Fan-In = Bring results together.
