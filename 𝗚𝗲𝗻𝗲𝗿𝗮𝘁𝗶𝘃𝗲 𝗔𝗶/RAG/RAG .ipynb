<h2 style="color:#B7D94B; border-bottom:2px solid #B7D94B; padding-bottom:8px;">
What is RAG?
</h2>

**RAG stands for <span style="color:#7C3AED;">Retrieval-Augmented Generation</span>.**

RAG is a technique that allows an **LLM to answer questions using relevant information retrieved from external sources**, such as PDFs, documents, databases, or knowledge bases.

### How RAG Works

<div style="background-color:#F3F4F6; padding:14px; border-left:4px solid #7C3AED; border-radius:6px; text-align:center;">

**User Question**  
↓  
<span style="color:#7C3AED;"><b>Retrieval</b></span> — Find relevant information  
↓  
<span style="color:#2563EB;"><b>Augmentation</b></span> — Add the retrieved information to the prompt  
↓  
<span style="color:#059669;"><b>Generation</b></span> — LLM generates an answer  
↓  
**Final Answer**

</div>

### Example: Ask Your PDF

Suppose you upload a PDF containing information about Machine Learning.

**User asks:**  
"What is supervised learning according to this PDF?"

**RAG application:**
1. Searches the PDF for relevant information.
2. Retrieves the matching text.
3. Sends the text and question to the LLM.
4. Generates an answer based on the retrieved content.

### Why Do We Need RAG?

- **External knowledge:** Use information beyond the model's training data.
- **Private documents:** Answer questions about your own files.
- **Updated information:** Retrieve newer information without retraining the LLM.
- **Grounded answers:** Help the model answer using relevant source material.

---

<div style="background-color:#F0FDF4; padding:12px; border-left:4px solid #16A34A; border-radius:6px;">

**In short:**

RAG = **Retrieve relevant information + Give it to the LLM + Generate an answer.**

</div>

<h2 style="color:#B7D94B; border-bottom:2px solid #B7D94B; padding-bottom:8px;">
Text Splitter
</h2>

### <span style="color:#7C3AED;">1. Why Do We Need Text Splitting?</span>

Most embedding models and language models have a **context window**, so they cannot process an infinite number of tokens at the same time.

Therefore, we use **Text Splitting** to divide large documents into smaller chunks.

### <span style="color:#2563EB;">2. Benefits of Chunking / Text Splitting</span>

**A. Better Retrieval in RAG**

In Retrieval-Augmented Generation (RAG), we search through vector embeddings.

If a document is too large:
- Embeddings may become less precise.
- Smaller chunks allow the system to retrieve only the most relevant piece of information instead of the whole document.

**B. More Accurate Embeddings**

Embeddings work best when the text represents one clear idea or topic.

If you embed very large text:
- Multiple topics may mix together.
- Semantic meaning may become blurred.

**Chunking ensures each embedding represents a focused concept.**

**C. Faster Processing**

Working with smaller chunks:
- Speeds up embedding creation.
- Speeds up similarity search in vector databases such as **Chroma, Pinecone, and FAISS**.

Large documents can make the pipeline slower and more expensive.

---

### <span style="color:#D97706;">3. Types of Text Splitting</span>

1. **Character-Based Splitting**
2. **Token-Based Splitting**
3. **Semantic / Meaning-Based Splitting**

### <span style="color:#7C3AED;">1. Character-Based Splitting</span>

Character-based splitting is the **simplest method of text splitting**.

It divides a large document into smaller chunks based on the **number of characters**, without understanding words, tokens, or meaning.

### <span style="color:#2563EB;">2. Token-Based Splitting</span>

Token-based splitting divides a document into chunks based on the **number of tokens, not characters**.

Since LLMs process tokens instead of words, this method aligns better with how models actually read text.

### <span style="color:#059669;">3. Semantic / Meaning-Based Splitting</span>

**Semantic / Meaning-Based Splitting** divides a document into chunks based on the **meaning and structure of the text**, rather than characters or tokens.

The goal is to keep **complete ideas together** so that each chunk contains a coherent piece of information.

<h2 style="color:#B7D94B; border-bottom:2px solid #B7D94B; padding-bottom:8px;">
1. What Are Vector Embeddings?
</h2>

**Vector embeddings** are numerical representations of text that capture its semantic meaning.

An embedding model converts text into a list of numbers called a **vector**.

**Example:**

Text: `"Gradient Descent Optimization"`

↓

Vector: `[0.23, -0.81, 0.44, 0.67, ...]`

These numbers represent patterns and relationships in the text. Text with similar meanings often has similar vector representations.

<div style="background-color:#F3E8FF; padding:12px; border-left:4px solid #7C3AED; border-radius:6px;">

**In short:** Text → Embedding Model → Numerical Vector

</div>

<h2 style="color:#B7D94B; border-bottom:2px solid #B7D94B; padding-bottom:8px;">
2. What Is a Vector Store?
</h2>

A **Vector Store** is a system used to store and search vector embeddings.

In a RAG application, it can store:

- **Vector embeddings** — numerical representations of text.
- **Text chunks** — the original text associated with each vector.
- **Metadata** — additional information, such as document names or page numbers.

It allows the application to find text chunks that are semantically similar to a user's question.

### Example: KrishiMitra AI

1. The system splits a PDF into smaller chunks.
2. An embedding model converts each chunk into a vector.
3. The vectors and associated information are stored in a vector store.
4. When a student asks a question, the system searches for relevant chunks.
5. The retrieved information is passed to the LLM to generate an answer.

<div style="background-color:#ECFDF5; padding:12px; border-left:4px solid #059669; border-radius:6px;">

**In short:**

- **Vector Embeddings →** Represent text as numbers.
- **Vector Store →** Stores and searches those vectors to find relevant information.

</div>

### <span style="color:#7C3AED;">Why Does a Vector Store Store Vector Embeddings?</span>

A vector store stores **vector embeddings** so that we can efficiently search for information based on its **semantic meaning**, rather than matching only exact keywords.

### How Does It Work?

1. **Convert text into vectors:** An embedding model converts each text chunk into a numerical vector.
2. **Store the vectors:** The vector store stores these embeddings along with their associated text and metadata.
3. **Convert the user's question:** The question is also converted into a vector using an embedding model.
4. **Find similar vectors:** The vector store searches for vectors that are close to the question's vector.
5. **Retrieve relevant text:** The corresponding text chunks are returned to the application.

### Example — KrishiMitra AI

Suppose a PDF contains this sentence:

> "Plants need nitrogen for healthy growth."

A student asks:

> "Which nutrient helps plants grow?"

Even though the wording is different, the system can retrieve the relevant sentence because their meanings are similar.

### Why Is This Useful?

- **Semantic search:** Finds relevant information even when the wording differs.
- **Fast retrieval:** Indexing techniques such as HNSW and IVF can speed up similarity searches.
- **RAG applications:** Retrieves relevant document chunks for an LLM to use when answering questions.

---

<div style="background-color:#F3E8FF; padding:12px; border-left:4px solid #7C3AED; border-radius:6px;">

**In short:** The embedding represents the text's meaning as numbers, while the vector store stores and indexes those numbers to help retrieve relevant information efficiently.

</div>

### <span style="color:#7C3AED;">Algorithms Used in Vector Stores</span>

Vector stores use **indexing algorithms** to find similar vector embeddings efficiently.

| Algorithm | Full Form / Purpose |
|---|---|
| **HNSW** | Hierarchical Navigable Small World — graph-based approximate nearest-neighbor search. |
| **IVF** | Inverted File Index — partitions vectors into clusters to narrow the search. |
| **PQ** | Product Quantization — compresses vectors to reduce memory usage. |

**Examples of vector search libraries and databases:**
- **FAISS** — supports IVF, PQ, and HNSW.
- **Chroma** — vector database that uses indexing for similarity search.
- **Pinecone** — managed vector database.

<div style="background-color:#F3E8FF; padding:12px; border-left:4px solid #7C3AED; border-radius:6px;">

**In short:** HNSW, IVF, and PQ are indexing/search techniques that help vector stores find relevant embeddings efficiently. The exact algorithm depends on the vector store and its configuration.

</div>

<h2 style="font-family:Georgia,serif; color:#2563EB; border-bottom:3px solid #2563EB; padding-bottom:10px;">
How IVF Works
</h2>

<p style="font-family:Verdana,sans-serif; font-size:15px;">
<strong style="color:#7C3AED;">IVF (Inverted File Index)</strong> is a vector indexing technique that divides a large collection of vectors into clusters, allowing the system to search only selected clusters instead of examining every vector.
</p>

<div style="background-color:#EFF6FF; padding:14px; border-radius:8px; border-left:5px solid #2563EB; font-family:Verdana,sans-serif;">

<strong>Step 1 — Create Clusters</strong>

The algorithm uses a clustering method, commonly <strong>K-means</strong>, to group similar vectors around cluster centers called <em>centroids</em>.

</div>

<div style="background-color:#F5F3FF; padding:14px; border-radius:8px; border-left:5px solid #7C3AED; font-family:Verdana,sans-serif;">

<strong>Step 2 — Assign Vectors</strong>

Each vector is assigned to its nearest centroid. The index records which vectors belong to each cluster.

</div>

<div style="background-color:#ECFDF5; padding:14px; border-radius:8px; border-left:5px solid #059669; font-family:Verdana,sans-serif;">

<strong>Step 3 — Receive a Query</strong>

When a user asks a question, its embedding is created. IVF finds the nearest cluster centers to that query vector.

</div>

<div style="background-color:#FFF7ED; padding:14px; border-radius:8px; border-left:5px solid #EA580C; font-family:Verdana,sans-serif;">

<strong>Step 4 — Search Selected Clusters</strong>

Instead of searching every vector, IVF searches only a selected number of nearby clusters. This setting is often called <code>nprobe</code>.

</div>

<div style="background-color:#FDF2F8; padding:14px; border-radius:8px; border-left:5px solid #DB2777; font-family:Verdana,sans-serif;">

<strong>Step 5 — Return the Nearest Vectors</strong>

The system ranks candidate vectors by similarity or distance and returns the best matches.

</div>

### Example

Imagine a vector store containing **100,000 vectors**, divided into 100 clusters.

- **Without IVF:** A search may compare the query against all 100,000 vectors.
- **With IVF:** If the search examines 5 clusters containing roughly 1,000 vectors each, it checks about 5,000 candidates instead.

<small>These numbers are illustrative; actual cluster sizes and search work vary.</small>

<div style="background-color:#F3F4F6; padding:12px; border-radius:8px; font-family:Georgia,serif;">

**Key idea:** IVF trades some search completeness for speed. If the relevant vector lies in an unsearched cluster, the search may miss it. Increasing `nprobe` generally improves recall but requires more work.

</div>

<h2 style="color:#B7D94B; border-bottom:2px solid #B7D94B; padding-bottom:8px;">
Retrievers
</h2>

A **Retriever** takes a user query and returns the most relevant documents or chunks from a database.

## Types of Retrievers

### <span style="color:#7C3AED;">1. By Data Source</span>

Retrievers can fetch information from specific external sources or databases instead of your own vector store.

They connect to APIs or datasets and return relevant documents for the query.

**Examples of data sources:**
- Wikipedia
- arXiv
- PubMed

### <span style="color:#2563EB;">2. By Retrieval Strategy</span>

Retrievers can also be classified by the strategy they use to find relevant information.

In most RAG (Retrieval-Augmented Generation) systems, three common retrieval strategies are:

1. **Similarity Search (Most Common)**
2. **MMR (Maximal Marginal Relevance)**
3. **MultiQuery Retriever**

<mark>1. **Similarity Search (Most Common)**


### <span style="color:#7C3AED;">1. Similarity Search</span>

The system compares the **query vector** with document vectors using similarity metrics like:

- **Cosine Similarity** (most common)
- **Dot Product**
- **Euclidean Distance**

The retriever finds the most similar vectors and retrieves the **Top-K documents**.

```python
retriever = vector_store.as_retriever()

<mark>2. **MMR (Maximal Marginal Relevance)**


<h2 style="font-family:Georgia,serif; color:#B7D94B; border-bottom:3px solid #B7D94B; padding-bottom:8px;">
2. MMR — Maximal Marginal Relevance
</h2>

### <span style="color:#7C3AED;">Understanding the Problem</span>

Imagine you are building a **RAG system** for students. Your knowledge base contains many chunks about **Gradient Descent**.

**Example: Database Chunks**

<div style="background-color:#F3F4F6; padding:12px; border-radius:8px; font-family:Verdana,sans-serif;">

**Chunk 1:** Gradient descent is an optimization algorithm used in machine learning.

**Chunk 2:** Gradient descent minimizes the loss function.

**Chunk 3:** Gradient descent is an optimization method that minimizes the loss function.

**Chunk 4:** Neural networks use gradient descent for training.

**Chunk 5:** Support Vector Machines are supervised learning algorithms.

</div>

### <span style="color:#2563EB;">What Happens with Normal Similarity Search?</span>

**User asks:** "What is Gradient Descent?"

Normal similarity search may return **Chunk 1, Chunk 2, and Chunk 3** because they are highly similar to the query.

However, all three chunks convey nearly the same information.

This wastes:

- **Context window**
- **Token usage**
- **Information diversity**

The LLM receives repetitive information instead of a broader range of relevant details.

### <span style="color:#059669;">How MMR Helps</span>

**MMR selects documents that are both relevant to the query and diverse from one another.**

Instead of selecting only the most similar chunks, MMR tries to reduce redundancy while preserving relevance.

<h2 style="font-family:Georgia,serif; color:#B7D94B; border-bottom:3px solid #B7D94B; padding-bottom:8px;">
3. MultiQuery Retriever
</h2>

### <span style="color:#7C3AED;">The Problem</span>

A single query might not capture all the ways a concept can be expressed.

**Example:**

Original query:

`What is gradient descent?`

It could also be asked as:

- Explain gradient descent.
- How does gradient descent work?
- What optimization algorithm is used in neural networks?

Different wording → Different embeddings → Different search results.

This is where the **MultiQuery Retriever** helps.

### <span style="color:#2563EB;">How Does It Work?</span>

Instead of using one query, it generates multiple variations of the query using an LLM.

**Pipeline:**

<div style="background-color:#F3F4F6; padding:14px; border-left:5px solid #2563EB; border-radius:8px; font-family:Verdana,sans-serif;">

**User Query**

↓

**LLM generates multiple queries**

↓

**Retriever searches for each query**

↓

**Combine all retrieved documents**

</div>

### <span style="color:#059669;">Example</span>

Suppose your database contains:

- **Chunk 1:** Gradient descent is an optimization algorithm used to minimize loss functions.
- **Chunk 2:** Neural networks are trained using optimization algorithms like gradient descent.
- **Chunk 3:** Weight updates in machine learning models are done using gradient descent.
- **Chunk 4:** Support Vector Machines are supervised learning algorithms.

For the query *"What is gradient descent?"*, the MultiQuery Retriever generates alternative queries to help find additional relevant chunks.

The retrieved documents are then combined, typically removing duplicates.

### <span style="color:#D97706;">Key Benefit</span>

**Improved retrieval coverage:** It can find relevant information that a single query might miss because of different wording.
