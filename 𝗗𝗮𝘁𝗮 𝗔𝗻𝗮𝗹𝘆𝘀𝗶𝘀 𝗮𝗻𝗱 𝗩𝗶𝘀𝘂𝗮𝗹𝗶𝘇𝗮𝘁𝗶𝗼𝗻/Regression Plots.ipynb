```python
#Main Goal : To predict some values 

# It have x and y axis 

# Y Axis : Prediction
# X Axis : Data Source of the prediction 
```

```python

tips = sns.load_dataset('tips')

tips
```

```python
sns.lmplot(x = 'total_bill' , y = 'tip' , data = tips)

#linear regression 
```

```python
sns.lmplot(x = 'total_bill' , y = 'tip' ,hue = 'sex', data = tips)
```

```python
#Styling 

sns.lmplot(
    x='total_bill',
    y='tip',
    data=tips,
    hue='sex',
    palette='coolwarm',
    markers=['o', 'v'],
    scatter_kws={'s':50}
)
```
