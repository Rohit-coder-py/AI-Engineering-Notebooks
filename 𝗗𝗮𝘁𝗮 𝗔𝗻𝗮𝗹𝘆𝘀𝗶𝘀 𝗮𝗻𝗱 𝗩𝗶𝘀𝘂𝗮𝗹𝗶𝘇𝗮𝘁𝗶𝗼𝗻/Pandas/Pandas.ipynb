<b> Interpolation in pandas 

# Interpolation in Pandas

Interpolation means filling missing values (`NaN`) by guessing values from nearby data.

Instead of putting random values like `0` or `"Unknown"`, pandas tries to estimate the missing value intelligently.

Example:

100 → NaN → 300

After interpolation:

100 → 200 → 300

Mostly used in:
- Time-Series Data
- Weather Data
- Sales Data
- Sensor Data

Pandas Function:

```python
df.interpolate()

`````python
data = {
    "Year": [2021,2022,2023,2034],
    "Resturant": ["China Bowl","Subway","Dominoz","Burger Singh"],
    "Sales": [500000,21000,None,34233],
    "Overall": ["Good","Good","Good","Bad"],
    "City": ["banglore","Mumbai","nanital","uttrakhand"]
} 

df = pd.DataFrame(data)

# df["Sales"]=df["Sales"].interpolate(method = "linear",axis=0)
df.interpolate(method="linear",axis=0,inplace=True)

print(df)
`````


# Common Interpolation Methods 📈

| Method              | Meaning                        |
| ------------------- | ------------------------------ |
| `"linear"`          | straight-line estimation       |
| `"polynomial"`      | curved mathematical estimation |
| `"time"`            | based on timestamps/time gaps  |
| `"nearest"`         | use nearest value              |
| `"spline"`          | smooth curve interpolation     |
| `"pad"` / `"ffill"` | forward fill                   |
| `"bfill"`           | backward fill                  |

---

# 1. Linear (Most Common) ✅

```python id="m1v8qp"
df.interpolate(method="linear")
```

Simple straight-line estimation.

Example:

```text id="w3u6ko"
100 → NaN → 300
```

becomes:

```text id="r9p4zc"
100 → 200 → 300
```

---

# 2. Polynomial 📈

Used when data follows curves instead of straight lines.

```python id="d5n1xt"
df.interpolate(method="polynomial", order=2)
```

More advanced statistical estimation.

Mostly used in:

* scientific data
* graphs
* signals

---

# 3. Time ⏰

Used for time-series datasets.

```python id="h2z9qk"
df.interpolate(method="time")
```

Works using timestamps/dates.

Very useful for:

* stock prices
* weather data
* sensor logs

---

# 4. Forward Fill (`ffill`) 🔥

```python id="n8y0tr"
df.fillna(method="ffill")
```

Copies previous value downward.

Example:

```text id="5x2cvu"
100
NaN
300
```

becomes:

```text id="4v7pda"
100
100
300
```

---

# 5. Backward Fill (`bfill`) 🚀

Uses next value upward.

```text id="0j4lmn"
100
NaN
300
```

becomes:

```text id="c8k1fx"
100
300
300
```

---

# Important Reality 🧠

In real industry:

* `linear`
* `ffill`
* `bfill`

are most commonly used.

Advanced methods:

* polynomial
* spline

are more specialized.

# </mark>Concatenation in Pandas<mark>

Concatenation means combining DataFrames together.

---

# Types of Concatenation

## 1. Vertical Concatenation (Row-wise)

Adds rows one below another.

```python
pd.concat([df1, df2], axis=0, ignore_index=True)
````

### Notes:

* `[df1, df2]` → DataFrames to combine
* `axis=0` → combine vertically (default)
* `ignore_index=True` → resets row indexes

---

## 2. Horizontal Concatenation (Column-wise)

Adds columns side-by-side.

```python
pd.concat([df1, df2], axis=1)
```

### Notes:

* `axis=1` → combine horizontally

---

# Easy Memory Trick 🧠

```text
axis = 0 → vertical ↓
axis = 1 → horizontal →
```

`````python
#Combining vertically 


df_Region1 = pd.DataFrame({
    'CustomerID': [1,2],
    'Name': ['Gopal', 'Raju']})

df_Region2= pd.DataFrame({
    'CustomerID': [3,4],
    'Name':['Shyam', 'Baburao']})


print(df_Region1)


print(df_Region2)

# '''axis=0  ↓
# Rows go downward


# axis=1  →
# Columns go sideways'''
print("\n \n \n")

combined = pd.concat([df_Region1,df_Region2],axis=0,ignore_index=True)

print(combined)

print("\n \n \n")
combined2 = pd.concat([df_Region1,df_Region2],axis=1,ignore_index=True)

print(combined2)
`````
