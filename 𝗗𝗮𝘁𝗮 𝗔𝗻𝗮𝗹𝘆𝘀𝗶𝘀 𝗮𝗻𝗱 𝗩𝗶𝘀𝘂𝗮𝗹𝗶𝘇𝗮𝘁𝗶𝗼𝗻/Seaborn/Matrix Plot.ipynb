<h1 align="center"> Heat Map</h1>

<div style="background-color:#ffffff;padding:15px;border-radius:10px;border-left:6px solid #FF5722;margin-bottom:20px;">

<p>
A <b>Heat Map</b> is a graphical representation of data where values are represented using different colors.
</p>

<p>
It helps to quickly identify patterns, correlations, and variations in the dataset.
</p>

<p>
Darker or brighter colors usually represent higher values, while lighter colors represent lower values.
</p>

</div>

<h1 align="center"> Cluster Map</h1>

<div style="background-color:#ffffff;padding:15px;border-radius:10px;border-left:6px solid #3F51B5;margin-bottom:20px;">

<p>
A <b>Cluster Map</b> is similar to a heat map, but it also groups similar rows and columns together using clustering techniques.
</p>

<p>
It helps in identifying similar patterns and relationships within the dataset.
</p>

</div>

<h1 align="center"> Pivot Table Heat Map</h1>

<div style="background-color:#ffffff;padding:15px;border-radius:10px;border-left:6px solid #009688;">

<p>
A <b>Pivot Table Heat Map</b> is created by applying a heat map to a pivot table.
</p>

<p>
It is commonly used to visualize summarized data and compare values across different categories.
</p>

</div>

</div>

<b><span style="background-color:green; color:white;">Pivot Table</span></b>

```python
import pandas as pd

data = {
    "Brand": ["Samsung", "Samsung", "Apple", "Apple", "Vivo"],
    "Year": [2024, 2025, 2024, 2025, 2025],
    "Sales": [100, 150, 200, 250, 180]
}

df = pd.DataFrame(data)
df.index= range(1,len(df)+1)

df
```

```python
pivot = df.pivot_table(
    values='Sales',
    index='Brand',
    columns='Year',
    aggfunc='sum'
)

pivot
```

```python
data = sns.load_dataset('flights')

pvt = data.pivot_table(
    values='passengers',
    index='month',
    columns='year'
)

sns.heatmap(pvt)
```

```python
pvt = data.pivot_table(values = "passengers", index = 'month', columns = 'year')

pvt
```

```python
sns.heatmap(pvt)
```
