<h1 align="center">📊 Distribution Plots in Data Visualization</h1>

<p align="center">
Distribution plots help us understand how numerical data is spread across a dataset.
</p>

<hr>

<h2>📌 What are Distribution Plots?</h2>

<p>
We use <b>distribution plots</b> to visualize the distribution of quantitative data.
These plots help us analyze:
</p>

<ul>
  <li>Data spread</li>
  <li>Frequency of values</li>
  <li>Patterns in data</li>
  <li>Skewness</li>
  <li>Outliers</li>
</ul>

<hr>

<h2>📚 Common Distribution Plots</h2>

<h3>1️⃣ Histplot / Distplot</h3>

<p>
A histogram is used to visualize the frequency distribution of numerical data.
It groups data into ranges called <b>bins</b>.
</p>

<h4>✅ Uses:</h4>

<ul>
  <li>Understanding data spread</li>
  <li>Finding concentration of values</li>
  <li>Detecting skewness</li>
</ul>

<hr>

<h3>2️⃣ Jointplot</h3>

<p>
A jointplot is used to visualize the relationship between two numerical variables.
It also displays their individual distributions.
</p>

<h4>✅ Uses:</h4>

<ul>
  <li>Comparing two variables</li>
  <li>Finding correlations</li>
  <li>Analyzing trends</li>
</ul>

<hr>

<h3>3️⃣ Pairplot</h3>

<p>
A pairplot creates multiple plots to visualize relationships between all numerical columns in a dataset.
</p>

<h4>✅ Uses:</h4>

<ul>
  <li>Exploratory Data Analysis (EDA)</li>
  <li>Feature comparison</li>
  <li>Pattern recognition</li>
</ul>

<hr>

<h3>4️⃣ Rugplot</h3>

<p>
A rugplot displays small marks along an axis to represent the distribution of individual data points.
</p>

<h4>✅ Uses:</h4>

<ul>
  <li>Visualizing density of points</li>
  <li>Understanding value concentration</li>
  <li>Supporting histograms and KDE plots</li>
</ul>

<hr>

<h2>🎯 Conclusion</h2>

<p>
Distribution plots are essential in data analysis because they help us understand the structure and behavior of data before applying machine learning or statistical techniques.
</p>

<b><span style="background-color:green; color:white;">  Distplot or Histplot  </span></b>

```python
df = sns.load_dataset("tips")

x = df["total_bill"]
sns.histplot(x,bins=10,kde = True)

#evry bar = bins

#kde = Kernal Density Estimation
```

<b><span style="background-color:green; color:white;"> Jointplot  </span></b>

```python
# Used to compare 2 values 
x1 = df["total_bill"]
x2 = df["tip"]
sns.jointplot(x = x1 , y = x2 , data = df , kind = 'hex',height = 6)
```

<b><span style="background-color:green; color:white;"> PairPlot  </span></b>

```python
# Most Liked for Data Analysation and Visualization

#Compare relationships between ALL numerical columns in a dataset

sns.pairplot(df,hue='sex',palette= 'rainbow')
```

<b><span style="background-color:green; color:white;"> RugPlot  </span></b>

```python
sns.rugplot(df)
```
