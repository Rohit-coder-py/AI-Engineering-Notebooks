<b><span style="background-color:green; color:white;">  Importing  </span></b>

```python
df = sns.load_dataset('tips')

pd.set_option('display.max_rows',None)

df
```

<b><span style="background-color:green; color:white;">  Countplot  </span></b>

```python
sns.countplot(x = df['sex'],hue=df['smoker'])
plt.title("Graph of smokers")
#to check the no. of counts
```

<b><span style="background-color:green; color:white;">  BarPlot  </span></b>

```python
sns.barplot(x = df["sex"],y = df["tip"])

plt.show()
```

```python
sns.barplot(x = df["sex"],y = df["tip"],    estimator = np.sum)

plt.show()


# x axis = quantitative  (numbers)
#y axis = qualitative
```

<b><span style="background-color:green; color:white;">  Boxplot  </span></b>

```python
sns.boxplot(x = df['tip'] , y = df['day'],data = df   ,palette = 'rainbow' )


# x axis = quantitative  (numbers)
#y axis = qualitative
```

<div style="border:2px solid #444;padding:20px;border-radius:12px;background-color:#f8f9fa;line-height:1.8;">

<h1 align="center"> Boxplot and Violinplot</h1>

<p style="font-size:16px;">
<b>Boxplots</b> and <b>violinplots</b> are used to shown the distribution of categorical data.
</p>

<p style="font-size:16px;">
A <b>box plot</b> (or <b>box-and-whisker plot</b>) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable.
</p>

<p style="font-size:16px;">
The <b>box</b> shows the quartiles of the dataset while the <b>whiskers</b> extend to show the rest of the distribution, except for points that are determined to be <b>"outliers"</b> using a method that is a function of the <b>inter-quartile range</b>.
</p>

</div>

<b><span style="background-color:green; color:white;">  Violin Plots  </span></b>

```python
sns.violinplot(x = 'day', y = "total_bill",data = df)
```

<div style="border:2px solid #444;padding:20px;border-radius:12px;background-color:#f8f9fa;line-height:1.8;">

<h1 align="center">Stripplot and Swarmplot</h1>

<div style="background-color:#ffffff;padding:15px;border-radius:10px;border-left:6px solid #4CAF50;margin-bottom:20px;">

<h2>🔹 Stripplot</h2>

<p>
The <b>stripplot</b> will draw a scatterplot where one variable is categorical.
</p>

<p>
A <b>strip plot</b> can be drawn on its own, but it is also a good complement to a <b>box plot</b> or <b>violin plot</b> in cases where you want to show all observations along with some representation of the underlying distribution.
</p>

</div>

<div style="background-color:#ffffff;padding:15px;border-radius:10px;border-left:6px solid #2196F3;">

<h2>🔹 Swarmplot</h2>

<p>
The <b>swarmplot</b> is similar to <b>stripplot()</b>, but the points are adjusted <b>(only along the categorical axis)</b> so that they don't overlap.
</p>

<p>
This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations.
</p>

<p>
It helps both in:
<br>✔️ Showing all the points clearly
<br>✔️ Better understanding the data distribution
</p>

<p>
However, it requires more computation to arrange the points properly.
</p>

</div>

</div>

```python
sns.stripplot(x = df['day'], y = df['tip'])
```

```python
sns.swarmplot(x = df['day'], y = df['tip'],data = df )

sns.violinplot(x = 'day', y = "tip",data = df)

#to show variation in stripplot
```
