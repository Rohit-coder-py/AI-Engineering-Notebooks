```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

<b><span style="background-color:green; color:white;">Matplotlib Basics</span></b>

```python
arr1 = np.arange(1,5)

arr2 = arr1**2

print(arr1)
print(arr2)
```

```python
plt.plot(arr1,arr1)
plt.title("Hello")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

<b><span style="background-color:green; color:white;">Subplots in Matplotlib</span></b>

Multiple graphs inside ONE figure/window.

```python
plt.subplot(2,2,1)
plt.plot(arr1,arr2)

plt.subplot(2,2,2)
plt.plot(arr1,arr1)

plt.subplot(2,2,3)
plt.plot(arr2,arr1)

plt.subplot(2,2,4)
plt.plot(arr2+1,arr1+1)

#rows,columns,position 
```

<b><span style="background-color:green; color:white;">Object Oriented Way</span></b>

<b>Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi
keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).</b>

```python
# fig = plt.figure()
# fig.add_axes([0.1,0.1 ,0.8  ,1])

# [left, bottom, width, height]


fig = plt.figure()                               #add a empty canvas

axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(arr1,arr2)

axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(arr1,arr2)
```

```python
#We can also adjust the size of canvas by : figsize 

# 6,8 = 600 * 800 pixels

fig = plt.figure( figsize = (6,6),dpi = 100)                  #( figsize(width,length))

axes = fig.add_axes( [ 0.1,0.1 , 0.8, 0.9] )

plot = axes.plot(arr1,arr2)



axes2 = fig.add_axes([0.2,0.2,0.3,0.9])

plot = axes2.plot(arr1,arr2)
```

<b><span style="background-color:green; color:white;">Types of charts and their usage</span></b>

# 🧠 <b>Simple Chart Selection Guide</b>

| <b>Question</b> | <b>Best Chart</b> |
|---|---|
| Compare categories? | 📊 <b>Bar Chart</b> |
| Trend over time? | 📈 <b>Line Chart</b> |
| Distribution of values? | 📉 <b>Histogram</b> |
| Relationship between variables? | ⭐ <b>Scatter Plot</b> |
| Percentage / Share? | 🥧 <b>Pie Chart</b> |

---

# 📌 <b>Quick Examples</b>

| <b>Example</b> | <b>Chart</b> |
|---|---|
| Brand-wise smartphone sales | 📊 <b>Bar Chart</b> |
| Monthly sales growth | 📈 <b>Line Chart</b> |
| Rating distribution | 📉 <b>Histogram</b> |
| Price vs Rating | ⭐ <b>Scatter Plot</b> |
| Smartphone market share | 🥧 <b>Pie Chart</b> |

---

# 🚀 <b>Golden Rule</b>

Before plotting, ask:

> <mark>“What story am I trying to show?”</mark>

<b><span style="background-color:green; color:white;"> Pie Charts</span></b>

```python
labels = ["Apple","Samsung","vivo"]
per = [33,33,33]

plt.pie(per,labels=labels,shadow = True)

plt.show()
```

```python
# Profesnial version

labels = ["Apple", "Samsung", "Xiaomi", "OnePlus"]
sizes = [40, 30, 20, 10]

plt.figure(figsize=(6,6))

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    shadow=True
)

plt.title("Market Share")

plt.show()
```
