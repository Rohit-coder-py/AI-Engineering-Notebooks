# HTTP Methods in FastAPI

| HTTP Method | Full Form / Meaning | Purpose | FastAPI Example |
|---|---|---|---|
| **GET** | Retrieve | Fetch data from the server | `@app.get("/users")` |
| **POST** | Submit / Create | Send data to the server, often to create a resource | `@app.post("/users")` |
| **PUT** | Replace | Replace an entire resource | `@app.put("/users/{id}")` |
| **PATCH** | Partial Update | Update specific fields of a resource | `@app.patch("/users/{id}")` |
| **DELETE** | Delete | Remove a resource | `@app.delete("/users/{id}")` |
| **HEAD** | Headers Only | Retrieve response headers without the response body | `@app.head("/users")` |
| **OPTIONS** | Communication Options | Discover supported methods or handle CORS preflight requests | `@app.options("/users")` |
| **TRACE** | Diagnostic Loopback | Diagnose how a request is received by the server | `@app.trace("/users")` |
| **CONNECT** | Establish Tunnel | Create a tunnel through a proxy, commonly for HTTPS | Usually handled by a proxy |

## Safe and Idempotent Methods

| Concept | Meaning | Methods |
|---|---|---|
| **Safe** | Intended not to modify resource state | GET, HEAD, OPTIONS, TRACE |
| **Idempotent** | Repeating the same request has the same intended effect on server state | GET, HEAD, PUT, DELETE, OPTIONS, TRACE |

## CRUD Operations

| CRUD Operation | HTTP Method | Example Endpoint |
|---|---|---|
| **Create** | POST | `/users` |
| **Read** | GET | `/users` |
| **Update (Full Replacement)** | PUT | `/users/{id}` |
| **Update (Partial)** | PATCH | `/users/{id}` |
| **Delete** | DELETE | `/users/{id}` |

> **Remember:** GET = Read, POST = Create/Submit, PUT = Replace, PATCH = Partial Update, DELETE = Remove.

# What Is an HTTP Status Code?

> **Definition:** An HTTP status code is a three-digit number returned by a server to indicate the result of a client's HTTP request.

## Why Do We Use Status Codes?

Status codes help the client understand whether a request was successful, requires further action, or failed.

## Example

When a client requests data from a FastAPI server:

- `200 OK` → The request was successful.
- `201 Created` → A resource was created successfully.
- `404 Not Found` → The requested resource could not be found.
- `500 Internal Server Error` → An unexpected server-side error occurred.

## Categories of HTTP Status Codes

| Range | Category | Meaning |
|---|---|---|
| `1xx` | Informational | The request is still in progress or requires further protocol action. |
| `2xx` | Success | The request was successfully received and processed. |
| `3xx` | Redirection | Further action is needed, often involving another URL. |
| `4xx` | Client Error | There is a problem with the request. |
| `5xx` | Server Error | The server failed to fulfill a valid request. |


# HTTP Status Codes in FastAPI

## 1. Informational Responses (1xx)

| Status Code | Name | Meaning |
|---|---|---|
| **100** | Continue | The client can continue sending the request. |
| **101** | Switching Protocols | The server is switching protocols. |

## 2. Success Responses (2xx)

| Status Code | Name | Meaning |
|---|---|---|
| **200** | OK | The request was successful. |
| **201** | Created | A new resource was created successfully. |
| **202** | Accepted | The request was accepted for processing, but processing may not be complete. |
| **204** | No Content | The request succeeded, but there is no response body. |

## 3. Redirection Responses (3xx)

| Status Code | Name | Meaning |
|---|---|---|
| **301** | Moved Permanently | The resource has permanently moved to another URL. |
| **302** | Found | The resource is temporarily available at another URL. |
| **304** | Not Modified | The cached resource has not changed. |

## 4. Client Error Responses (4xx)

| Status Code | Name | Meaning |
|---|---|---|
| **400** | Bad Request | The request is invalid or malformed. |
| **401** | Unauthorized | Authentication is required or has failed. |
| **403** | Forbidden | The server refuses to authorize the request. |
| **404** | Not Found | The requested resource does not exist or cannot be found. |
| **405** | Method Not Allowed | The HTTP method is not supported for this endpoint. |
| **408** | Request Timeout | The server timed out waiting for the request. |
| **409** | Conflict | The request conflicts with the current resource state. |
| **413** | Content Too Large | The request body exceeds the allowed size. |
| **415** | Unsupported Media Type | The request uses an unsupported content format. |
| **422** | Unprocessable Content | The request content cannot be processed as submitted. FastAPI commonly uses this for request validation errors. |
| **429** | Too Many Requests | The client has sent too many requests in a given period. |

## 5. Server Error Responses (5xx)

| Status Code | Name | Meaning |
|---|---|---|
| **500** | Internal Server Error | An unexpected error occurred on the server. |
| **501** | Not Implemented | The server does not support the requested functionality. |
| **502** | Bad Gateway | A gateway received an invalid response from an upstream server. |
| **503** | Service Unavailable | The server is temporarily unable to handle the request. |
| **504** | Gateway Timeout | A gateway did not receive a timely response from an upstream server. |

# Difference Between GET and POST Methods

| Feature | GET Method | POST Method |
|---|---|---|
| Purpose | Retrieve data from the server | Submit data to the server |
| Main Use | Fetch or read data | Create or process data |
| Data Location | Usually sent through URL parameters | Usually sent through the request body |
| Request Body | Generally not used | Commonly used |
| Data Visibility | URL parameters may be visible in the browser | Request body is not shown in the URL |
| Data Size | URL length may be limited | Supports larger request bodies, subject to server limits |
| Security | HTTPS is needed for secure transmission | HTTPS is needed for secure transmission |
| Data Modification | Should not modify server data | May modify server data |
| Idempotent | Yes, by HTTP definition | Not necessarily |
| Browser Usage | Can be opened directly in the browser | Usually requires a form, API client, or JavaScript |
| FastAPI Decorator | `@app.get()` | `@app.post()` |
| Example Endpoint | `/products` | `/userprofile` |
| Example Use Case | View available products | Submit customer details |
| Typical Response | Requested data | Confirmation or processed result |

# Path Parameters vs Query Parameters in FastAPI

In **FastAPI**, we use Path Parameters and Query Parameters to pass data through URLs.

The main difference is where the data appears and what it represents.

---

## 1. Path Parameters

Use **Path** when a value is required to identify a specific resource or item.

**Example URL:**

```text
/products/101
````

Here, `101` is the product ID.

### Syntax

```python
from fastapi import FastAPI, Pathapp = FastAPI()
@app.get("/products/{product_id}")
    def get_product(product_id: int = Path(...)): 
        return {"product_id": product_id}
```

### When to use Path Parameters?

- Fetch a specific user by ID.
- Get details of a particular product.
- Access a specific blog post.
- Delete a particular record.

## 2. Query Parameters

Use Query when you want to filter, sort, search, or control the results returned by an endpoint.

Example URL:

```
/products?limit=10&category=electronics
```

Here:

- `limit=10` → Show only 10 products.
- `category=electronics` → Filter products by category.

### Syntax

```
from fastapi import FastAPI, Queryapp = FastAPI()@app.get("/products")def get_products(    limit: int = Query(10),    category: str = Query(None)):    return {        "limit": limit,        "category": category    }
```

### When to use Query Parameters?

- Limit the number of results.
- Search products by name.
- Filter products by category.
- Sort results by price.
- Implement pagination.

## 3. Difference Between Path and Query

| Feature           | Path Parameters     | Query Parameters         |
| ----------------- | ------------------- | ------------------------ |
| Purpose           | Identify a resource | Filter or modify results |
| URL symbol        | `/`                 | `?`                      |
| Multiple values   | Possible            | Possible using `&`       |
| Usually required? | Yes                 | Usually optional         |
| Example           | `/users/101`        | `/users?age=18`          |
| FastAPI           | `Path()`            | `Query()`                |

## 4. Using Both Together

You can use Path and Query Parameters in the same endpoint.

### Code

```python
from fastapi import FastAPI, Path, Query
app = FastAPI()
@app.get("/users/{user_id}/orders")
    def get_orders(    user_id: int = Path(...),    limit: int = Query(10),    status: str = Query(None)):
            return {        "user_id": user_id,        "limit": limit,        "status": status    }
```

### URL

```
http://127.0.0.1:8000/users/101/orders?limit=5&status=delivered
```

### Output

```
{
    "user_id": 101,
    "limit": 5,
    "status": "delivered"
}
```

## 5. Real-World Example: E-commerce API

| Requirement            | URL                       | Parameter |
| ---------------------- | ------------------------- | --------- |
| Get product ID 101     | `/products/101`           | Path      |
| Show 10 products       | `/products?limit=10`      | Query     |
| Search laptop          | `/products?search=laptop` | Query     |
| Get user 5             | `/users/5`                | Path      |
| Get 5 orders of user 5 | `/users/5/orders?limit=5` | Both      |

## 6. Important Rule to Remember

> Path = Which resource do you want?
>
> Query = How do you want to retrieve that resource?

Example:

```
/products/101?currency=INR
```

- `101` → Identifies the product (Path).
- `currency=INR` → Controls how its price is displayed (Query).

Conclusion:

- Use `Path()` to identify a particular resource.
- Use `Query()` to filter, search, sort, or limit the results.

```
```

<div style="background:linear-gradient(135deg,#EDE9FE,#DBEAFE,#E0F2FE); padding:22px; border:2px solid #7C3AED; border-left:8px solid #7C3AED; border-radius:14px; box-shadow:0 3px 10px rgba(124,58,237,0.12);">

<h1 style="color:#6D28D9; font-size:30px; font-weight:900; margin:0; letter-spacing:1px;">
PHASE 4
</h1>

<div style="height:3px; background:linear-gradient(90deg,#7C3AED,#2563EB,#06B6D4); border-radius:5px; margin:12px 0;"></div>

<h2 style="color:#1E3A8A; font-size:24px; font-weight:800; margin:0;">
Request Body
</h2>

<p style="color:#475569; font-size:14px; font-weight:500; margin:10px 0 0 0;">
Learn how to send structured data from a client to a FastAPI application.
</p>

<div style="display:flex; gap:7px; margin-top:16px;">
<span style="background:#7C3AED; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">FASTAPI</span>
<span style="background:#2563EB; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">REQUEST DATA</span>
<span style="background:#0891B2; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">PYDANTIC</span>
</div>

</div>

# Request Body in FastAPI

## 1. What is a Request Body?

A **Request Body** is the data that a client sends to a server inside an HTTP request.

In FastAPI, we generally use a Request Body when we want to **send data to the server**, usually through the `POST`, `PUT`, or `PATCH` HTTP methods.

**Examples:**
- Registering a new user.
- Creating a product.
- Submitting login credentials.
- Updating customer information.

---

## 2. How Does It Work?

Suppose a user wants to register on an application.

The client sends the following JSON data to the server:

```json
{
    "user_id": 101,
    "user_name": "Rohit",
    "user_age": 17,
    "user_city": "Lucknow"
}
```
The json data is called **Request Body**

<div style="background:linear-gradient(135deg,#FCE7F3,#FDF2F8,#EDE9FE); padding:22px; border:2px solid #DB2777; border-left:8px solid #DB2777; border-radius:14px; box-shadow:0 3px 10px rgba(219,39,119,0.12);">

<h1 style="color:#BE185D; font-size:30px; font-weight:900; margin:0; letter-spacing:1px;">
PHASE 5
</h1>

<div style="height:3px; background:linear-gradient(90deg,#DB2777,#9333EA,#6366F1); border-radius:5px; margin:12px 0;"></div>

<h2 style="color:#831843; font-size:24px; font-weight:800; margin:0;">
Exception Handling in FastAPI
</h2>

<p style="color:#6B4B5B; font-size:14px; font-weight:500; margin:10px 0 0 0;">
Learn how to handle errors, manage exceptions, and return meaningful error responses in FastAPI.
</p>

<div style="display:flex; flex-wrap:wrap; gap:7px; margin-top:16px;">
<span style="background:#BE185D; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">FASTAPI</span>
<span style="background:#9333EA; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">ERROR HANDLING</span>
<span style="background:#4F46E5; color:white; padding:5px 11px; border-radius:20px; font-size:12px; font-weight:bold;">EXCEPTIONS</span>
</div>

</div>

we use http exeption for this and it returen a message + status code
