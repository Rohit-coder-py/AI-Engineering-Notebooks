<h2 style="color:#7C3AED;">1. What Are HTTP Methods?</h2>

HTTP (Hypertext Transfer Protocol) methods specify what a client wants to do with a resource on a server.

For example, in a student management API, a client might want to:

- Retrieve student information.
- Create a new student.
- Update an existing student's information.
- Delete a student record.

Different HTTP methods are used for these different operations.

<h2 style="color:#7C3AED;">2. Types of HTTP Methods</h2>

| HTTP Method | Purpose | Example |
|:--|:--|:--|
| GET | Retrieve data | Get all students |
| POST | Create a resource or submit data for processing | Add a student |
| PUT | Replace a resource | Replace a student's details |
| PATCH | Partially update a resource | Update a student's email |
| DELETE | Delete a resource | Remove a student |
| HEAD | Retrieve response headers without the response body | Check resource headers |
| OPTIONS | Discover communication options supported by a resource | Check supported methods |
| TRACE | Diagnostic loopback of a request through the server | Inspect request handling |
| CONNECT | Establish a tunnel to a server, commonly through a proxy | Create an HTTPS tunnel |

<strong>Important:</strong> GET, POST, PUT, PATCH and DELETE are the methods you will use most frequently when building everyday FastAPI applications.

<h2 style="color:#7C3AED;">3. GET Method</h2>

<strong>Purpose:</strong> Retrieve information from the server.

GET should not be used to modify the resource being requested.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
GET /students
</pre>

<strong>FastAPI Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
from fastapi import FastAPI

app = FastAPI()

students = [
    {"id": 1, "name": "Aman"},
    {"id": 2, "name": "Rahul"}
]

@app.get("/students")
def get_students():
    return students
</pre>

<strong>Response:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
[
    {"id": 1, "name": "Aman"},
    {"id": 2, "name": "Rahul"}
]
</pre>

<strong>Common use cases:</strong>

- Fetching product details.
- Retrieving database records.
- Getting predictions from an existing model.
- Reading information from an API.

<h2 style="color:#7C3AED;">4. POST Method</h2>

<strong>Purpose:</strong> Submit data to the server, commonly to create a resource or initiate processing.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
POST /students
</pre>

<strong>FastAPI Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Student(BaseModel):
    name: str
    age: int

@app.post("/students")
def create_student(student: Student):
    return {
        "message": "Student received successfully",
        "student": student
    }
</pre>

<strong>Example request body:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{
    "name": "Aman",
    "age": 20
}
</pre>

<strong>Common use cases:</strong>

- Registering a new user.
- Adding a product.
- Uploading information for processing.
- Sending a prompt to an AI service.

<strong>Note:</strong> POST does not automatically save data to a database. You must implement the storage or processing logic.

<h2 style="color:#7C3AED;">5. PUT Method</h2>

<strong>Purpose:</strong> Replace the current representation of a resource with the supplied representation.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
PUT /students/1
</pre>

<strong>FastAPI Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI()

students = {
    1: {"name": "Aman", "age": 20}
}

class Student(BaseModel):
    name: str
    age: int

@app.put("/students/{student_id}")
def replace_student(student_id: int, student: Student):
    if student_id not in students:
        raise HTTPException(status_code=404, detail="Student not found")

    students[student_id] = student.model_dump()
    return students[student_id]
</pre>

<strong>Example request body:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{
    "name": "Aman Kumar",
    "age": 21
}
</pre>

<strong>Before:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{"name": "Aman", "age": 20}
</pre>

<strong>After:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{"name": "Aman Kumar", "age": 21}
</pre>

<strong>Important:</strong> PUT generally represents full replacement. In this example, both fields are required. The exact behavior when a resource does not exist depends on the API's implementation.

<h2 style="color:#7C3AED;">6. PATCH Method</h2>

<strong>Purpose:</strong> Apply a partial modification to a resource.

Unlike PUT, PATCH can update only the fields you supply.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
PATCH /students/1
</pre>

<strong>FastAPI Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI()

students = {
    1: {"name": "Aman", "age": 20}
}

class StudentUpdate(BaseModel):
    name: str | None = None
    age: int | None = None

@app.patch("/students/{student_id}")
def update_student(student_id: int, student: StudentUpdate):
    if student_id not in students:
        raise HTTPException(status_code=404, detail="Student not found")

    updates = student.model_dump(exclude_unset=True)
    students[student_id].update(updates)

    return students[student_id]
</pre>

<strong>Example request body:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{
    "age": 21
}
</pre>

<strong>Result:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{
    "name": "Aman",
    "age": 21
}
</pre>

The student's name remains unchanged because only the age was provided.

<strong>Note:</strong> This example allows explicit null values because the fields are nullable. In a production API, you may want stricter validation.

<h2 style="color:#7C3AED;">7. DELETE Method</h2>

<strong>Purpose:</strong> Delete a resource.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
DELETE /students/1
</pre>

<strong>FastAPI Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
from fastapi import FastAPI, HTTPException

app = FastAPI()

students = {
    1: {"name": "Aman", "age": 20},
    2: {"name": "Rahul", "age": 22}
}

@app.delete("/students/{student_id}")
def delete_student(student_id: int):
    if student_id not in students:
        raise HTTPException(status_code=404, detail="Student not found")

    deleted_student = students.pop(student_id)

    return {
        "message": "Student deleted successfully",
        "student": deleted_student
    }
</pre>

<strong>Response:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
{
    "message": "Student deleted successfully",
    "student": {
        "name": "Aman",
        "age": 20
    }
}
</pre>

<strong>Common use cases:</strong>

- Removing a user.
- Deleting a product.
- Removing a database record.

<h2 style="color:#7C3AED;">8. HEAD Method</h2>

<strong>Purpose:</strong> Retrieve the headers that a GET request would return, without returning the response body.

It is useful when you want to inspect response metadata without downloading the full content.

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
HEAD /students
</pre>

<strong>Example:</strong>

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
HTTP/1.1 200 OK
Content-Type: application/json
</pre>

The actual headers depend on the server and response.

<h2 style="color:#7C3AED;">9. OPTIONS Method</h2>

<strong>Purpose:</strong> Discover communication options available for a resource or server.

For example, a server might respond with an Allow header:

<pre style="background-color:#F8FAFC; padding:12px; border-radius:8px; color:#334155;">
OPTIONS /students

Allow: GET, POST, PUT, PATCH, DELETE
</pre>

OPTIONS is also important for browser-based cross-origin requests, including CORS preflight requests.

<strong>Important:</strong> FastAPI can handle CORS preflight requests when configured with CORSMiddleware. You do not usually need to write a separate OPTIONS endpoint for every route.

<h2 style="color:#7C3AED;">10. TRACE Method</h2>

<strong>Purpose:</strong> Perform a diagnostic loopback of a request to inspect how it is received by the server.

TRACE is rarely used in ordinary application development and is commonly disabled for security reasons.

<h2 style="color:#7C3AED;">11. CONNECT Method</h2>

<strong>Purpose:</strong> Establish a tunnel to a destination server through an intermediary proxy.

It is commonly used to create a tunnel for HTTPS traffic through an HTTP proxy.

CONNECT is generally handled by proxy infrastructure rather than ordinary FastAPI endpoints.

<h2 style="color:#7C3AED;">12. Safe and Idempotent Methods</h2>

These are two important HTTP concepts.

<h3 style="color:#D97706;">Safe Methods</h3>

A safe method is intended to retrieve information without requesting a change to the resource's state.

- GET
- HEAD
- OPTIONS
- TRACE

<h3 style="color:#D97706;">Idempotent Methods</h3>

A method is idempotent when repeating the same request has the same intended effect on server state as making it once.

| Method | Safe? | Idempotent? |
|:--|:--:|:--:|
| GET | Yes | Yes |
| HEAD | Yes | Yes |
| POST | No | Not guaranteed |
| PUT | No | Yes |
| PATCH | No | Not guaranteed |
| DELETE | No | Yes |
| OPTIONS | Yes | Yes |
| TRACE | Yes | Yes |
| CONNECT | No | Not generally |

<strong>Example:</strong>

If you send the same PUT request five times, the intended final representation should remain the same as after the first request.

If you send the same POST request five times, the server might create five records unless the API implements duplicate protection or idempotency handling.

For DELETE, repeating the request should leave the resource deleted, even if later requests return a 404 response.

The decorators above are available on FastAPI applications. You can also use the corresponding methods with an `APIRouter`.

# What Is an HTTP Status Code?

> **Definition:** An HTTP status code is a three-digit number returned by a server to indicate the result of a client's HTTP request.

## Why Do We Use Status Codes?

Status codes help the client understand whether a request was successful, requires further action, or failed.

## Example

When a client requests data from a FastAPI server:

- `200 OK` → The request was successful.
- `201 Created` → A resource was created successfully.
- `404 Not Found` → The requested resource could not be found.
- `500 Internal Server Error` → An unexpected server-side error occurred.

## Categories of HTTP Status Codes

| Range | Category | Meaning |
|---|---|---|
| `1xx` | Informational | The request is still in progress or requires further protocol action. |
| `2xx` | Success | The request was successfully received and processed. |
| `3xx` | Redirection | Further action is needed, often involving another URL. |
| `4xx` | Client Error | There is a problem with the request. |
| `5xx` | Server Error | The server failed to fulfill a valid request. |

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello, World!"}

# HTTP Status Codes in FastAPI

## 1. Informational Responses (1xx)

| Status Code | Name | Meaning |
|---|---|---|
| **100** | Continue | The client can continue sending the request. |
| **101** | Switching Protocols | The server is switching protocols. |

## 2. Success Responses (2xx)

| Status Code | Name | Meaning |
|---|---|---|
| **200** | OK | The request was successful. |
| **201** | Created | A new resource was created successfully. |
| **202** | Accepted | The request was accepted for processing, but processing may not be complete. |
| **204** | No Content | The request succeeded, but there is no response body. |

## 3. Redirection Responses (3xx)

| Status Code | Name | Meaning |
|---|---|---|
| **301** | Moved Permanently | The resource has permanently moved to another URL. |
| **302** | Found | The resource is temporarily available at another URL. |
| **304** | Not Modified | The cached resource has not changed. |

## 4. Client Error Responses (4xx)

| Status Code | Name | Meaning |
|---|---|---|
| **400** | Bad Request | The request is invalid or malformed. |
| **401** | Unauthorized | Authentication is required or has failed. |
| **403** | Forbidden | The server refuses to authorize the request. |
| **404** | Not Found | The requested resource does not exist or cannot be found. |
| **405** | Method Not Allowed | The HTTP method is not supported for this endpoint. |
| **408** | Request Timeout | The server timed out waiting for the request. |
| **409** | Conflict | The request conflicts with the current resource state. |
| **413** | Content Too Large | The request body exceeds the allowed size. |
| **415** | Unsupported Media Type | The request uses an unsupported content format. |
| **422** | Unprocessable Content | The request content cannot be processed as submitted. FastAPI commonly uses this for request validation errors. |
| **429** | Too Many Requests | The client has sent too many requests in a given period. |

## 5. Server Error Responses (5xx)

| Status Code | Name | Meaning |
|---|---|---|
| **500** | Internal Server Error | An unexpected error occurred on the server. |
| **501** | Not Implemented | The server does not support the requested functionality. |
| **502** | Bad Gateway | A gateway received an invalid response from an upstream server. |
| **503** | Service Unavailable | The server is temporarily unable to handle the request. |
| **504** | Gateway Timeout | A gateway did not receive a timely response from an upstream server. |

---

## Common Status Codes in FastAPI

| Status Code | When to Use |
|---|---|
| **200 OK** | Successfully retrieve or process data. |
| **201 Created** | Successfully create a resource. |
| **204 No Content** | Successfully complete a request without returning a body. |
| **400 Bad Request** | Reject an invalid request. |
| **401 Unauthorized** | Authentication is missing or invalid. |
| **403 Forbidden** | The user is authenticated but lacks permission. |
| **404 Not Found** | The requested resource does not exist. |
| **409 Conflict** | A request conflicts with existing data. |
| **422 Unprocessable Content** | Request validation fails. |
| **500 Internal Server Error** | An unexpected server-side error occurs. |

---

## Example: Using Status Codes in FastAPI

### 1. Returning a Custom Status Code

```python
from fastapi import FastAPI, status

app = FastAPI()

@app.post("/users", status_code=status.HTTP_201_CREATED)
def create_user():
    return {"message": "User created successfully"}
