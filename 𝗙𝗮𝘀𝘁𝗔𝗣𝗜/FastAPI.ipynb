# HTTP Methods in FastAPI

| HTTP Method | Full Form / Meaning | Purpose | FastAPI Example |
|---|---|---|---|
| **GET** | Retrieve | Fetch data from the server | `@app.get("/users")` |
| **POST** | Submit / Create | Send data to the server, often to create a resource | `@app.post("/users")` |
| **PUT** | Replace | Replace an entire resource | `@app.put("/users/{id}")` |
| **PATCH** | Partial Update | Update specific fields of a resource | `@app.patch("/users/{id}")` |
| **DELETE** | Delete | Remove a resource | `@app.delete("/users/{id}")` |
| **HEAD** | Headers Only | Retrieve response headers without the response body | `@app.head("/users")` |
| **OPTIONS** | Communication Options | Discover supported methods or handle CORS preflight requests | `@app.options("/users")` |
| **TRACE** | Diagnostic Loopback | Diagnose how a request is received by the server | `@app.trace("/users")` |
| **CONNECT** | Establish Tunnel | Create a tunnel through a proxy, commonly for HTTPS | Usually handled by a proxy |

## Safe and Idempotent Methods

| Concept | Meaning | Methods |
|---|---|---|
| **Safe** | Intended not to modify resource state | GET, HEAD, OPTIONS, TRACE |
| **Idempotent** | Repeating the same request has the same intended effect on server state | GET, HEAD, PUT, DELETE, OPTIONS, TRACE |

## CRUD Operations

| CRUD Operation | HTTP Method | Example Endpoint |
|---|---|---|
| **Create** | POST | `/users` |
| **Read** | GET | `/users` |
| **Update (Full Replacement)** | PUT | `/users/{id}` |
| **Update (Partial)** | PATCH | `/users/{id}` |
| **Delete** | DELETE | `/users/{id}` |

> **Remember:** GET = Read, POST = Create/Submit, PUT = Replace, PATCH = Partial Update, DELETE = Remove.

# What Is an HTTP Status Code?

> **Definition:** An HTTP status code is a three-digit number returned by a server to indicate the result of a client's HTTP request.

## Why Do We Use Status Codes?

Status codes help the client understand whether a request was successful, requires further action, or failed.

## Example

When a client requests data from a FastAPI server:

- `200 OK` → The request was successful.
- `201 Created` → A resource was created successfully.
- `404 Not Found` → The requested resource could not be found.
- `500 Internal Server Error` → An unexpected server-side error occurred.

## Categories of HTTP Status Codes

| Range | Category | Meaning |
|---|---|---|
| `1xx` | Informational | The request is still in progress or requires further protocol action. |
| `2xx` | Success | The request was successfully received and processed. |
| `3xx` | Redirection | Further action is needed, often involving another URL. |
| `4xx` | Client Error | There is a problem with the request. |
| `5xx` | Server Error | The server failed to fulfill a valid request. |


# HTTP Status Codes in FastAPI

## 1. Informational Responses (1xx)

| Status Code | Name | Meaning |
|---|---|---|
| **100** | Continue | The client can continue sending the request. |
| **101** | Switching Protocols | The server is switching protocols. |

## 2. Success Responses (2xx)

| Status Code | Name | Meaning |
|---|---|---|
| **200** | OK | The request was successful. |
| **201** | Created | A new resource was created successfully. |
| **202** | Accepted | The request was accepted for processing, but processing may not be complete. |
| **204** | No Content | The request succeeded, but there is no response body. |

## 3. Redirection Responses (3xx)

| Status Code | Name | Meaning |
|---|---|---|
| **301** | Moved Permanently | The resource has permanently moved to another URL. |
| **302** | Found | The resource is temporarily available at another URL. |
| **304** | Not Modified | The cached resource has not changed. |

## 4. Client Error Responses (4xx)

| Status Code | Name | Meaning |
|---|---|---|
| **400** | Bad Request | The request is invalid or malformed. |
| **401** | Unauthorized | Authentication is required or has failed. |
| **403** | Forbidden | The server refuses to authorize the request. |
| **404** | Not Found | The requested resource does not exist or cannot be found. |
| **405** | Method Not Allowed | The HTTP method is not supported for this endpoint. |
| **408** | Request Timeout | The server timed out waiting for the request. |
| **409** | Conflict | The request conflicts with the current resource state. |
| **413** | Content Too Large | The request body exceeds the allowed size. |
| **415** | Unsupported Media Type | The request uses an unsupported content format. |
| **422** | Unprocessable Content | The request content cannot be processed as submitted. FastAPI commonly uses this for request validation errors. |
| **429** | Too Many Requests | The client has sent too many requests in a given period. |

## 5. Server Error Responses (5xx)

| Status Code | Name | Meaning |
|---|---|---|
| **500** | Internal Server Error | An unexpected error occurred on the server. |
| **501** | Not Implemented | The server does not support the requested functionality. |
| **502** | Bad Gateway | A gateway received an invalid response from an upstream server. |
| **503** | Service Unavailable | The server is temporarily unable to handle the request. |
| **504** | Gateway Timeout | A gateway did not receive a timely response from an upstream server. |

## Common Status Codes in FastAPI

| Status Code | When to Use |
|---|---|
| **200 OK** | Successfully retrieve or process data. |
| **201 Created** | Successfully create a resource. |
| **204 No Content** | Successfully complete a request without returning a body. |

| **401 Unauthorized** | Authentication is missing or invalid. |
| **403 Forbidden** | The user is authenticated but lacks permission. |

| **409 Conflict** | A request conflicts with existing data. |
| **422 Unprocessable Content** | Request validation fails. |
| **500 Internal Server Error** | An unexpected server-side error occurs. |

# Difference Between GET and POST Methods

| Feature | GET Method | POST Method |
|---|---|---|
| Purpose | Retrieve data from the server | Submit data to the server |
| Main Use | Fetch or read data | Create or process data |
| Data Location | Usually sent through URL parameters | Usually sent through the request body |
| Request Body | Generally not used | Commonly used |
| Data Visibility | URL parameters may be visible in the browser | Request body is not shown in the URL |
| Data Size | URL length may be limited | Supports larger request bodies, subject to server limits |
| Security | HTTPS is needed for secure transmission | HTTPS is needed for secure transmission |
| Data Modification | Should not modify server data | May modify server data |
| Idempotent | Yes, by HTTP definition | Not necessarily |
| Browser Usage | Can be opened directly in the browser | Usually requires a form, API client, or JavaScript |
| FastAPI Decorator | `@app.get()` | `@app.post()` |
| Example Endpoint | `/products` | `/userprofile` |
| Example Use Case | View available products | Submit customer details |
| Typical Response | Requested data | Confirmation or processed result |



# Special Symbols in URLs

| Symbol | Name | Purpose | Example |
|:---:|---|---|---|
| `/` | Forward Slash | Separates URL paths and represents the root path. | `/users/10` |
| `?` | Question Mark | Starts the query parameters in a URL. | `/users?age=18` |
| `&` | Ampersand | Separates multiple query parameters. | `/users?age=18&city=Delhi` |
| `=` | Equals Sign | Assigns a value to a query parameter. | `/users?name=Rohit` |
| `#` | Hash / Fragment | Identifies a section within a webpage. | `/docs#examples` |
| `*` | Asterisk | Used as a wildcard or special symbol in certain contexts; not a standard URL separator. | Depends on the framework |
