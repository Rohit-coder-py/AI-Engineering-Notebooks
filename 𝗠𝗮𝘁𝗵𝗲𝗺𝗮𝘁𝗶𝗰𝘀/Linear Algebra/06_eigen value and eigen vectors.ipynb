#### An eigenvector is a non-zero vector whose direction remains unchanged after multiplication by a matrix; only its magnitude changes by the eigenvalue.

<B>Av=λv

<b>Where:

A = Matrix (the transformation)

v = Eigenvector

λ (lambda) = Eigenvalue (how much the vector is scaled)

<b>Easy way to remember :<b>

Eigenvector = the special direction that a matrix doesn't rotate.

Eigenvalue = how much the matrix stretches or shrinks that special direction.
