<b>What are Determinants?

A determinant is a single number calculated from a square matrix.

<b>Basically,

Matrix → a table of numbers 

Determinant → one special value obtained from that table 

# <span style="color:#2563EB;">Determinants</span>

## Definition

A **determinant** is a **single numerical value** calculated from a **square matrix**.

**Notation:**

\[
A=
\begin{bmatrix}
a & b\\
c & d
\end{bmatrix}
\qquad
|A|
\]

- **[ ]** → Matrix
- **| |** → Determinant

---

## Conditions

- ✔️ Defined only for **square matrices** (n × n)
- ❌ Not defined for **non-square matrices**

---

## 1 × 1 Determinant

If

\[
A=[a]
\]

then

\[
|A|=a
\]

---

## 2 × 2 Determinant

If

\[
A=
\begin{bmatrix}
a & b\\
c & d
\end{bmatrix}
\]

then

\[
|A|=ad-bc
\]

### Shortcut

\[
\begin{matrix}
a & b\\
c & d
\end{matrix}
\]

\[
(a \times d) - (b \times c)
\]

> **Main Diagonal − Secondary Diagonal**

---

## Example

\[
\begin{vmatrix}
2 & 5\\
3 & 4
\end{vmatrix}
=(2\times4)-(5\times3)
=8-15
=-7
\]

---

## Important Points

- If **|A| = 0** → **Singular Matrix** (No Inverse)
- If **|A| \neq 0** → **Non-Singular Matrix** (Inverse Exists)

---

## Formulas

- Determinant of a 2×2 matrix:

\[
|A| = ad - bc
\]

- If **|A| = 0** → **Singular Matrix** (No Inverse)

- If **|A| ≠ 0** → **Non-Singular Matrix** (Inverse Exists)

# <span style="color:#2563EB;">Minor</span>

## Definition

The **minor** of an element is the **determinant of the matrix obtained after deleting that element's row and column**.

### Steps to Find a Minor

1. Choose an element.
2. Delete its **row** and **column**.
3. Find the determinant of the remaining matrix.
