# Pythagorean Theorem

→ Used only for a **Right-Angled Triangle (90°)**.

> **Hypotenuse² = Base² + Perpendicular²**

---

## Formula

<div align="center">

$$
c^2=a^2+b^2
$$

$$
c=\sqrt{a^2+b^2}
$$

$$
a=\sqrt{c^2-b^2}
$$

$$
b=\sqrt{c^2-a^2}
$$

</div>

---

## Where

- **a** → Base
- **b** → Perpendicular (Height)
- **c** → Hypotenuse (Longest Side)

---

## Remember

→ Applicable **only for right-angled triangles.**

→ **Hypotenuse** is always opposite the **90° angle**.

# General Form of a Line

## Why Convert?

Sometimes the line is given as:

<div align="center">

\[
y = mx + c
\]

</div>

But many Coordinate Geometry formulas require the **General Form**:

<div align="center">

\[
ax + by + c = 0
\]

</div>

**Used in:**
- → Distance of a Point from a Line
- → Distance between Two Parallel Lines
- → Angle between Two Lines

---

## How to Convert?

**Step 1:** Move all terms to one side.

<div align="center">

\[
y = mx + c
\]

↓

\[
mx - y + c = 0
\]

</div>

---

## Example 1

<div align="center">

\[
y = 2x + 3
\]

↓

\[
2x - y + 3 = 0
\]

</div>

---

## Example 2

<div align="center">

\[
y = -4x + 5
\]

↓

\[
4x + y - 5 = 0
\]

</div>

---

## Reverse Conversion

If given:

<div align="center">

\[
ax + by + c = 0
\]

</div>

Solve for **y** to get:

<div align="center">

\[
y = mx + c
\]

</div>

---

## Remember

- **Slope-Intercept Form:** \(y = mx + c\)
- **General Form:** \(ax + by + c = 0\)
- Both represent the **same line**, only written in different forms.
