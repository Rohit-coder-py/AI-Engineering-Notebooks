# Vectors - Basics

---

## What is a Vector?

- A **vector** is a quantity that has:
  - → **Magnitude** (Length / Size)
  - → **Direction**

**Examples:**
- Velocity
- Force
- Acceleration
- Displacement

---

## Magnitude

**Definition:**

> Magnitude is the **length**, **size**, or **amount** of a vector.

A vector tells:
- → **How much?** (Magnitude)
- → **Which way?** (Direction)

**Examples:**

- **5 m East**
  - Magnitude = 5 m
  - Direction = East

- **20 N Right**
  - Magnitude = 20 N
  - Direction = Right

---

## Symbol of Magnitude

If the vector is:

<div align="center">

$$
\vec{A}
$$

</div>

Its magnitude is:

<div align="center">

$$
|\vec{A}|
$$

</div>

---

## Vector Representation

A vector can be written as:

<div align="center">

$$
\vec{A}=\langle x,y\rangle
$$

</div>

**Example:**

<div align="center">

$$
\vec{A}=\langle3,4\rangle
$$

</div>

**Meaning:**

- → x-component = 3
- → y-component = 4

It means:

- Move **3 units** along the **x-axis**.
- Move **4 units** along the **y-axis**.

The vector starts at **(0,0)** and ends at **(3,4)**.

---

## Components of a Vector

For

<div align="center">

$$
\vec{A}=\langle3,4\rangle
$$

</div>

- x-component = 3
- y-component = 4

The numbers inside **⟨ ⟩** are called the **components** of the vector.

---

## Magnitude Formula (2D)

<div align="center">

$$
|\vec{A}|=\sqrt{x^2+y^2}
$$

</div>

**Example:**

<div align="center">

$$
|\vec{A}|=\sqrt{3^2+4^2}
=\sqrt{25}
=5
$$

</div>

---

## Magnitude Formula (3D)

<div align="center">

$$
|\vec{A}|=\sqrt{x^2+y^2+z^2}
$$

</div>

---

## Difference Between Point and Vector

| Point | Vector |
|:------:|:------:|
| Represents a location | Represents movement |
| $(3,4)$ | $\langle3,4\rangle$ |
| Position only | Length + Direction |

> **Note:** A vector **⟨3,4⟩** drawn from the origin ends at the point **(3,4)**.

---

## Memory Tricks

- **Magnitude = "How much?"**
- **Direction = "Which way?"**
- **Components = Movement along each axis**
- **⟨3,4⟩ = Move 3 right, 4 up**
- **|A| = Length of the vector**

# Dot Product (Scalar Product)

## Definition

The dot product is an operation between two vectors that produces a **scalar (number)**.

---

## Symbol

$$
\vec{A}\cdot\vec{B}
$$

---

## Formula (2D)

$$
\vec{A}\cdot\vec{B}=x_1x_2+y_1y_2
$$

---

## Formula (3D)

$$
\vec{A}\cdot\vec{B}=x_1x_2+y_1y_2+z_1z_2
$$

---

## Example

$$
\vec{A}=\langle2,3\rangle
$$

$$
\vec{B}=\langle4,5\rangle
$$

$$
\vec{A}\cdot\vec{B}
=(2\times4)+(3\times5)
=23
$$

---

## Remember

- Vector + Vector → Vector
- Vector · Vector → Scalar (Number)
- Dot Product measures how much two vectors point in the same direction.

````python
A = np.array([2, 3])
B = np.array([4, 5])

dot_product = np.dot(A, B)

print(dot_product)
````

````python
import numpy as np

A = np.array([2, 3])
B = np.array([4, 5])

print(A @ B)       #matrix/vector multiplication operator 
````

<b>The cross product is an operation between two 3D vectors that produces another vector.

---

Unlike the dot product, whose answer is a number, the cross product gives a new vector.

| Dot Product                                   | Cross Product                          |

````python

A = np.array([1,2,3])
B = np.array([4,5,6])

np.cross(A, B)
````

# Cross Product (Vector Product)

## Definition

The cross product is an operation between **two 3D vectors** that produces a **new vector**.

> **Vector × Vector = Vector**

---

## Symbol

$$
\vec{A}\times\vec{B}
$$

---

## Formula (Magnitude)

$$
|\vec{A}\times\vec{B}|
=
|\vec{A}||\vec{B}|\sin\theta
$$

Where:
- $|\vec{A}|$ = Magnitude of A
- $|\vec{B}|$ = Magnitude of B
- $\theta$ = Angle between the vectors

---

## Component Formula (3D)

$$
\vec A\times\vec B=
\begin{vmatrix}
\hat i&\hat j&\hat k\\
a_1&a_2&a_3\\
b_1&b_2&b_3
\end{vmatrix}
$$

---

## Important Facts

- Works **only for 3D vectors**
- Result is **perpendicular (90°)** to both vectors
- Parallel vectors → Cross Product = **0**
- Perpendicular vectors → Cross Product is **maximum**

---

## NumPy

```python
np.cross(A, B)
```
