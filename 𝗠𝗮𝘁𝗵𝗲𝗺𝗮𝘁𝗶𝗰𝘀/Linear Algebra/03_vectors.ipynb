# <span style="color:#2563EB; font-family:Georgia;">What are Vectors?</span>

---

## <span style="color:#7C3AED; font-family:'Trebuchet MS';">Definition</span>

A **<span style="color:#2563EB;">vector</span>** is a **<span style="color:#DC2626;">quantity</span>** that has:

- ➜ **<span style="color:#16A34A; font-family:Verdana;">Magnitude</span>** → <span style="color:#EA580C;"><b><i>Size / Length</i></b></span>

- ➜ **<span style="color:#7C3AED; font-family:Verdana;">Direction</span>** → <span style="color:#0F766E;"><b><i>Where it points</i></b></span>

---

## <span style="color:#F59E0B; font-family:Georgia;">Key Difference</span>

Unlike a **<span style="color:#DC2626;">scalar</span>**, a **<span style="color:#2563EB;">vector</span>** tells you:

- ➜ **<span style="color:#16A34A;"><b>How much</b></span>**
- ➜ **<span style="color:#7C3AED;"><b>Which way</b></span>**

---

## <span style="color:#059669; font-family:'Trebuchet MS';">Examples</span>

- **<span style="color:#2563EB;">Velocity</span>**
- **<span style="color:#DC2626;">Force</span>**
- **<span style="color:#16A34A;">Acceleration</span>**
- **<span style="color:#EA580C;">Displacement</span>**

---

### <span style="color:#B91C1C; font-family:Georgia;">Remember</span>

> **<span style="color:#2563EB;">Vector</span> = <span style="color:#16A34A;">Magnitude</span> + <span style="color:#7C3AED;">Direction</span>**

## <span style="color:#7C3AED; font-family:'Trebuchet MS';">Meaning of the Cap ( ^ )</span>

The **<span style="color:#DC2626;">cap ( ^ )</span>** means the vector is a **<span style="color:#16A34A;">unit vector</span>**.

## <span style="color:#059669; font-family:Georgia;">What is a Unit Vector?</span>

A **<span style="color:#2563EB;">unit vector</span>** is a vector whose **<span style="color:#EA580C;">magnitude (length)</span>** is **<span style="color:#DC2626;"><b>exactly 1</b></span>**.

# Vectors - Basics

---

## What is a Vector?

- A **vector** is a quantity that has:
  - → **Magnitude** (Length / Size)
  - → **Direction**

**Examples:**
- Velocity
- Force
- Acceleration
- Displacement

---

## Magnitude

**Definition:**

> Magnitude is the **length**, **size**, or **amount** of a vector.

A vector tells:
- → **How much?** (Magnitude)
- → **Which way?** (Direction)

**Examples:**

- **5 m East**
  - Magnitude = 5 m
  - Direction = East

- **20 N Right**
  - Magnitude = 20 N
  - Direction = Right

---

## Symbol of Magnitude

If the vector is:

<div align="center">

$$
\vec{A}
$$

</div>

Its magnitude is:

<div align="center">

$$
|\vec{A}|
$$

</div>

---

## Vector Representation

A vector can be written as:

<div align="center">

$$
\vec{A}=\langle x,y\rangle
$$

</div>

**Example:**

<div align="center">

$$
\vec{A}=\langle3,4\rangle
$$

</div>

**Meaning:**

- → x-component = 3
- → y-component = 4

It means:

- Move **3 units** along the **x-axis**.
- Move **4 units** along the **y-axis**.

The vector starts at **(0,0)** and ends at **(3,4)**.

---

## Components of a Vector

For

<div align="center">

$$
\vec{A}=\langle3,4\rangle
$$

</div>

- x-component = 3
- y-component = 4

The numbers inside **⟨ ⟩** are called the **components** of the vector.

---

## Magnitude Formula (2D)

<div align="center">

$$
|\vec{A}|=\sqrt{x^2+y^2}
$$

</div>

**Example:**

<div align="center">

$$
|\vec{A}|=\sqrt{3^2+4^2}
=\sqrt{25}
=5
$$

</div>

---

## Magnitude Formula (3D)

<div align="center">

$$
|\vec{A}|=\sqrt{x^2+y^2+z^2}
$$

</div>

---

## Difference Between Point and Vector

| Point | Vector |
|:------:|:------:|
| Represents a location | Represents movement |
| $(3,4)$ | $\langle3,4\rangle$ |
| Position only | Length + Direction |

> **Note:** A vector **⟨3,4⟩** drawn from the origin ends at the point **(3,4)**.

---

## Memory Tricks

- **Magnitude = "How much?"**
- **Direction = "Which way?"**
- **Components = Movement along each axis**
- **⟨3,4⟩ = Move 3 right, 4 up**
- **|A| = Length of the vector**
