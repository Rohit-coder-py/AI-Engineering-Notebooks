# Straight Line Equation

## Equation

\[
y = mx + c
\]

### Where:

- **x** → Independent variable (horizontal axis)
- **y** → Dependent variable (vertical axis)
- **m** → Slope (Gradient) of the line <b>(m = slope = how fast y changes when x increases by 1.)
- **c** → Y-intercept (Point where the line cuts the y-axis)

### Interpretation

- Increase **m** → Line becomes steeper.
- Positive **m** → Line rises from left to right.
- Negative **m** → Line falls from left to right.
- Increase **c** → Entire line shifts upward.
- Decrease **c** → Entire line shifts downward.

## Example

Given:

\[
y = 2x + 3
\]

Here,

- **m = 2** (Slope)
- **c = 3** (Y-intercept)

The line crosses the y-axis at **(0, 3)**.

# Straight Line — Linear Algebra

## 1. Coordinate Plane

A point in 2D space is represented as:

`P(x, y)`

* **x** → horizontal position
* **y** → vertical position

Example: `P(3, 4)`

---

## 2. Slope of a Line

Slope tells us the **direction and steepness** of a line.

[
m = \frac{y_2-y_1}{x_2-x_1}
]

For points `(x₁, y₁)` and `(x₂, y₂)`:

* `m > 0` → line goes upward
* `m < 0` → line goes downward
* `m = 0` → horizontal line
* `x₁ = x₂` → vertical line, slope undefined

---

## 3. Equation of a Straight Line

The most common form is:

[
y = mx + c
]

Where:

* `m` = slope
* `c` = y-intercept
* `x, y` = coordinates of a point on the line

Example:

[
y = 2x + 3
]

Here, slope = `2` and y-intercept = `3`.

---

## 4. Point-Slope Form

If slope `m` and one point `(x₁, y₁)` are known:

[
y-y_1=m(x-x_1)
]

---

## 5. Parallel & Perpendicular Lines

### Parallel Lines

Parallel lines have the **same slope**:

[
m_1=m_2
]

### Perpendicular Lines

Their slopes satisfy:

[
m_1m_2=-1
]

---

## 6. Distance Between Two Points

For points `(x₁, y₁)` and `(x₂, y₂)`:

[
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
]

---

## AI/ML Connection

Straight lines appear frequently in Machine Learning:

* **Linear Regression** → best-fit line
* **Logistic Regression** → linear decision boundary
* **SVM** → separating hyperplane
* **Neural Networks** → linear transformations followed by nonlinear activations

### Key Idea

> A straight line represents a linear relationship between variables.

# 📈 Finding the Slope (**m**)

> 💡 **Slope (m)** tells us **how steep a line is**.

---

## ✨ Case 1: Equation is Given

If the equation is in the form:

<div align="center">

$$
\boxed{y = mx + c}
$$

</div>

Then,

* **m** = Coefficient of **x**
* **c** = Y-intercept

### Examples

|  Equation | Slope (**m**) |
| :-------: | :-----------: |
|  $y=2x+3$ |     **2**     |
| $y=-5x+1$ |     **-5**    |
|  $y=x+7$  |     **1**     |
|  $y=-x+4$ |     **-1**    |

> ✅ **Remember:** **Just look at the number before `x`.**

---

## ✨ Case 2: Two Points are Given

If two points are given:

$$
(x_1,y_1),\ (x_2,y_2)
$$

Use the slope formula:

<div align="center">

$$
\boxed{
m=\frac{y_2-y_1}{x_2-x_1}
}
$$

</div>

### Example

Given:

$$
(2,3)\ \text{and}\ (4,7)
$$

<div align="center">

$$
m=\frac{7-3}{4-2}
=\frac{4}{2}
=2
$$

</div>

> ✅ **Remember:**
> **Rise = Difference of y-values**
> **Run = Difference of x-values**

---

## ✨ Case 3: From the Graph

Choose **any two points** on the line.

### Step 1️⃣ Find **Run**

➡️ Move **left/right** (Horizontal)

<div align="center">

$$
\boxed{\text{Run}=x_2-x_1}
$$

</div>

### Step 2️⃣ Find **Rise**

⬆️ Move **up/down** (Vertical)

<div align="center">

$$
\boxed{\text{Rise}=y_2-y_1}
$$

</div>

### Step 3️⃣ Calculate Slope

<div align="center">

$$
\boxed{
m=\frac{\text{Rise}}{\text{Run}}
=\frac{\Delta y}{\Delta x}
}
$$

</div>

### Example

```text
        ● B(4,5)
        |
        |  Rise = 3
        |
        ●────────────►
     A(2,2)
        Run = 2
```

<div align="center">

$$
m=\frac{3}{2}
$$

</div>

---

# 🎯 Quick Revision

| Given      | How to Find **m**                              |
| :--------- | :--------------------------------------------- |
| Equation   | Coefficient of **x**                           |
| Two Points | $\displaystyle \frac{y_2-y_1}{x_2-x_1}$        |
| Graph      | $\displaystyle \frac{\text{Rise}}{\text{Run}}$ |

---

## ⭐ Memory Trick

<div align="center">

### 👉 **Run = X-axis (Horizontal)**

### ⬆️ **Rise = Y-axis (Vertical)**

### 🔥 **Slope = Rise ÷ Run**

</div>

> 🧠 **Easy Rule:**
> **X → Run → Left/Right**
> **Y → Rise → Up/Down**
