## Calculus is the branch of mathematics that studies change and accumulation.


Calculus is the mathematics of understanding how things change and how quantities accumulate over time.

# <span style="color:#2563EB;">📘 FUNCTIONS</span>

### <span style="color:#6B7280;">The Foundation of Calculus & Artificial Intelligence</span>

</div>

<div style="background-color:#E0F2FE; border-left:6px solid #0284C7; padding:15px; border-radius:8px;">

# <span style="color:#0369A1;">💡 What is a Function?</span>

A <b style="color:#DC2626;">Function</b> is a <mark>mathematical rule</mark> that takes an <b>input</b>, performs a specific operation, and produces <span style="color:green;"><b>exactly one output</b></span>.

<b>Definition:</b>

A function is a relationship in which <mark>every input has one and only one output.</mark>

</div>

---

<div style="background:#F8FAFC; border:2px solid #CBD5E1; padding:15px; border-radius:8px;">

# <span style="color:#7C3AED;">⚙️ How Does a Function Work?</span>

```text
Input (x)
    │
    ▼
┌────────────────┐
│    Function    │
│  (Rule/Logic)  │
└────────────────┘
    │
    ▼
Output (f(x))
```

### Step 1
Receive an <b style="color:#2563EB;">Input</b>

↓

### Step 2
Apply the <b style="color:#DC2626;">Mathematical Rule</b>

↓

### Step 3
Return the <span style="color:green;"><b>Output</b></span>

</div>

---

<div style="background:#FEFCE8; border-left:6px solid #EAB308; padding:15px; border-radius:8px;">

# <span style="color:#CA8A04;">📝 Example</span>

Given

\[
f(x)=2x+3
\]

Input

\[
x=4
\]

The function applies its rule

\[
2(4)+3
\]

Output

\[
f(4)=11
\]

<b style="color:green;">Input → Rule → Output</b>

</div>

---

<div style="background:#F3E8FF; border-left:6px solid #9333EA; padding:15px; border-radius:8px;">

# <span style="color:#7E22CE;">📌 Function Notation</span>

\[
f(x)
\]

<b style="color:#DC2626;">Important:</b>

<mark>f(x) does NOT mean multiplication.</mark>

It simply means

> <b>"The value produced by the function when the input is x."</b>

</div>

---

<div style="background:#ECFDF5; border-left:6px solid #10B981; padding:15px; border-radius:8px;">

# <span style="color:#047857;">🧩 General Form</span>

\[
y=f(x)
\]

Where

- <b>x</b> → <span style="color:#2563EB;">Input (Independent Variable)</span>

- <b>f(x)</b> or <b>y</b> → <span style="color:green;">Output (Dependent Variable)</span>

</div>

---

<div style="background:#FEF2F2; border-left:6px solid #EF4444; padding:15px; border-radius:8px;">

# <span style="color:#B91C1C;">🚨 Most Important Rule</span>

## <mark>One Input → One Output</mark>

### ✔ Valid Function

```text
2 → 5

3 → 8

4 → 11
```

### ❌ Not a Function

```text
2 → 5

2 → 9
```

<b style="color:#DC2626;">Reason:</b>

The same input cannot produce two different outputs.

</div>

---

<div style="background:#EFF6FF; border-left:6px solid #2563EB; padding:15px; border-radius:8px;">

# <span style="color:#1D4ED8;">📦 Components of a Function</span>

| Component | Meaning |
|-----------|---------|
| <b>Input (x)</b> | Value given to the function |
| <b>Rule</b> | Mathematical operation performed |
| <b>Output (f(x))</b> | Final answer after applying the rule |

</div>

---

<div style="background:#F0FDF4; border-left:6px solid #22C55E; padding:15px; border-radius:8px;">

# <span style="color:#15803D;">🎯 Purpose of Functions</span>

Functions are used to describe the <mark>relationship between quantities.</mark>

Instead of solving every problem separately,

<b style="color:#2563EB;">one mathematical rule</b>

can solve

<span style="color:green;"><b>infinitely many inputs.</b></span>

</div>

---

<div style="background:#EEF2FF; border-left:6px solid #4F46E5; padding:15px; border-radius:8px;">

# <span style="color:#4338CA;">📚 Why Functions Matter in Calculus</span>

Every topic in calculus begins with a function.

```text
Function
     │
     ▼
Limits
     │
     ▼
Derivatives
     │
     ▼
Integration
```

Without functions,

<b style="color:#DC2626;">there is no Calculus.</b>

</div>

---

<div style="background:#F5F3FF; border-left:6px solid #8B5CF6; padding:15px; border-radius:8px;">

# <span style="color:#6D28D9;">🤖 Why Functions Matter in AI</span>

Every Machine Learning model is a <b>Mathematical Function.</b>

```text
Input Data
      │
      ▼
Function (ML Model)
      │
      ▼
Prediction
```

Examples

• Image → Cat

• Email → Spam

• Marks → Grade

• Features → Prediction

</div>

<div style="background:#FFF7ED; border-left:6px solid #F97316; padding:15px; border-radius:8px;">

# <span style="color:#C2410C;">⭐ Quick Summary</span>

✅ A function is a <b>rule.</b>

✅ It takes an <b>input.</b>

✅ It applies a <b>mathematical operation.</b>

✅ It produces <mark>exactly one output.</mark>

✅ Functions describe relationships between quantities.

✅ Calculus studies how functions change.

✅ AI models are mathematical functions.

</div>

<h1 style="color:#1E3A8A;margin-bottom:5px;">📘 Domain & Range</h1>

<p style="font-size:16px;">
A <b>function</b> takes an <span style="color:#2563EB;"><b>input (x)</b></span>,
processes it, and produces an
<span style="color:#16A34A;"><b>output (y)</b></span>.
</p>

<hr>

<h2 style="color:#1D4ED8;">📥 Domain</h2>

<p>
<span style="background:#DBEAFE;padding:4px 8px;border-radius:5px;">
<b>Definition:</b>
The <b>Domain</b> is the set of <b>all possible input values (x)</b>
for which the function is defined.
</span>
</p>

<p>
<b>Think of it as:</b> <span style="color:#2563EB;">"What values am I allowed to enter?"</span>
</p>

<hr>

<h2 style="color:#15803D;">📤 Range</h2>

<p>
<span style="background:#DCFCE7;padding:4px 8px;border-radius:5px;">
<b>Definition:</b>
The <b>Range</b> is the set of <b>all possible output values (y)</b>
produced by the function.
</span>
</p>

<p>
<b>Think of it as:</b> <span style="color:#15803D;">"What values can the function return?"</span>
</p>

<hr>

<p>

<b>Function:</b>

</p>

<p align="center">

<b style="font-size:20px;">f(x) = x²</b>

</p>
