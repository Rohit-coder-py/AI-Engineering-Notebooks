<b> What is limits ? 

<div style="background:#F8FAFC;border-left:8px solid #2563EB;padding:18px;border-radius:10px;">

<h1 style="color:#1E3A8A;">Limits</h1>

<p>
A <b>limit</b> tells us the value that a function <b>approaches</b> as the input (<b>x</b>) approaches a particular value.
</p>

<hr>

<h2 style="color:#15803D;">📘 Definition</h2>

<p>If a function is</p>

<p align="center">
<b>f(x)</b>
</p>

<p>then the limit is written as</p>

<p align="center">
<b>lim<sub>x→a</sub> f(x) = L</b>
</p>

<div style="background:#DBEAFE;padding:10px;border-radius:6px;">

<b>Meaning:</b>

As <b>x</b> gets closer and closer to <b>a</b>, the value of <b>f(x)</b> gets closer and closer to <b>L</b>.

</div>

<hr>

<h2 style="color:#9333EA;">📝 Example</h2>

<p><b>Given:</b></p>

<p align="center">
<b>f(x)=x+2</b>
</p>

<p>Find the limit as x approaches 3.</p>

<table style="border-collapse:collapse;text-align:center;">
<tr>
<th style="border:1px solid black;padding:6px;">x</th>
<th style="border:1px solid black;padding:6px;">f(x)</th>
</tr>
<tr><td style="border:1px solid black;">2.9</td><td style="border:1px solid black;">4.9</td></tr>
<tr><td style="border:1px solid black;">2.99</td><td style="border:1px solid black;">4.99</td></tr>
<tr><td style="border:1px solid black;">2.999</td><td style="border:1px solid black;">4.999</td></tr>
<tr><td style="border:1px solid black;">3.001</td><td style="border:1px solid black;">5.001</td></tr>
<tr><td style="border:1px solid black;">3.01</td><td style="border:1px solid black;">5.01</td></tr>
</table>

<p align="center">
<b>lim<sub>x→3</sub>(x+2)=5</b>
</p>

<hr>

<div style="background:#FEF3C7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#92400E;">💡 Key Idea</h3>

<ul>
<li>The keyword is <b>"Approaches"</b>, not <b>"Equals"</b>.</li>
<li>A limit describes the value the function is getting closer to.</li>
<li>The function may or may not actually reach that value.</li>
</ul>

</div>

<hr>

<div style="background:#E0F2FE;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#0369A1;">🎯 Why Do We Need Limits?</h3>

<ul>
<li>To find the slope of a curve at a single point.</li>
<li>To define derivatives (Differentiation).</li>
<li>Used in Gradient Descent and Deep Learning.</li>
</ul>

</div>

<hr>

<div style="background:#DCFCE7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#166534;">🧠 Memory Trick</h3>

<b>Limit = Destination</b>

<p>
You don't care how you travel.<br>
You only care where you're heading.
</p>

</div>

<h3 style="margin-top:0;color:#6B21A8;">✅ Summary</h3>

<ul>
<li>Limit means <b>approaching a value</b>.</li>
<li>The important word is <b>Approaches</b>.</li>
<li>Notation: <b>lim<sub>x→a</sub>f(x)=L</b></li>
<li>Limits are the foundation of <b>Differentiation</b> and <b>Calculus</b>.</li>
</ul>
