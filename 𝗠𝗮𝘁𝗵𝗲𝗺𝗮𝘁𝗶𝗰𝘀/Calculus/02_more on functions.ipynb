<b> Nested Functions

Notation : <b> f(g(x))

<div style="background:#eef7ff;border-left:6px solid #2196F3;padding:16px;border-radius:8px;">

<h2 style="color:#1565C0;">📌 Nested (Composite) Functions</h2>

<p>
A <span style="color:#D32F2F;font-weight:bold;">nested function</span> is a function
placed <b>inside another function</b>.
The <span style="background:#FFF59D;padding:2px 6px;"><b>output</b></span> of one function
becomes the <span style="background:#C8E6C9;padding:2px 6px;"><b>input</b></span> of another.
</p>

<hr>

<h3 style="color:#2E7D32;">General Form</h3>

<p align="center">

<b style="font-size:18px;">f(g(x))</b>

</p>

<p>
<b>Rule:</b>
<span style="color:#E65100;"><b>Solve the inner function first</b></span>,
then apply the outer function.
</p>

<hr>

<h3 style="color:#2E7D32;">Example</h3>

<p>

g(x) = x + 2

<br>

f(x) = x²

</p>

<p>

<b>f(g(x))</b>

<br>

= f(x + 2)

<br>

= (x + 2)²

</p>

<hr>

<div style="background:#FFF3CD;padding:10px;border-radius:6px;">

<b>⚠️ Important:</b>

<br>

<b>f(g(x)) ≠ g(f(x))</b>

<br>

The <span style="color:#D32F2F;">order matters</span>.

</div>

<hr>

<div style="background:#E8F5E9;padding:10px;border-radius:6px;">

<b>🤖 AI Connection</b>

<br>

Neural Networks are made of many nested functions:

<br><br>

Input → Layer 1 → ReLU → Layer 2 → Softmax → Output

</div>

</div>

<div style="background:#F8FAFC;border-left:8px solid #2563EB;padding:18px;border-radius:10px;">

<h1 style="color:#1E3A8A;">📘 Finding the Domain of a Function</h1>

<p>
The <b>Domain</b> is the set of <b>all valid input values (x)</b> for which a function is defined.
</p>

<hr>

<h2 style="color:#1D4ED8;">📝 Example</h2>

<p align="center">

<b style="font-size:20px;">f(g(x)) = √(x + 1)</b>

</p>

<hr>

<h2 style="color:#15803D;">Step 1 — Apply the Rule</h2>

<div style="background:#E8F5E9;padding:10px;border-radius:6px;">

For a square root function,

<b>the value inside the root must always be ≥ 0.</b>

</div>

<p align="center">

<b>x + 1 ≥ 0</b>

</p>

<hr>

<h2 style="color:#15803D;">Step 2 — Solve the Inequality</h2>

<p>

Subtract <b>1</b> from both sides:

</p>

<p align="center">

<b>x + 1 − 1 ≥ 0 − 1</b>

<br><br>

<b>x ≥ −1</b>

</p>

<div style="background:#DBEAFE;padding:10px;border-radius:6px;">

<b>Why subtract 1?</b>

<br>

Because the expression inside the root is <b>x + 1</b>.
To isolate <b>x</b>, subtract the same number from both sides.

</div>

<hr>

<h2 style="color:#9333EA;">Step 3 — Write the Domain</h2>

<p>

Since

<b>x ≥ −1</b>,

the domain is

</p>

<p align="center">

<b style="font-size:20px;">[-1, ∞)</b>

</p>

<hr>

<div style="background:#FEF3C7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#92400E;">📌 Interval Notation</h3>

<b>[</b> or <b>]</b> → Endpoint is <b>included</b> (≤ or ≥)

<br><br>

<b>(</b> or <b>)</b> → Endpoint is <b>not included</b> (< or >)

</div>

<hr>

<div style="background:#DCFCE7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#166534;">✅ Final Answer</h3>

<b>Condition:</b>

<br>

x + 1 ≥ 0

<br><br>

<b>Solution:</b>

<br>

x ≥ −1

<br><br>

<b>Domain:</b>

<br>

<span style="font-size:18px;"><b>[-1, ∞)</b></span>

</div>

<hr>

<div style="background:#F3F4F6;padding:12px;border-radius:8px;">

<b>💡 Remember</b>

<ul>
<li><b>Condition</b> → Restriction on the input.</li>
<li><b>Solution</b> → Solve the restriction for <b>x</b>.</li>
<li><b>Domain</b> → Write the solution in <b>interval notation</b>.</li>
</ul>

</div>

<b> <mark> Functions Operations 

1.<b> Multiplying a Functions with Scaler 

<div style="background:#F8FAFC;border-left:8px solid #2563EB;padding:18px;border-radius:10px;">

<h1 style="color:#1E3A8A;">Multiplying a Function by a Scalar</h1>

<p>
A <b>scalar</b> is simply a <b>constant number</b> (e.g., 2, -5, 0.5, π).
</p>

<hr>

<h2 style="color:#15803D;"> Definition</h2>

<p>

If

</p>

<p align="center">

<b>f(x)</b>

</p>

<p>

is a function and <b>k</b> is a scalar, then

</p>

<p align="center">

<b>(kf)(x) = k · f(x)</b>

</p>

<div style="background:#DBEAFE;padding:10px;border-radius:6px;">

<b>Meaning:</b>

Multiply the <b>output</b> of the function by the scalar.

</div>

<hr>

<h2 style="color:#9333EA;">📝 Example 1</h2>

<p>

<b>Given:</b>

</p>

<p align="center">

<b>f(x) = x²</b>

</p>

<p>

Multiply by <b>3</b>:

</p>

<p align="center">

<b>3f(x) = 3x²</b>

</p>

<hr>

<h2 style="color:#9333EA;">📝 Example 2</h2>

<p>

<b>Given:</b>

</p>

<p align="center">

<b>f(x) = sin(x)</b>

</p>

<p>

Multiply by <b>5</b>:

</p>

<p align="center">

<b>5f(x) = 5sin(x)</b>

</p>

<hr>

<div style="background:#FEF3C7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#92400E;">📈 Effect on the Graph</h3>

<ul>
<li><b>k > 1</b> → Vertical Stretch</li>
<li><b>0 < k < 1</b> → Vertical Compression</li>
<li><b>k < 0</b> → Reflection about the x-axis + Stretch/Compression</li>
</ul>

</div>

<hr>

<div style="background:#DCFCE7;padding:12px;border-radius:8px;">

<h3 style="margin-top:0;color:#166534;">✅ Key Points</h3>

<ul>
<li><b>Scalar</b> = Constant number.</li>
<li>Multiply the <b>output</b>, <b>not the input</b>.</li>
<li><b>Formula:</b> (kf)(x) = k · f(x)</li>
<li>The <b>domain remains unchanged</b>.</li>
<li>The <b>range is multiplied by the scalar</b>.</li>
</ul>

</div>

2. <b> Internal Scaling 

# 📘 Internal Scaling (Multiplying x Inside the Function)

## Definition

Internal Scaling means **multiplying the input `x` inside the function** by a constant.

\[
f(x)\rightarrow f(kx)
\]

where `k` is called the **scaler**.

---

## Rule

- **Outside multiplier** → Vertical Scaling
  \[
  kf(x)
  \]

- **Inside multiplier** → Horizontal (Internal) Scaling
  \[
  f(kx)
  \]

---

## Steps to Solve

1. Write the original function.
2. Replace every `x` with `kx`.
3. Simplify using algebra.
4. Write the final answer.

---

## Example 1

Given

\[
f(x)=x^2
\]

Find

\[
f(2x)
\]

### Step 1: Replace x with 2x

\[
f(2x)=(2x)^2
\]

### Step 2: Expand

\[
(2x)^2=(2x)(2x)
\]

### Step 3: Multiply

\[
=2\times2\times x\times x
\]

\[
=4x^2
\]

### Final Answer

\[
\boxed{f(2x)=4x^2}
\]

---

## Example 2

\[
f(x)=x^2
\]

Find

\[
f(3x)
\]

\[
=(3x)^2
\]

\[
=3^2x^2
\]

\[
=9x^2
\]

---

## Example 3

\[
f(x)=x^2
\]

Find

\[
f\left(\frac{x}{2}\right)
\]

\[
=\left(\frac{x}{2}\right)^2
\]

\[
=\frac{x^2}{4}
\]

---

# Important Algebra Rule

\[
(ax)^n=a^nx^n
\]

Examples

\[
(2x)^2=4x^2
\]

\[
(3x)^2=9x^2
\]

\[
(2x)^3=8x^3
\]

\[
(5x)^2=25x^2
\]

---

# Graph Effect

| Function | Effect |
|----------|--------|
| \(f(2x)\) | Horizontal Compression |
| \(f(3x)\) | More Horizontal Compression |
| \(f\left(\frac{x}{2}\right)\) | Horizontal Stretch |

---

# Vertical vs Horizontal Scaling

| Vertical Scaling | Horizontal (Internal) Scaling |
|------------------|-------------------------------|
| \(2f(x)\) | \(f(2x)\) |
| Changes Height (y-values) | Changes Width (x-values) |

---

# Memory Trick 🧠

- **Outside multiplier** → Height changes 📈
- **Inside multiplier** → Width changes ↔️

---

# Summary

- Replace every **x** with **kx**.
- Simplify using algebra.
- Use the rule

\[
(ax)^n=a^nx^n
\]

- \(k>1\) → Horizontal Compression.
- \(0<k<1\) → Horizontal Stretch.
