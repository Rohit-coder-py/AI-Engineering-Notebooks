# Product Rule

The **Product Rule** is used when **two functions are multiplied together**.

## Formula

$$
\frac{d}{dx}[f(x)g(x)] = f'(x)g(x) + f(x)g'(x)
$$

## When to Use?

Use the Product Rule whenever your function is in the form:

$$
f(x)\times g(x)
$$

## Examples

- $x^2\sin x$
- $xe^x$
- $(x+1)(x^2+3)$
- $(\ln x)(x^3)$
- $(x^2+1)\cos x$

> **Tip:** If you see **one function × another function**, think **Product Rule**.

<h1 style="color:#1E3A8A;">Product Rule</h1>

<p>
The <b>Product Rule</b> is used when <b>two functions are multiplied together</b>.
</p>

<hr>

<h2 style="color:#15803D;">📘 Definition</h2>

<p>If</p>

<p align="center">
<b>y = f(x)g(x)</b>
</p>

<p>then</p>

<p align="center">
<b>(fg)' = f'g + fg'</b>
</p>

<div style="background:#DBEAFE;padding:10px;border-radius:6px;">

<b>Meaning:</b>

Differentiate the <b>first function</b> and keep the <b>second</b> the same, then add the <b>first function</b> multiplied by the <b>derivative of the second</b>.

</div>

<hr>

<h2 style="color:#9333EA;">📝 Formula</h2>

<p align="center">

<b>Derivative of First × Second + First × Derivative of Second</b>

</p>

<p align="center">

<b>f'g + fg'</b>

</p>

<hr>

<h2 style="color:#9333EA;">📝 Example</h2>

<p><b>Given:</b></p>

<p align="center">
<b>y = x² sin(x)</b>
</p>

<p><b>Step 1:</b></p>

<ul>
<li>First Function: <b>f = x²</b></li>
<li>Second Function: <b>g = sin(x)</b></li>
</ul>

<p><b>Step 2:</b></p>

<ul>
<li><b>f' = 2x</b></li>
<li><b>g' = cos(x)</b></li>
</ul>

<p><b>Step 3:</b> Apply Product Rule</p>

<p align="center">
<b>(2x)(sin(x)) + (x²)(cos(x))</b>
</p>

<p><b>Answer:</b></p>

<p align="center">
<b>2x sin(x) + x² cos(x)</b>
</p>

<hr>

Use the Product Rule whenever <b>two functions are multiplied</b>.

<ul>
<li>x sin(x)</li>
<li>x² e<sup>x</sup></li>
<li>(x+1)(x²+3)</li>
<li>x ln(x)</li>
</ul>

Do <b>NOT</b> multiply the derivatives.

<p align="center">

</p>

Always use

<p align="center">
✅ f'g + fg'
</p>

<ul>
<li>Used when <b>two functions are multiplied</b>.</li>

<b>2.<mark> Chain Rule

<b>The Chain Rule is used when one function is inside another function.

# 📦 Example

Suppose

$$
y=(x^2+1)^5
$$

Notice this has **two functions**.

### Outer Function

$$
(\square)^5
$$

The outer function is:

$$
u^5
$$

### Inner Function

$$
x^2+1
$$

### Why Use the Chain Rule?

Since **one function is inside another function**, we use the **Chain Rule**.

> **Tip:** If you can replace the inside part with a single variable (like \(u\)), it's a Chain Rule problem.

# Formula

$$
\frac{dy}{dx}=\frac{dy}{du}\times\frac{du}{dx}
$$

### Or simply

$$
\left(f(g(x))\right)' = f'(g(x))\cdot g'(x)
$$

### Read it as:

> **Differentiate the outside function first, then multiply by the derivative of the inside function.**

### Easy Way to Remember

**Outside Derivative × Inside Derivative**

or

**Outside → Inside**

| Rule             | When to Use                    | Example      |
| ---------------- | ------------------------------ | ------------ |
| **Sum Rule**     | Functions are added/subtracted | (x^2+\sin x) |
| **Product Rule** | Functions are multiplied       | (x^2\sin x)  |
| **Chain Rule**   | One function is inside another | ((x^2+1)^5)  |
