# 📌 Application of $\dfrac{dy}{dx}$

> <span style="color:#1E90FF; font-size:18px;"><b>Derivative tells us how one quantity changes with respect to another.</b></span>

---

## 1️⃣ Slope of a Curve 📈

For

$$
y=f(x)
$$

the derivative

$$
\frac{dy}{dx}
$$

gives the <span style="color:#2E8B57;"><b>slope of the tangent</b></span> at any point.

**Example**

$$
y=x^2
$$

$$
\frac{dy}{dx}=2x
$$

At $x=3$,

$$
\frac{dy}{dx}=6
$$

✅ **Slope = 6**

---

## 2️⃣ Rate of Change ⚡

<span style="background-color: #FFF3CD; padding:2px;"><b>Derivative measures how fast something changes.</b></span>

Examples:

- 🚗 Distance → Speed
- ⚡ Speed → Acceleration
- 📈 Population → Growth Rate
- 🌡️ Temperature → Cooling Rate

---

## 3️⃣ Maximum & Minimum ⭐

To find the highest or lowest value:

1. Find the derivative.
2. Set

$$
\frac{dy}{dx}=0
$$

3. Solve for $x$.

<span style="color:#DC143C;"><b>These are called Critical Points.</b></span>

---

## 4️⃣ Increasing & Decreasing Functions 📊

If

$$
\frac{dy}{dx}>0
$$

➡️ <span style="color:green;"><b>Function is Increasing</b></span>

If

$$
\frac{dy}{dx}<0
$$

➡️ <span style="color:red;"><b>Function is Decreasing</b></span>

---

## 5️⃣ Velocity 🚗

If

$$
s(t)=t^2
$$

then

$$
v=\frac{ds}{dt}=2t
$$

> <span style="color:#8A2BE2;"><b>Velocity = Derivative of Distance</b></span>

---

## 6️⃣ Acceleration 🚀

Differentiate velocity again:

$$
a=\frac{dv}{dt}
$$

> <span style="color:#FF8C00;"><b>Acceleration = Derivative of Velocity</b></span>

---

## 7️⃣ Tangent Line 📏

Derivative gives the <b>slope of the tangent line</b>.

Tangent equation:

$$
y-y_1=m(x-x_1)
$$

where

$$
m=\frac{dy}{dx}
$$

---

## 8️⃣ Normal Line 📐

Normal line is <span style="color:#20B2AA;"><b>perpendicular</b></span> to the tangent.

If tangent slope is

$$
m
$$

then

$$
m_{\text{normal}}=-\frac1m
$$

---

## 9️⃣ Optimization 💰

Derivatives help solve:

- Maximum Profit
- Minimum Cost
- Largest Area
- Shortest Distance
- Best Design

---

## 🔟 AI & Machine Learning 🤖

<span style="background-color:#D4EDDA; padding:3px;"><b>Deep Learning is built on derivatives.</b></span>

Used in:

- Gradient Descent
- Backpropagation
- Loss Function
- Optimizers (SGD, Adam)

Example:

$$
\frac{\partial \text{Loss}}{\partial \text{Weight}}
$$

This tells the model **how to update its weights** during training.
