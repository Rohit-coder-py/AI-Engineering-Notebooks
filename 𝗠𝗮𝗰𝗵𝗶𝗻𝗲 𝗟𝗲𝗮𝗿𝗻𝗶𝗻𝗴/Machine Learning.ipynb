# Artificial Intelligence vs Machine Learning vs Deep Learning

### Understanding the Difference

</div>

---

## Artificial Intelligence (AI)

<b>Artificial Intelligence</b> is the <mark>broadest field</mark> focused on creating systems that can perform tasks requiring human intelligence.

### What AI Can Do
- Reason and make decisions
- Understand language
- Solve problems
- Automate tasks

### Examples
- ChatGPT
- Virtual Assistants
- Self-Driving Cars
- Recommendation Systems

### Key Point

<b>AI aims to make machines behave intelligently.</b>

---

## Machine Learning (ML)

<b>Machine Learning</b> is a <mark>subset of AI</mark> that allows computers to learn patterns from data instead of being explicitly programmed.

### What ML Does
- Learns from historical data
- Makes predictions
- Finds patterns
- Improves with experience

### Examples
- Spam Detection
- Stock Prediction
- House Price Prediction
- Customer Churn Prediction

### Key Point

<b>ML focuses on learning from data.</b>

---

## Deep Learning (DL)

<b>Deep Learning</b> is a <mark>subset of Machine Learning</mark> that uses multi-layered neural networks to learn complex patterns automatically.

### What DL Does
- Processes large datasets
- Learns features automatically
- Handles images, audio, and text
- Powers modern AI systems

### Examples
- Face Recognition
- Speech Recognition
- Image Classification
- Large Language Models (LLMs)

### Key Point

<b>DL uses deep neural networks to solve complex problems.</b>

<b> <mark style="background-color: #90EE90;">Feature Scaling</mark></b>

Feature Scaling is the process of bringing all numerical features to a similar scale so that no feature dominates others because of its larger values.

### syntax only

# Standard Scaler

# Values are not restricted between 0 and 1.
# Negative values are possible

    
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_scaled = scaler.fit_transform()


#MinMax Scaler

# All values lie between 0 and 1.
# No negative values

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform()


| Use **StandardScaler**               | Use **MinMaxScaler**                                    |
| ------------------------------------ | ------------------------------------------------------- |
| Data is roughly normally distributed | Data is not normally distributed                        |
| Using Linear Regression              | Using Neural Networks                                   |
| Using Logistic Regression            | Image pixel data (0–255)                                |
| Using SVM                            | Need values strictly between 0 and 1                    |
| Using PCA                            | Want to preserve the original shape of the distribution |
| Features can have negative values    | Prefer all positive values                              |

<h2>
    <mark style="background-color:#dbeafe;">
        Standardization (Z-Score) Formula
    </mark>
</h2>

<h1>
    <span style="color:#2563eb;">z</span>
    =
    <span style="color:#dc2626;">(x − μ)</span>
    /
    <span style="color:#16a34a;">σ</span>
</h1>

<p>
    <b style="color:#2563eb;">z</b> =
    <mark style="background-color:#bfdbfe;">
        Standardized Value
    </mark>
</p>

<p>
    <b style="color:#dc2626;">x</b> =
    <mark style="background-color:#fecaca;">
        Original Value
    </mark>
</p>

<p>
    <b style="color:#f59e0b;">μ (Mu)</b> =
    <mark style="background-color:#fde68a;">
        Mean of the Feature
    </mark>
</p>

<p>
    <b style="color:#16a34a;">σ (Sigma)</b> =
    <mark style="background-color:#bbf7d0;">
        Standard Deviation
    </mark>
</p>

</div>

```python
data = {
    'StudyHours': [1, 2, 3, 4, 5],
    'TestScore': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)


from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)


print("Scaled Data: ")

pd.DataFrame(scaled_data,columns = ['StudyHours','TestScore'])
```

<h2>
    <mark style="background-color:#dcfce7;">

</h1>

<p>
    <b style="color:#2563eb;">x'</b> =
    <mark style="background-color:#bfdbfe;">
        Scaled Value (Between 0 and 1)
    </mark>
</p>

<p>
    <b style="color:#dc2626;">x</b> =
    <mark style="background-color:#fecaca;">
        Original Value
    </mark>
</p>

<p>
    <b style="color:#f59e0b;">x<sub>min</sub></b> =
    <mark style="background-color:#fde68a;">
        Minimum Value in Dataset
    </mark>
</p>

<p>
    <b style="color:#16a34a;">x<sub>max</sub></b> =
    <mark style="background-color:#bbf7d0;">
        Maximum Value in Dataset
    </mark>
</p>

<hr>

</div>

```python
# use minmax scaler when you want values lies between 0 to 1

from sklearn.preprocessing import MinMaxScaler

data = {
    'StudyHours': [1, 2, 3, 4, 5],
    'TestScore': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

minmax = MinMaxScaler()

minmax_scaled = minmax.fit_transform(df)

scaled_df = pd.DataFrame(minmax_scaled , columns = ['StudyHours','TestScore'])

scaled_df
```

<b> <mark style="background-color: #90EE90;">Split Data</mark></b>

```python
from sklearn.model_selection import train_test_split

# Remember :

# x = input and y = output to train your model

data = {
    'StudyHours': [1, 2, 3, 4, 5],
    'TestScore': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

X = df[['StudyHours']]                  # passing it in dataframe
Y = df[['TestScore']]                   # passing it in dataframe

X_train , X_test , Y_train , Y_test = train_test_split(X,Y , test_size = 0.2  , random_state=42 )


print("Train Data :\n")

print(X_train)

print("\nTest Data :\n")

print(X_test)

print("Train Data :\n")

print(Y_train)

print("\nTest Data :\n")

Y_test

```

```python
how_it_works ="""

1- take the mistake difference
2- remove the minus sign
3- add
4- divide 

"""
```

```python
from sklearn.metrics import mean_absolute_error

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mae = mean_absolute_error(y_true, y_pred)

print("MAE:", mae)

# The model predictions differ from the actual values by about 6.25 units on average
# USE in Real World : Judge Model Quality
```

<b>2. <b><mark>MSE (Mean Squared Error)

```python
p='''
MSE Mean Squared Error
1- Mistakes square them
2- add
3- divide total'''
```

```python
# just square the mistakes finded by MAE so that machine can find patterens and errors easily 

# MSE measures the average squared prediction error and is used because it penalizes large errors much more heavily than MAE

from sklearn.metrics import mean_squared_error

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mse = mean_squared_error(y_true, y_pred)

print("MSE:", mse)

# Use in Real World: Punishes large mistakes more heavily
```

<b>3. <b><mark> RSME - Root Mean Square Error 

```python
from sklearn.metrics import mean_squared_error
import math

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mse = mean_squared_error(y_true, y_pred)

rmse = math.sqrt(mse)

print("RMSE:", rmse)
```

# Variance

## 🔹 Definition

<b style="color:#2563eb;">Variance</b> measures how far data points are <b style="color:#16a34a;">spread from the mean</b>.

---

## 🔹 Formula

σ² = Σ(x - μ)² / N

Where:

- <b style="color:#2563eb;">σ²</b> = Variance
- <b style="color:#2563eb;">x</b> = Data Value
- <b style="color:#2563eb;">μ</b> = Mean
- <b style="color:#2563eb;">N</b> = Number of Values

---

## 🔹 Interpretation

- <b style="color:#16a34a;">Low Variance</b> → Data is close to the mean.
- <b style="color:#dc2626;">High Variance</b> → Data is spread out.

---

## 🔹 When to Use

- Measure Data Spread
- Statistics
- Data Analysis
- Machine Learning

---

## 📌 Key Idea

<b style="color:#7c3aed;">Variance</b> tells us how much the data varies from its <b style="color:#16a34a;">average value (mean)</b>.

### 5 Steps Of How K-Mean Model works

<div style="border:1px solid #ddd; border-radius:12px; padding:20px;">

# <span style="color:#2563eb;">K-Means Clustering: 5-Step Formula</span>

## <span style="color:#7c3aed;">Step 1: Choose K</span>

Tell the machine how many <b style="color:#16a34a;">clusters (K)</b> you want.

**Example:**

K = 3

---

## <span style="color:#7c3aed;">Step 2: Select Random Centroids</span>

Randomly place <b style="color:#16a34a;">K centroids</b> in the dataset.

**Example:**

K = 3 → Select 3 random centroids.

---

## <span style="color:#7c3aed;">Step 3: Assign Points</span>

Calculate the <b style="color:#16a34a;">Euclidean Distance</b> from each data point to every centroid.

Assign the point to the <b style="color:#16a34a;">nearest centroid</b>.

---

## <span style="color:#7c3aed;">Step 4: Update Centroids</span>

Find the <b style="color:#16a34a;">mean</b> of all points in each cluster.

The new mean becomes the new <b style="color:#16a34a;">centroid</b>.

---

## <span style="color:#7c3aed;">Step 5: Repeat</span>

Repeat <b style="color:#16a34a;">Steps 3 and 4</b> until centroids stop changing significantly.

At this point, the clusters are finalized.

---

## 📌 <span style="color:#2563eb;">Key Idea</span>

<b style="color:#2563eb;">Choose K</b>

⬇️

<b style="color:#7c3aed;">Random Centroids</b>

⬇️

<b style="color:#16a34a;">Assign Points</b>

⬇️

<b style="color:#ea580c;">Update Centroids</b>

⬇️

<b style="color:#dc2626;">Repeat Until Stable</b>

</div>


```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Sample data
X = [
    [1, 2],[1, 3],[2, 2],[8, 8],[9, 8],[8, 9]]

# Create model
model = KMeans(n_clusters=2, random_state=42)

# Train model
model.fit(X)

# Cluster assigned to each point
print("Clusters:", model.labels_)

# Centroids
print("Centroids:")
print(model.cluster_centers_)

# Visualization
for point, cluster in zip(X, model.labels_):
    plt.scatter(point[0], point[1])

plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker="X",
    s=200
)

plt.show()
```

<b><mark>PCA (Principal Component Analysis)

<b><b>PCA (Principal Component Analysis) is a dimensionality reduction technique that transforms correlated features into a smaller set of uncorrelated principal components while preserving as much information (variance) as possible.


```python
#syntax

# from sklearn.decomposition import PCA

# pca = PCA(n_components=2)

# X_pca = pca.fit_transform(X)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# why to use StandardScaler before PCA?

# "Because PCA is variance-based. Features with
# larger scales have larger variances and can dominate the principal components.
# Standardizing the data ensures that each feature contributes fairly

data = {
    'Age': [25, 30, 35, 40, 45, 50],
    'Income': [30000, 40000, 50000, 60000, 70000, 80000],
    'Spending': [70, 60, 50, 40, 30, 20],
    'Savings': [1000, 5000, 8000, 10000, 15000, 20000]
}

df = pd.DataFrame(data)

# first scaling

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

# pca

pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])

# now Variance explained

# How much information (variance) a Principal Component captures from the original data. it is % 

exp_var = pca.explained_variance_ratio_

print(exp_var)

# total variance explained

print(sum(exp_var)) # always 1
```

WHY PCA IS USED

- Reduces the number of features (dimensions)
- Removes redundant information from correlated features
- Preserves most of the important information (variance)
- Makes model training faster
- Reduces memory usage
- Helps reduce noise in the data
- Makes high-dimensional data easier to visualize
- Can improve the performance of some ML algorithms

WHEN TO USE PCA

- When the dataset has many features (10, 20, 100+)
- When features are highly correlated
- Before clustering algorithms like K-Means
- When training is slow due to many features
- When you want to visualize high-dimensional data in 2D or 3D
- When you want to reduce storage and computation requirements

WHEN NOT TO USE PCA

- When the dataset has only a few features
- When feature interpretability is important
- When original feature names must be preserved
- Often unnecessary for Random Forest, XGBoost, and Decision Trees
- When the dataset is already small and simple

```python
# plot of PCA

plt.figure(figsize=(8,6))
plt.scatter(pca_df['PCA1'], pca_df['PCA2'], color='blue',s = 80)
plt.title("PCA Projection (2D View)")
plt.xlabel('PCA1 Main Pattern')
plt.ylabel('PCA2 Minor Pattern')
plt.grid(True)
plt.show()
```
