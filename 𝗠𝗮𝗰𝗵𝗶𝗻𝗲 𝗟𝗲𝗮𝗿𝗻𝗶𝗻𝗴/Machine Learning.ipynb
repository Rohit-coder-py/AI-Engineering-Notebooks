# Artificial Intelligence vs Machine Learning vs Deep Learning

### Understanding the Difference

</div>

---

## Artificial Intelligence (AI)

<b>Artificial Intelligence</b> is the <mark>broadest field</mark> focused on creating systems that can perform tasks requiring human intelligence.

### What AI Can Do
- Reason and make decisions
- Understand language
- Solve problems
- Automate tasks

### Examples
- ChatGPT
- Virtual Assistants
- Self-Driving Cars
- Recommendation Systems

### Key Point

<b>AI aims to make machines behave intelligently.</b>

---

## Machine Learning (ML)

<b>Machine Learning</b> is a <mark>subset of AI</mark> that allows computers to learn patterns from data instead of being explicitly programmed.

### What ML Does
- Learns from historical data
- Makes predictions
- Finds patterns
- Improves with experience

### Examples
- Spam Detection
- Stock Prediction
- House Price Prediction
- Customer Churn Prediction

### Key Point

<b>ML focuses on learning from data.</b>

---

## Deep Learning (DL)

<b>Deep Learning</b> is a <mark>subset of Machine Learning</mark> that uses multi-layered neural networks to learn complex patterns automatically.

### What DL Does
- Processes large datasets
- Learns features automatically
- Handles images, audio, and text
- Powers modern AI systems

### Examples
- Face Recognition
- Speech Recognition
- Image Classification
- Large Language Models (LLMs)

### Key Point

<b>DL uses deep neural networks to solve complex problems.</b>

<b> <mark style="background-color: #90EE90;">Feature Scaling</mark></b>

Feature scaling is the process of bringing all numerical features to a similar scale so that no feature disproportionately influences others because of its larger values

### syntax only

# Standard Scaler

# Values are not restricted between 0 and 1.
# Negative values are possible

    
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_scaled = scaler.fit_transform()


## MinMax Scaler

# All values lie between 0 and 1.
# No negative values

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform()


| Use **StandardScaler**               | Use **MinMaxScaler**                                    |
| ------------------------------------ | ------------------------------------------------------- |
| Data is roughly normally distributed | Data is not normally distributed                        |
| Using Linear Regression              | Using Neural Networks                                   |
| Using Logistic Regression            | Image pixel data (0–255)                                |
| Using SVM                            | Need values strictly between 0 and 1                    |
| Using PCA                            | Want to preserve the original shape of the distribution |
| Features can have negative values    | Prefer all positive values                              |

<h2>
    <mark style="background-color:#dbeafe;">
        Standardization (Z-Score) Formula
    </mark>
</h2>

<h1>
    <span style="color:#2563eb;">z</span>
    =
    <span style="color:#dc2626;">(x − μ)</span>
    /
    <span style="color:#16a34a;">σ</span>
</h1>

<p>
    <b style="color:#2563eb;">z</b> =
    <mark style="background-color:#bfdbfe;">
        Standardized Value
    </mark>
</p>

<p>
    <b style="color:#dc2626;">x</b> =
    <mark style="background-color:#fecaca;">
        Original Value
    </mark>
</p>

<p>
    <b style="color:#f59e0b;">μ (Mu)</b> =
    <mark style="background-color:#fde68a;">
        Mean of the Feature
    </mark>
</p>

<p>
    <b style="color:#16a34a;">σ (Sigma)</b> =
    <mark style="background-color:#bbf7d0;">
        Standard Deviation
    </mark>
</p>

</div>

````python
data = {
    'StudyHours': [1, 2, 3, 4, 5],
    'TestScore': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)


from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)


print("Scaled Data: ")

pd.DataFrame(scaled_data,columns = ['StudyHours','TestScore'])
````

<b> <strong><mark>2. MinMaxScaler

    background:#ffffff;

<h2>
    <mark style="background-color:#dcfce7;">
        Min-Max Scaling Formula
    </mark>
</h2>

<h1>
    <span style="color:#2563eb;">x'</span>
    =
    <span style="color:#dc2626;">(x − x<sub>min</sub>)</span>
    /
    <span style="color:#16a34a;">(x<sub>max</sub> − x<sub>min</sub>)</span>
</h1>

<p>
    <b style="color:#2563eb;">x'</b> =
    <mark style="background-color:#bfdbfe;">
        Scaled Value (Between 0 and 1)
    </mark>
</p>

<p>
    <b style="color:#dc2626;">x</b> =
    <mark style="background-color:#fecaca;">
        Original Value
    </mark>
</p>

<p>
    <b style="color:#f59e0b;">x<sub>min</sub></b> =
    <mark style="background-color:#fde68a;">
        Minimum Value in Dataset
    </mark>
</p>

<p>
    <b style="color:#16a34a;">x<sub>max</sub></b> =
    <mark style="background-color:#bbf7d0;">
        Maximum Value in Dataset
    </mark>
</p>

<hr>

</div>

````python
# use minmax scaler when you want values lies between 0 to 1

from sklearn.preprocessing import MinMaxScaler

data = {
    'StudyHours': [1, 2, 3, 4, 5],
    'TestScore': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

minmax = MinMaxScaler()

minmax_scaled = minmax.fit_transform(df)

scaled_df = pd.DataFrame(minmax_scaled , columns = ['StudyHours','TestScore'])

scaled_df
````

<b> Cost Function 

<mark>Cost Function </mark>= Sum of all Residual Errors


## Cost Function

The Cost Function measures how well the regression line fits the training data.

It calculates the average of the squared residual errors for all training examples.

### Formula

$$
J(\theta_0,\theta_1)
=
\frac{1}{2m}
\sum_{i=1}^{m}
\left(
h_\theta(x^{(i)}) - y^{(i)}
\right)^2
$$

### Where:

- $J(\theta_0,\theta_1)$ = Cost Function (total error)
- $\theta_0$ = Intercept (bias)
- $\theta_1$ = Slope (weight)
- $m$ = Number of training examples
- $h_\theta(x^{(i)})$ = Predicted value for the $i^{th}$ example
- $y^{(i)}$ = Actual value for the $i^{th}$ example
- $\sum$ = Sum of errors for all training examples

### Goal

Find the values of $\theta_0$ and $\theta_1$ that minimize the Cost Function.

### Intuition

- Large prediction errors → Large Cost
- Small prediction errors → Small Cost
- Lowest Cost → Best-Fit Line

To find best fit line we have to minimize the value of cost function by changing value of $\theta_0$ and $\theta_1$

To do that we use : <mark> Gradient Descent

# Gradient Descent

Gradient Descent is an optimization algorithm used to find the best values of $\theta_0$ and $\theta_1$ by minimizing the Cost Function.

## Simple Idea

Imagine you are standing on a hill and want to reach the lowest point.

* Look at the slope around you.
* Take a small step downhill.
* Repeat until you reach the bottom.

Gradient Descent works the same way:

* Start with random values of $\theta_0$ and $\theta_1$
* Calculate the Cost Function
* Update $\theta_0$ and $\theta_1$
* Reduce the Cost
* Repeat until the Cost is minimum

## Goal

Find the values of $\theta_0$ and $\theta_1$ that produce the lowest Cost Function value.

## Workflow

```text
Choose θ₀ and θ₁
        ↓
Calculate Cost Function
        ↓
Update θ₀ and θ₁
        ↓
Cost Decreases
        ↓
Repeat
        ↓
Minimum Cost Found
        ↓
Best-Fit Line Obtained
```

## In One Sentence

Gradient Descent is the method used to minimize the Cost Function and find the best values of $\theta_0$ and $\theta_1$.


## Repeat Convergence Theorem

Gradient Descent repeatedly updates the values of $\theta_0$ and $\theta_1$ until the Cost Function stops decreasing significantly.

At this point, the algorithm is said to have **converged**.

### Convergence

Convergence occurs when further updates to $\theta_0$ and $\theta_1$ produce little or no reduction in the Cost Function.

### Workflow

Choose $\theta_0$ and $\theta_1$
↓
Calculate Cost Function
↓
Update $\theta_0$ and $\theta_1$
↓
Cost Decreases
↓
Repeat
↓
Convergence Reached
↓
Best-Fit Line Found

### In One Sentence

Repeat until convergence means continuously updating the model parameters until the Cost Function reaches its minimum value or stops improving.

# <span style="color:#2563eb;">Why Logistic Regression Uses an S-Curve Instead of a Straight Line</span>

## Problem with a Straight Line

A <b style="color:#dc2626;">straight line</b> can produce any value:

```text
-5, 2, 10, 100, ...
```

But probabilities must always be between:

```text
0 and 1
```

Example:

```text
Probability = -3 ❌

Probability = 7 ❌
```

These values are impossible.

---

## Solution: Sigmoid Function

The <b style="color:#16a34a;">Sigmoid Function</b> converts any value into a valid probability.

Output is always:

```text
0 ≤ P ≤ 1
```

---

## Why the S-Shape?

The <b style="color:#7c3aed;">S-Curve</b> naturally represents classification confidence.

```text
Very Negative z → Probability ≈ 0

z = 0 → Probability = 0.5

Very Positive z → Probability ≈ 1
```

This makes it easy to separate classes.

---

## Example

```text
Probability = 0.10 → Class 0

Probability = 0.50 → Boundary

Probability = 0.90 → Class 1
```

---

## Key Idea

<b style="color:#2563eb;">Straight Line</b> → Can produce any value

<b style="color:#7c3aed;">Sigmoid (S-Curve)</b> → Always produces values between 0 and 1

---

## One-Line Answer

<b style="color:#2563eb;">Logistic Regression</b> uses an <b style="color:#7c3aed;">S-Curve</b> because classification requires probabilities between <b>0 and 1</b>, while a straight line can produce invalid probability values.

</div>


To make a straight line into S-curved we use<mark> SIMBOID Function

$$
h_\theta(x) = g(\theta_0 + \theta_1 x_1)
$$
Where:
$h_\theta(x)$ = Predicted probability

$g$ = Sigmoid (Logistic) Function

$\theta_0$ = Bias (Intercept)

$\theta_1$ = Weight (Coefficient)

$x_1$ = Input Feature


<div align="center">

## Sigmoid Function

$$
g(z) = \frac{1}{1 + e^{-z}}
$$

</div>

### Where:

- $g(z)$ = Output of sigmoid function
- $e$ = Euler's number ($\approx 2.718$)
- $z$ = Input to the sigmoid function

For Logistic Regression:

$$
z = \theta_0 + \theta_1 x_1
$$

Substituting \(z\):

$$
h_\theta(x) = g(\theta_0 + \theta_1 x_1)
$$

$$
h_\theta(x) = \frac{1}{1 + e^{-(\theta_0 + \theta_1 x_1)}}
$$

</div>

# <span style="color:#2563eb;">Support Vector Machine (SVM)</span>

## <span style="color:#2563eb;">Definition</span>

<b style="color:#7c3aed;">Support Vector Machine (SVM)</b> is a <b style="color:#16a34a;">Supervised Machine Learning Algorithm</b> mainly used for <b style="color:#dc2626;">Classification</b>.

---

## <span style="color:#2563eb;">Main Idea</span>

SVM finds the <b style="color:#16a34a;">best hyperplane</b> that separates different classes with the <b style="color:#dc2626;">maximum margin</b>.

---

## <span style="color:#2563eb;">Key Terms</span>

### <span style="color:#7c3aed;">Hyperplane</span>

The <b style="color:#16a34a;">decision boundary</b> that separates classes.

### <span style="color:#7c3aed;">Support Vectors</span>

The data points <b style="color:#dc2626;">closest to the hyperplane</b>.

### <span style="color:#7c3aed;">Margin</span>

The distance between the <b style="color:#16a34a;">support vectors</b> and the <b style="color:#16a34a;">hyperplane</b>.

<div style="background:#eff6ff; border-left:4px solid #2563eb; padding:12px; border-radius:8px;">

<b>Goal: Maximize the Margin</b>

</div>

---

## <span style="color:#2563eb;">Types of SVM</span>

### <span style="color:#16a34a;">Linear SVM</span>

Used when data can be separated by a <b>straight line</b>.

### <span style="color:#16a34a;">Non-Linear SVM</span>

Uses the <b style="color:#7c3aed;">Kernel Trick</b> to handle complex data.

### Common Kernels

* Linear
* Polynomial
* RBF
* Sigmoid

---

## <span style="color:#2563eb;">Advantages</span>

* High Accuracy
* Works Well with High-Dimensional Data
* Effective for Small and Medium Datasets

---

## <span style="color:#2563eb;">Disadvantages</span>

* Slow on Large Datasets
* Sensitive to Feature Scaling
* Difficult to Interpret

---

## <span style="color:#2563eb;">Applications</span>

* Spam Detection
* Sentiment Analysis

---

## <span style="color:#2563eb;">One-Line Summary</span>

<b style="color:#7c3aed;">SVM</b> separates classes by finding the <b style="color:#16a34a;">optimal hyperplane</b> with the <b style="color:#dc2626;">maximum margin</b> between classes.


2d Line - Hyperplain

3d line - Hyper Plane
</div>


$$
w_1x_1 + w_2x_2 + \cdots + w_nx_n + b = 0
$$

### Where:

- \(w_1, w_2, \ldots, w_n\) = Weights (importance of each feature)
- \(x_1, x_2, \ldots, x_n\) = Input feature values
- \(b\) = Bias (intercept)
- \(n\) = Number of features

### Meaning

This equation represents a **Hyperplane** in SVM.

SVM tries to find the **best hyperplane** that separates different classes with the **maximum margin**.

### Example (2 Features)

$$
w_1x_1 + w_2x_2 + b = 0
$$

where:

- \(x_1\) = Height
- \(x_2\) = Weight
- \(w_1, w_2\) = Learned weights
- \(b\) = Bias

The resulting line (hyperplane) separates the classes.

````python
from sklearn.metrics import mean_absolute_error

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mae = mean_absolute_error(y_true, y_pred)

print("MAE:", mae)

# The model predictions differ from the actual values by about 6.25 units on average
# USE in Real World : Judge Model Quality
````

<b>2. <b><mark>MSE (Mean Squared Error)

````python
p='''
MSE Mean Squared Error
1- Mistakes square them
2- add
3- divide total'''
````

````python
# just square the mistakes finded by MAE so that machine can find patterens and errors easily 

# MSE measures the average squared prediction error and is used because it penalizes large errors much more heavily than MAE

from sklearn.metrics import mean_squared_error

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mse = mean_squared_error(y_true, y_pred)

print("MSE:", mse)

# Use in Real World: Punishes large mistakes more heavily
````

<b>3. <b><mark> RSME - Root Mean Square Error 

````python
from sklearn.metrics import mean_squared_error
import math

y_true = [80, 60, 90, 70]
y_pred = [75, 70, 85, 65]

mse = mean_squared_error(y_true, y_pred)

rmse = math.sqrt(mse)

print("RMSE:", rmse)
````

<h2 style="color:#2563eb;">🔹 Supervised Learning</h2>

Supervised Learning is a type of
<b style="color:#dc2626;">Machine Learning</b>
in which the model is trained using
<b style="color:#16a34a;">labeled data</b>.

The dataset contains both
<b style="color:#ea580c;">input features (X)</b>
and the
<b style="color:#9333ea;">target/output variable (y)</b>.

The model learns the relationship between inputs and outputs and then makes predictions on new data.

<b style="color:#16a34a;">Main Goal:</b>
Predict a known outcome.

<hr>

<h2 style="color:#7c3aed;">🔹 Unsupervised Learning</h2>

Unsupervised Learning is a type of
<b style="color:#dc2626;">Machine Learning</b>
in which the model is trained using
<b style="color:#16a34a;">unlabeled data</b>.

There is
<b style="color:#dc2626;">no target/output variable</b>.

The model tries to discover
<b style="color:#ea580c;">hidden patterns</b>,
<b style="color:#ea580c;">relationships</b>,

<b style="color:#ea580c;">groups</b>
within the data.

<h3 style="color:#7c3aed;">Examples</h3>

</ul>

<b style="color:#16a34a;">Main Goal:</b>
Find hidden patterns or groups in data.

<h2 style="color:#0f766e;">Key Differences</h2>

| Supervised Learning | Unsupervised Learning |
|---------------------|----------------------|
| Uses <b style="color:#16a34a;">labeled data</b> | Uses <b style="color:#16a34a;">unlabeled data</b> |
| <b style="color:#9333ea;">Target variable</b> is present | <b style="color:#dc2626;">Target variable</b> is absent |

Use <b style="color:#2563eb;">Supervised Learning</b> when:

### Examples
- Salary Prediction

## 🔹 Unsupervised Learning

Use <b style="color:#7c3aed;">Unsupervised Learning</b> when:

- You want to find <b style="color:#7c3aed;">patterns, relationships, or groups</b> in the data.

### Examples
- Customer Segmentation
- Market Basket Analysis
- Customer Grouping

---

## 📌 Quick Rule

❌ No → <b style="color:#7c3aed;">Unsupervised Learning</b>

# Variance

## 🔹 Definition

<b style="color:#2563eb;">Variance</b> measures how far data points are <b style="color:#16a34a;">spread from the mean</b>.

---

## 🔹 Formula

σ² = Σ(x - μ)² / N

Where:

- <b style="color:#2563eb;">σ²</b> = Variance
- <b style="color:#2563eb;">x</b> = Data Value
- <b style="color:#2563eb;">μ</b> = Mean
- <b style="color:#2563eb;">N</b> = Number of Values

---

## 🔹 Interpretation

- <b style="color:#16a34a;">Low Variance</b> → Data is close to the mean.
- <b style="color:#dc2626;">High Variance</b> → Data is spread out.

---

## 🔹 When to Use

- Measure Data Spread
- Statistics
- Data Analysis
- Machine Learning

---

## 📌 Key Idea

<b style="color:#7c3aed;">Variance</b> tells us how much the data varies from its <b style="color:#16a34a;">average value (mean)</b>.

<h3 style="color:#7c3aed;">The Reality of Machine Learning</h3>

When we build a machine learning model for the first time, it is usually <b style="color:#dc2626;">not the best version</b> of that model.

This is where <b style="color:#2563eb;">Model Tuning</b> becomes important.

<div style="background:#f8fafc; border-radius:12px; padding:15px; border-left:4px solid #2563eb;">

<b>Model Tuning</b> is the process of improving a model by finding the best values for its hyperparameters.

</div>

<br>

<h3 style="color:#7c3aed;">What Are Hyperparameters?</h3>

Hyperparameters are settings that control how a model learns.

Examples:

<ul>
<li><b style="color:#16a34a;">KNN</b> → Number of Neighbors (K)</li>
<li><b style="color:#16a34a;">Decision Tree</b> → Maximum Depth</li>
<li><b style="color:#16a34a;">Gradient Descent</b> → Learning Rate</li>
</ul>

These settings directly affect model performance.

<br>

<h3 style="color:#7c3aed;">Why Tune a Model?</h3>

Using random hyperparameter values may cause:

<table>
<tr>
<td>❌ <b>Underfitting</b></td>
<td>Model learns too little</td>
</tr>

<tr>
<td>❌ <b>Overfitting</b></td>
<td>Model memorizes training data</td>
</tr>

<tr>
<td>❌ <b>Poor Performance</b></td>
<td>Weak predictions on new data</td>
</tr>
</table>

<br>

<h3 style="color:#7c3aed;">Main Goal</h3>

<div style="background:#eef2ff; padding:15px; border-radius:12px;">

Find the <b style="color:#2563eb;">best combination of hyperparameters</b>

<br><br>

So the model performs well on <b style="color:#16a34a;">unseen data</b>, not just the training data.

</div>

<p style="font-family:'Trebuchet MS',sans-serif; font-size:18px; color:#222; margin:0;">

<span style="background:#DDD6FE; padding:2px 6px; border-radius:4px;">

The model is trained

</p>

</p>

<p style="font-family:'Segoe UI',sans-serif; font-size:18px; color:#222; margin-bottom:0;">
✔ More reliable than a single train-test split
</p>

</div>

</div>

### 5 Steps Of How K-Mean Model works

<div style="border:1px solid #ddd; border-radius:12px; padding:20px;">

# <span style="color:#2563eb;">K-Means Clustering: 5-Step Formula</span>

## <span style="color:#7c3aed;">Step 1: Choose K</span>

Tell the machine how many <b style="color:#16a34a;">clusters (K)</b> you want.

**Example:**

K = 3

---

## <span style="color:#7c3aed;">Step 2: Select Random Centroids</span>

Randomly place <b style="color:#16a34a;">K centroids</b> in the dataset.

**Example:**

K = 3 → Select 3 random centroids.

---

## <span style="color:#7c3aed;">Step 3: Assign Points</span>

Calculate the <b style="color:#16a34a;">Euclidean Distance</b> from each data point to every centroid.

Assign the point to the <b style="color:#16a34a;">nearest centroid</b>.

---

## <span style="color:#7c3aed;">Step 4: Update Centroids</span>

Find the <b style="color:#16a34a;">mean</b> of all points in each cluster.

The new mean becomes the new <b style="color:#16a34a;">centroid</b>.

---

## <span style="color:#7c3aed;">Step 5: Repeat</span>

Repeat <b style="color:#16a34a;">Steps 3 and 4</b> until centroids stop changing significantly.

At this point, the clusters are finalized.

---

## 📌 <span style="color:#2563eb;">Key Idea</span>

<b style="color:#2563eb;">Choose K</b>

⬇️

<b style="color:#7c3aed;">Random Centroids</b>

⬇️

<b style="color:#16a34a;">Assign Points</b>

⬇️

<b style="color:#ea580c;">Update Centroids</b>

⬇️

<b style="color:#dc2626;">Repeat Until Stable</b>

</div>


````python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Sample data
X = [
    [1, 2],[1, 3],[2, 2],[8, 8],[9, 8],[8, 9]]

# Create model

model = KMeans(n_clusters=2, random_state=42)

# Train model

model.fit(X)

# Cluster assigned to each point

print("Clusters:", model.labels_)

# Centroids

print("Centroids:")
print(model.cluster_centers_)

# Visualization

for point, cluster in zip(X, model.labels_):
    plt.scatter(point[0], point[1])

plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker="X",
    s=200
)

plt.show()
````

<b><mark>PCA (Principal Component Analysis)

<b><b>PCA (Principal Component Analysis) is a dimensionality reduction technique that transforms correlated features into a smaller set of uncorrelated principal components while preserving as much information (variance) as possible.


````python
#syntax

# from sklearn.decomposition import PCA

# pca = PCA(n_components=2)

# X_pca = pca.fit_transform(X)
````

````python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# why to use StandardScaler before PCA?

# "Because PCA is variance-based. Features with
# larger scales have larger variances and can dominate the principal components.
# Standardizing the data ensures that each feature contributes fairly

data = {
    'Age': [25, 30, 35, 40, 45, 50],
    'Income': [30000, 40000, 50000, 60000, 70000, 80000],
    'Spending': [70, 60, 50, 40, 30, 20],
    'Savings': [1000, 5000, 8000, 10000, 15000, 20000]
}

df = pd.DataFrame(data)

# first scaling

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

# pca

pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])

# now Variance explained

# How much information (variance) a Principal Component captures from the original data. it is % 

exp_var = pca.explained_variance_ratio_

print(exp_var)

# total variance explained

print(sum(exp_var)) # always 1
````

WHY PCA IS USED

- Reduces the number of features (dimensions)
- Removes redundant information from correlated features
- Preserves most of the important information (variance)
- Makes model training faster
- Reduces memory usage
- Helps reduce noise in the data
- Makes high-dimensional data easier to visualize
- Can improve the performance of some ML algorithms

WHEN TO USE PCA

- When the dataset has many features (10, 20, 100+)
- When features are highly correlated
- Before clustering algorithms like K-Means
- When training is slow due to many features
- When you want to visualize high-dimensional data in 2D or 3D
- When you want to reduce storage and computation requirements

WHEN NOT TO USE PCA

- When the dataset has only a few features
- When feature interpretability is important
- When original feature names must be preserved
- Often unnecessary for Random Forest, XGBoost, and Decision Trees
- When the dataset is already small and simple

````python
# plot of PCA

plt.figure(figsize=(8,6))
plt.scatter(pca_df['PCA1'], pca_df['PCA2'], color='blue',s = 80)
plt.title("PCA Projection (2D View)")
plt.xlabel('PCA1 Main Pattern')
plt.ylabel('PCA2 Minor Pattern')
plt.grid(True)
plt.show()
````
