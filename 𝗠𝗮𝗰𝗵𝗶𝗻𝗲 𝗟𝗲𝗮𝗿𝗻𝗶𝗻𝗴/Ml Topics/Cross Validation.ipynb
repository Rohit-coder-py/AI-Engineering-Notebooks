<h1 align="center" style="font-family:Georgia,serif; color:#2563EB;">
Cross Validation
</h1>

<div style="background:#EFF6FF; border-left:6px solid #2563EB; padding:18px; border-radius:10px;">

<p style="font-family:'Trebuchet MS',sans-serif; font-size:18px; color:#222; margin:0;">
Cross Validation is a technique used to check how well a machine learning model will perform on
<span style="background:#BFDBFE; padding:2px 6px; border-radius:4px;">
<b>unseen data</b>
</span>.
</p>

</div>

<br>

<div style="background:#F5F3FF; border-left:6px solid #7C3AED; padding:18px; border-radius:10px;">

<p style="font-family:'Palatino Linotype',serif; font-size:19px; color:#222; margin:0;">
Instead of training and testing the model only once, we train and test it multiple times on different parts of the dataset and then take the
<span style="color:#7C3AED;">
<b>average performance</b>
</span>.
</p>

</div>

</div>


<div style="background:white; border:3px solid #7C3AED; border-radius:18px; padding:30px;">

<h1 align="center" style="font-family:Georgia,serif; color:#7C3AED;">
K-Fold Cross Validation
</h1>

<div style="background:#F5F3FF; border-left:6px solid #7C3AED; padding:18px; border-radius:10px;">

<p style="font-family:'Trebuchet MS',sans-serif; font-size:18px; color:#222; margin:0;">
K-Fold Cross Validation is a model evaluation technique where the dataset is divided into
<span style="background:#DDD6FE; padding:2px 6px; border-radius:4px;">
<b>K equal parts (folds)</b>
</span>.
</p>

</div>

<br>

<div style="background:#EFF6FF; border-left:6px solid #2563EB; padding:18px; border-radius:10px;">

<p style="font-family:'Palatino Linotype',serif; font-size:19px; color:#222; margin:0;">
The model is trained
<span style="color:#2563EB;"><b>K times</b></span>.
In each iteration:
</p>

<br>

<p style="font-family:Verdana,sans-serif; font-size:18px; color:#222; margin:0;">
🔹 <b>1 fold</b> is used for testing.
</p>

<p style="font-family:Verdana,sans-serif; font-size:18px; color:#222;">
🔹 <b>K − 1 folds</b> are used for training.
</p>

</div>

<br>

<div style="background:#ECFDF5; border-left:6px solid #10B981; padding:18px; border-radius:10px;">

<p style="font-family:'Book Antiqua',serif; font-size:19px; color:#222; margin:0;">
The final performance is the
<span style="color:#059669;"><b>average of all K results</b></span>.
</p>

</div>

<br>

<h2 style="font-family:Georgia,serif; color:#DC2626;">
✅ Advantages
</h2>

<div style="background:#FEF2F2; border:2px solid #FCA5A5; border-radius:12px; padding:18px;">

<p style="font-family:'Segoe UI',sans-serif; font-size:18px; color:#222;">
✔ Better evaluation of model performance
</p>

<p style="font-family:'Segoe UI',sans-serif; font-size:18px; color:#222;">
✔ Reduces overfitting risk
</p>

<p style="font-family:'Segoe UI',sans-serif; font-size:18px; color:#222;">
✔ Uses all data for both training and testing
</p>

<p style="font-family:'Segoe UI',sans-serif; font-size:18px; color:#222; margin-bottom:0;">
✔ More reliable than a single train-test split
</p>

</div>

</div>


```python
# code implementation 

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

data = load_iris()
X = data.data
y = data.target

model = DecisionTreeClassifier()

scores = cross_val_score(model ,X ,y, cv=5)
```
