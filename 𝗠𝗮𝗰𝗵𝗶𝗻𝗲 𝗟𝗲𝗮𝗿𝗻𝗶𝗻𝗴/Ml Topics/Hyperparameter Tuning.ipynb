<h2 style="font-size:30px; color:#374151;">
    <strong>What is GridSearchCV?</strong>
</h2>

<p><strong>Definition</strong></p>

<p>
<strong>GridSearchCV</strong> automatically tries <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>every combination</strong></span> of the <strong>hyperparameters</strong> you provide, <strong>evaluates</strong> each one using <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>cross-validation</strong></span>, and <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>selects the best combination</strong></span>.
</p>

| Point      | Explanation                                 |
| ---------- | ------------------------------------------- |
| Purpose    | Find the best hyperparameters               |
| Method     | Tests every possible combination            |
| Evaluation | Uses Cross Validation                       |
| Output     | Best hyperparameters and best trained model |
| Advantage  | Removes manual trial-and-error              |

<h2 style="font-size:30px; color:#374151;">
    <strong>What does GridSearchCV.fit() do?</strong>
</h2>

<p><strong>Answer:</strong></p>

<p>
<strong>GridSearchCV.fit()</strong> <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>trains the model</strong></span> using <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>every possible hyperparameter combination</strong></span>, <strong>evaluates</strong> each combination with <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>cross-validation</strong></span>, <strong>compares</strong> their scores, and <span style="background-color:#FCE7F3; padding:2px 6px; border-radius:4px;"><strong>stores the best-performing model and hyperparameters</strong></span>.
</p>

| Parameter    | Meaning                                                        |
| ------------ | -------------------------------------------------------------- |
| `estimator`  | Model to tune                                                  |
| `param_grid` | Hyperparameters and values to test                             |
| `cv`         | Number of cross-validation folds                               |
| `scoring`    | Metric used to compare combinations                            |
| `fit()`      | Runs the complete grid search and selects the best combination |

    <strong>After Training GridSearchCV</strong>

Now we simply ask <strong>GridSearchCV</strong>:

➜ Which <strong>trained model</strong> was the best? <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>.best_estimator_</strong></span>
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>GridSearchCV Functions</strong>
</h2>

<p>
<strong>grid.best_params_</strong> = <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>It returns the best hyperparameter combination</strong></span> found during <strong>GridSearchCV</strong>.
</p>

<p>
<strong>best_params_</strong> returns the <strong>hyperparameter values</strong> that achieved the <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>best cross-validation performance</strong></span> during <strong>GridSearchCV</strong>.
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>GridSearchCV Functions</strong>
</h2>

<p>
<strong>grid.best_score_</strong> = <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>It returns the highest average cross-validation score</strong></span> achieved by the <strong>best hyperparameter combination</strong>.
</p>

<p>
⚠️ <strong>Important:</strong> This is <span style="background-color:#FEE2E2; padding:2px 6px; border-radius:4px;"><strong>NOT</strong></span> the <strong>test accuracy</strong>.
</p>

<p>
It is the <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>average CV score</strong></span>.
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>GridSearchCV Functions</strong>
</h2>

<p>
<strong>grid.best_estimator_</strong> = <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>It returns the fully trained model</strong></span> using the <strong>best hyperparameters</strong>.
</p>

| Attribute         | Returns                             |
| ----------------- | ----------------------------------- |
| `best_params_`    | Best hyperparameter values          |
| `best_score_`     | Best average cross-validation score |
| `best_estimator_` | Best trained model                  |

```python
#example code

from sklearn.neighbors import KNeighborsClassifier

import seaborn as sns
df = sns.load_dataset('iris')

df['species'].unique()

#using train test split

X = df.drop('species',axis=1)
y = df['species']

from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# using Kneighbours

model_knn = KNeighborsClassifier(n_neighbors=23)

model_knn.fit(X_train , y_train)

model_knn.score(X_test,y_test)


# Now lets use grid search CV

from sklearn.model_selection import GridSearchCV

params = {
    'n_neighbors': [3,5,7,9,11,13,15],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

classifier = GridSearchCV((model_knn), params,cv=5,return_train_score=False)


classifier.fit(X,y)

best_params = classifier.best_params_

best_model = classifier.best_estimator_

print(best_model.score(X_test, y_test))

```

<h2 style="font-size:30px; color:#374151;">
    <strong>GridSearchCV vs RandomizedSearchCV</strong>
</h2>

<p>
<strong>GridSearchCV</strong> <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>checks every possible combination</strong></span>.
</p>

<p>
<strong>RandomizedSearchCV</strong> <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>checks only a random selection of combinations</strong></span>.
</p>

| GridSearchCV                            | RandomizedSearchCV                                     |
| --------------------------------------- | ------------------------------------------------------ |
| Tests every combination                 | Tests random combinations                              |
| Slower                                  | Faster                                                 |
| Finds the global best (within the grid) | Finds a good solution quickly                          |
| No `n_iter`                             | Uses `n_iter` to control how many combinations to test |
| Better for small search spaces          | Better for large search spaces                         |


RandomizedSearchCV randomly samples a specified number of hyperparameter combinations, evaluates them using cross-validation, and returns the best combination among those sampled.

```python
from sklearn.model_selection import RandomizedSearchCV


#only n_iter will be added from gridsearchcv rest evrything is same 



# grid search cv checks all combination but randomized check only random combinations


```

```python
n_iter = Test only 20 randomly selected hyperparameter combinations 
```

```python
#exmaple code

from sklearn.model_selection import RandomizedSearchCV

# using RandomizedSearchCV

from sklearn.model_selection import RandomizedSearchCV

params = {
    'n_neighbors': [3,5,7,9,11,13,15],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

classifier = RandomizedSearchCV(model_knn,param_distributions=params,n_iter=10,cv=5,random_state=42,return_train_score=False)

classifier.fit(X, y)

best_params = classifier.best_params_

best_model = classifier.best_estimator_

print("Best Parameters:", best_params)
print("Test Accuracy:", best_model.score(X_test, y_test))
```
