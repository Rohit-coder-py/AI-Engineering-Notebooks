<h2 style="font-size:30px; color:#374151;">
    <strong>What is Train-Test Split?</strong>
</h2>

<p>
<strong>Train-Test Split</strong> is the process of <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>dividing a dataset into two parts</strong></span>:
</p>

<p>
➥ <strong>Training Set</strong> → Used to <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>train the machine learning model</strong></span>.
</p>

<p>
➥ <strong>Testing Set</strong> → Used to <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>evaluate how well the trained model performs on unseen data</strong></span>.
</p>

<b><mark>Our goal is not to measure memorization. Our goal is to measure generalization—how well the model performs on data it has never seen before.

| Term         | Meaning                           |
| ------------ | --------------------------------- |
| Training Set | Used to teach the model.          |
| Testing Set  | Used to evaluate the model.       |
| Seen Data    | Data used during training.        |
| Unseen Data  | Data never shown during training. |


<h2 style="font-size:30px; color:#374151;">
    <strong>Train-Test Split</strong>
</h2>

<p>
➜ Splits the <strong>dataset</strong> into <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>Training Data</strong></span> and <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>Testing Data</strong></span>.
</p>

<p>
➜ <strong>Training Data</strong> is used to <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>train the model</strong></span>.
</p>

<p>
➜ <strong>Testing Data</strong> is used to <span style="background-color:#FCE7F3; padding:2px 6px; border-radius:4px;"><strong>evaluate the model</strong></span>.
</p>

<p>
➜ Prevents <strong>overfitting</strong> and helps measure how well the <strong>model</strong> generalizes to <strong>unseen data</strong>.
</p>

<p>
➜ Common <strong>split ratios</strong> are <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>80:20</strong></span> and <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>70:30</strong></span>.
</p>

```python
#now lets start

# stratify :

# Used mainly for classification problems when the target classes are imbalanced
# Without stratify, the training and testing sets might end up with very different class proportions

# With:

# stratify=y

# both sets maintain approximately the same class distribution as the original dataset
```

"stratify=y ensures that the class distribution in the training and testing sets is approximately the same as in the original dataset. This is especially important for classification problems with imbalanced classes, because it leads to more reliable training and evaluation."

| Variable      | Used In                                                        | Purpose                                                                                                              |
| ------------- | -------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------- |
| **`X_train`** | Feature Scaling (`fit`), Model Training, Hyperparameter Tuning | Used to **teach** the model.                                                                                         |
| **`y_train`** | Model Training, Hyperparameter Tuning                          | Provides the **correct target values** for learning.                                                                 |
| **`X_test`**  | Feature Scaling (`transform`), Prediction                      | Used to **test** the trained model on unseen data.                                                                   |
| **`y_test`**  | Model Evaluation                                               | Used to compare with predictions (`y_pred`) and calculate metrics such as Accuracy, Precision, Recall, and F1-Score. |
