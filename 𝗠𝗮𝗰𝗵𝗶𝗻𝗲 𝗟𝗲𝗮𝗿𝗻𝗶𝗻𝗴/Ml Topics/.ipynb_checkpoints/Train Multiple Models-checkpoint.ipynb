<h2 style="font-size:30px; color:#374151;">
    <strong>Why do we train multiple models?</strong>
</h2>

<p><strong>Answer:</strong></p>

<p>
We train <strong>multiple machine learning models</strong> because different <strong>algorithms</strong> perform differently on different <strong>datasets</strong>. By comparing their performance using <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>evaluation metrics</strong></span>, we can choose the <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>model that generalizes best to unseen data</strong></span>.
</p>

```python
# Same syntax for all model ( no changes required ) 

# Same syntax for almost all Classification Models

# 1. Import
from sklearn.<module> import ModelName

# 2. Create object
model = ModelName()

# 3. Train the model
model.fit(X_train, y_train)

# 4. Make predictions
y_pred = model.predict(X_test)

# 5. Evaluate the model
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report)

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```
