## Load a model

```python
import joblib

model = joblib.load("model.pkl")

scaler = joblib.load("scaler.pkl")

encoder = joblib.load("encoder.pkl")


#load everything together
```

| Object                     |   Save?  | Why?                                   |
| -------------------------- | :------: | -------------------------------------- |
| Trained Model              |     ✅    | Needed for prediction                  |
| StandardScaler             |     ✅    | Scale new data exactly like training   |
| LabelEncoder               |     ✅    | Encode categorical values consistently |
| OneHotEncoder              |     ✅    | Same category mapping                  |
| Target Encoder (if used)   |     ✅    | Decode/encode target labels            |
| PCA (if used)              |     ✅    | Apply same dimensionality reduction    |
| Feature Selector (if used) |     ✅    | Keep the same selected features        |
| Hyperparameters            | Optional | Useful for reproducibility             |

| Object         | During Training   | During Prediction |
| -------------- | ----------------- | ----------------- |
| LabelEncoder   | `fit_transform()` | `transform()`     |
| OneHotEncoder  | `fit_transform()` | `transform()`     |
| StandardScaler | `fit_transform()` | `transform()`     |
| MinMaxScaler   | `fit_transform()` | `transform()`     |
| PCA            | `fit_transform()` | `transform()`     |
| **Model**      | `fit()`           | `predict()`       |
