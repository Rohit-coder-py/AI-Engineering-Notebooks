<h2 style="font-size:30px; color:#374151;">
    <strong>1. Label Encoding</strong>
</h2>

<p>
➥ Assigns a <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>unique integer</strong></span> to each <strong>category</strong>.

<h2 style="font-size:30px; color:#374151;">
    <strong>2. One-Hot Encoding</strong>
</h2>

<p>
➥ Creates a <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>separate binary (0/1) column</strong></span> for each <strong>category</strong>.
</p>

<p>
➥ Best for <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>Nominal data</strong></span> (where <strong>no order exists</strong>).
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>3. Ordinal Encoding</strong>
</h2>

<p>
➥ Assigns <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>user-defined numerical values</strong></span> to each <strong>category</strong> based on their <strong>order</strong>.
</p>

<p>
➥ Best for <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>Ordinal data</strong></span> (where <strong>a natural order exists</strong>).
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>3. Ordinal Encoding</strong>
</h2>

<p>
➥ Assigns <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>user-defined numerical values</strong></span> to each <strong>category</strong> based on their <strong>order</strong>.
</p>

<p>
➥ Best for <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>Ordinal data</strong></span> (where <strong>a natural order exists</strong>).
</p>

Ordinal Encoding is used when categorical values have a natural order or ranking. Instead of assigning random numbers, we assign numbers according to their actual order



Ex: School college job
small big massive



```python
# Only for ordinal data


import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Small", "Large", "Medium"]
})

print("Original DataFrame:")
print(df)

encoder = OrdinalEncoder(
    categories=[["Small", "Medium", "Large"]]
)

df["Size_Encoded"] = encoder.fit_transform(df[["Size"]])

print("\nAfter Ordinal Encoding:")
df
```

| Size   | Size_Encoded |
| ------ | -----------: |
| Small  |          0.0 |
| Medium |          1.0 |
| Large  |          2.0 |
| Small  |          0.0 |
| Large  |          2.0 |
| Medium |          1.0 |


| Feature                   | Label Encoder                       | Ordinal Encoder                 |
| ------------------------- | ----------------------------------- | ------------------------------- |
| Purpose                   | Encodes the **target variable (y)** | Encodes **input features (X)**  |
| Part of Dataset           | Usually **y**                       | Usually **X**                   |
| Best For                  | Class labels                        | Ordinal categorical features    |
| Supports Multiple Columns | ❌ No (one column at a time)         | ✅ Yes                           |
| Available In              | `LabelEncoder()`                    | `OrdinalEncoder()`              |
| User Defines Order?       | ❌ No                                | ✅ Yes (`categories=` parameter) |
