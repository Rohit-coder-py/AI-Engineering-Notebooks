<h2>
<font color="DeepSkyBlue">Regression Metrics</font>
</h2>

<p>
Learn to evaluate regression models using
<mark style="background-color:#FFF59D;">MAE</mark>,
<mark style="background-color:#FFE082;">MSE</mark>, and
<mark style="background-color:#FFCCBC;">RMSE</mark>.
</p>

<ul>

<li><b>MAE</b> → Mean Absolute Error</li>

<li><b>MSE</b> → Mean Squared Error</li>

<li><b>RMSE</b> → Root Mean Squared Error</li>

</ul>

<hr>

<h2>
<font color="YellowGreen">Confusion Matrix</font>
</h2>

<p>
Understand how to use a
<mark style="background-color:#DCEDC8;">Confusion Matrix</mark>
to analyze
<mark style="background-color:#C8E6C9;">model performance</mark>.
</p>

<p>
It helps visualize:

</p>

<ul>

<li>True Positives (TP)</li>

<li>True Negatives (TN)</li>

<li>False Positives (FP)</li>

<li>False Negatives (FN)</li>

</ul>

<hr>

<h2>
<font color="MediumSeaGreen">Classification Metrics</font>
</h2>

<p>
Learn to evaluate classification models using
<mark style="background-color:#C8E6C9;">Accuracy</mark>,
<mark style="background-color:#B3E5FC;">Precision</mark>,
<mark style="background-color:#FFE082;">Recall</mark>, and
<mark style="background-color:#E1BEE7;">F1-Score</mark>.
</p>

<ul>

<li><b>Accuracy</b> → Overall correctness of predictions.</li>

<li><b>Precision</b> → How many predicted positives are actually positive.</li>

<li><b>Recall</b> → How many actual positives were correctly identified.</li>

<li><b>F1-Score</b> → Balance between Precision and Recall.</li>

</ul>

<li><b>Accuracy</b> → Overall prediction correctness.</li>

<li><b>Precision</b> → Quality of positive predictions.</li>

<li><b>Recall</b> → Ability to find positive cases.</li>

<li><b>F1-Score</b> → Precision-Recall balance.</li>

</ul>
