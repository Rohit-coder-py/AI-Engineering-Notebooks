Feature Scaling is the process of bringing numerical features to a similar scale so that no feature dominates another because of its large values.

Only numerical features need scaling

| Topic                          | Notes                                                                 |
| ------------------------------ | --------------------------------------------------------------------- |
| Feature Scaling                | Brings numerical features to a similar scale                          |
| Why?                           | Prevents large-value features from dominating                         |
| Scale Which Columns?           | Numerical features only                                               |
| Common Scalers                 | StandardScaler, MinMaxScaler                                          |
| Most Used                      | StandardScaler                                                        |
| StandardScaler                 | Mean = 0, Std = 1                                                     |
| MinMaxScaler                   | Values between 0 and 1                                                |
| Models That Need Scaling       | Logistic Regression, KNN, SVM, Neural Networks, K-Means, PCA          |
| Models That Don't Need Scaling | Decision Tree, Random Forest, XGBoost, LightGBM, CatBoost             |
| Best Practice                  | `fit` on `X_train` only, then `transform` both `X_train` and `X_test` |


Feature Scaling is used to bring all numerical features to a similar range so that no feature dominates the model because of its larger values.

| Feature              | StandardScaler   | MinMaxScaler      |
| -------------------- | ---------------- | ----------------- |
| Output Range         | No fixed range   | 0 to 1            |
| Mean                 | 0                | Not necessarily 0 |
| Standard Deviation   | 1                | Not fixed         |
| Most Used In         | Machine Learning | Deep Learning     |
| Affected by Outliers | Yes              | Yes               |


StandardScaler is used for most machine learning algorithms because it standardizes features to have a mean of 0 and a standard deviation of 1.
MinMaxScaler is used when features need to be scaled to a fixed range (usually 0–1), making it especially common in deep learning and image processing.
