## <span style="color:#2C3E50; font-weight:700;">6. Removing Stopwords :</span>

````python
import nltk

nltk.download('stopwords')
````

````python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return ' '.join(words)
````

````python
#apply to our dataset

df['review'] = df['review'].apply(remove_stopwords)
````

````python
#verify

text = "I am watching a very good movie"

print(remove_stopwords(text))
````

Important for sentiment analysis: don't automatically remove every stopword. Words such as "not", "no", and "never" can be extremely important for sentiment.

## <span style="color:#2C3E50; font-weight:700;">6. Tokenization :</span>

# <span style="color:#2c3e50;">Tokenization</span>

**Tokenization** = breaking text into smaller units called **tokens**.

### Example

> **"I love this movie"**

↓  

`["I", "love", "this", "movie"]`

### <span style="color:#2980b9;">Why Tokenization?</span>

- Makes text easier for NLP models to process.
- Converts text into individual units.
- Tokens can later be converted into **numbers/vectors** for ML/DL models.

### <span style="color:#27ae60;">Flow</span>

**Raw Text** → **Tokenization** → **Tokens** → **Numerical Representation** → **Model**

# <span style="color:#2c3e50;">Problems During Tokenization</span>

Tokenization is not always as simple as splitting text by spaces.  
**Punctuation and special cases can create problems.**

### <span style="color:#2980b9;">1. Prefix</span>

**Prefix** = Character(s) at the **beginning** of a word/token.

Examples:

`$100` → `$` is a prefix  
`“hello` → `“` is a prefix

---

### <span style="color:#8e44ad;">2. Suffix</span>

**Suffix** = Character(s) at the **end** of a word/token.

Examples:

`100km` → `km` is a suffix  
`hello!` → `!` is a suffix

---

### <span style="color:#16a085;">3. Infix</span>

**Infix** = Character(s) **in between** a word/token.

Examples:

`--`  
`/...`

These characters may occur inside text and affect how a tokenizer splits it.

---

### <span style="color:#c0392b;">4. Exceptions</span>

**Exception** = A special-case rule used to:

- Split a string into several tokens.
- Prevent a token from being split when punctuation rules are applied.

Examples:

`let's` → may need special handling

`U.S.` → may need to remain as a single token.

### <span style="color:#27ae60;">Key Point</span>

> **Prefix → Beginning**  
> **Suffix → End**  
> **Infix → In between**  
> **Exception → Special-case tokenization rule**

## <span style="color:#2C3E50; font-weight:700;">7. Stemming :</span>

<b>In grammar, <mark><u>inflection</u></mark> is the modification of a word to
express different grammatical categories such as tense,
case, voice, aspect, person, number, gender, and mood

# <span style="color:#2c3e50;">What is Stemming?</span>

**Stemming** is a text-preprocessing technique used to reduce a word to its **root/base form** by removing prefixes or suffixes.

### <span style="color:#2980b9;">Examples</span>

| Word      | Stem   |
|-----------|--------|
| playing   | play   |
| played    | play   |
| plays     | play   |
| studies   | studi  |
| studying  | studi  |
| happily   | happi  |

> **Key Point:** Stemming removes prefixes or suffixes to obtain a **root/base form**, which may not always be a proper English word.

>Stemming is the process of reducing inflection in words
to their root forms such as mapping a group of words to
the same stem even if the stem itself is not a valid word in
the Language."

````python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies"]

for word in words:
    print(word, "→", stemmer.stem(word))
````

# <span style="color:#2c3e50;">Stemming vs Lemmatization</span>

Both **stemming** and **lemmatization** are NLP techniques used to reduce different forms of a word to a common base form.

---

## <span style="color:#2980b9;">1. Stemming</span>

**Stemming** removes prefixes/suffixes from a word using simple rules.

It does **not necessarily produce a valid English word**.

### Examples

| Word | Stem |
|---|---|
| playing | play |
| played | play |
| plays | play |
| studies | studi |
| happily | happili |
| movie | movi |

### <span style="color:#27ae60;">Advantages</span>

- Fast
- Simple
- Useful for large amounts of text
- Reduces vocabulary size

### <span style="color:#c0392b;">Disadvantage</span>

The resulting stem may not be a real word.

```text
studies → studi
movie   → movi
```

---

## <span style="color:#8e44ad;">2. Lemmatization</span>

**Lemmatization** reduces a word to its **meaningful dictionary/base form**, called a **lemma**.

It uses linguistic information rather than simply chopping characters.

### Examples

| Word | Lemma |
|---|---|
| playing | play |
| played | play |
| studies | study |
| mice | mouse |

### <span style="color:#27ae60;">Advantages</span>

- More linguistically accurate
- Usually produces a valid word
- Better when the actual meaning/base form matters

### <span style="color:#c0392b;">Disadvantage</span>

- Slower than stemming
- More computationally involved

---

# <span style="color:#2c3e50;">Stemming vs Lemmatization</span>

| Feature | Stemming | Lemmatization |
|---|---|---|
| Method | Chops word endings | Linguistic analysis |
| Output | Stem | Lemma |
| Accuracy | Lower | Higher |
| Speed | Faster | Slower |
| Valid English word? | Not guaranteed | Usually |
| Example | `studies → studi` | `studies → study` |
| Example | `mice → mice` | `mice → mouse` |

---

# <span style="color:#2980b9;">When to Use Stemming?</span>

Use **stemming** when:

- Speed is important
- You are processing a large amount of text
- Exact linguistic correctness is not very important
- You mainly want related word forms to match
- Working with search/information-retrieval tasks

### Example

```text
connect
connected
connecting
```

Stemming can reduce these to similar forms so they can be treated as related.

---

# <span style="color:#8e44ad;">When to Use Lemmatization?</span>

Use **lemmatization** when:

- Linguistic correctness is important
- The actual base word matters
- You are doing linguistic analysis
- You need cleaner/meaningful words
- Tasks such as text classification or sentiment analysis benefit from meaningful normalization

```text
studies → study
mice → mouse
```

---

# <span style="color:#2c3e50;">Easy Way to Remember</span>

```text
STEMMING
    ↓
Chop the word
    ↓
Fast but rough
    ↓
studies → studi


LEMMATIZATION
    ↓
Find the meaningful base word
    ↓
Slower but accurate
    ↓
studies → study
```

> **Stemming = Fast + Rough**
>
> **Lemmatization = Slower + Linguistically Accurate**

---

## <span style="color:#16a085;">Important: Modern NLP</span>

For modern **Transformer models** such as BERT and GPT, we generally do **not manually apply stemming or lemmatization**.

These models use **subword tokenization** instead.

```text
Traditional NLP:
Text → Cleaning → Tokenization → Stemming/Lemmatization → Numerical Representation → Model

Modern Transformer NLP:
Text → Tokenization → Subword Tokens → Model
```

## <span style="color:#2C3E50; font-weight:700;">8.Lemmatization  :</span>

<b>Lemmatization, unlike Stemming, reduces the inflected
words<mark> properly ensuring that the root word belongs to
the language.</mark>. In Lemmatization root word is called
Lemma. A lemma (plural lemmas or lemmata) is the
canonical form, dictionary form, or citation form of a set
of words.

````python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
````
