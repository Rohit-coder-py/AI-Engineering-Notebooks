# Dependency Labels in NLP (spaCy)

## 1. What are Dependency Labels?

Dependency labels describe the **grammatical relationship between two words** in a sentence.

In dependency parsing, every word (except the ROOT) is connected to another word through a dependency relationship.

For example:

Sentence: "I am learning NLP."

| Word | Dependency Label | Meaning |
|---|---|---|
| I | nsubj | Nominal subject |
| am | aux | Auxiliary verb |
| learning | ROOT | Main word of the sentence |
| NLP | dobj | Direct object |

Here:
- `learning` is the ROOT.
- `I` is the subject of `learning`.
- `am` is an auxiliary verb.
- `NLP` is the object of `learning`.

---

## 2. How many dependency labels are there?

There is **no single universal number** of dependency labels.

The number depends on:
- The language being processed.
- The annotation scheme used.
- The NLP library and model.

spaCy's English model (`en_core_web_sm`) uses dependency labels based on the English linguistic annotation scheme.

Universal Dependencies (UD), another widely used annotation framework, has a different set of dependency relations.

---

## 3. Common Dependency Labels in spaCy

| Label | Meaning | Example |
|---|---|---|
| ROOT | Main word of the sentence | **eat** in "I eat rice." |
| nsubj | Nominal subject | **I** in "I eat rice." |
| nsubjpass | Passive nominal subject | **Rice** in "Rice was eaten." |
| dobj | Direct object | **rice** in "I eat rice." |
| pobj | Object of a preposition | **school** in "at school" |
| aux | Auxiliary verb | **will** in "I will go." |
| auxpass | Passive auxiliary | **was** in "It was eaten." |
| acomp | Adjectival complement | **happy** in "She is happy." |
| attr | Attribute | **teacher** in "She is a teacher." |
| advmod | Adverbial modifier | **quickly** in "Run quickly." |
| amod | Adjectival modifier | **beautiful** in "beautiful flower" |
| compound | Compound modifier | **POS** in "POS Tagging" |
| prep | Preposition | **in** in "in the room" |
| det | Determiner | **the** in "the book" |
| conj | Conjunct | **tea** in "tea and coffee" |
| cc | Coordinating conjunction | **and** in "tea and coffee" |
| punct | Punctuation | **.** in "Hello." |
| xcomp | Open clausal complement | **study** in "I want to study." |
| ccomp | Clausal complement | **left** in "I know she left." |
| advcl | Adverbial clause modifier | **rained** in "I stayed because it rained." |
| relcl | Relative clause modifier | **bought** in "The book that I bought" |
| mark | Marker | **because** in "because it rained" |
| poss | Possessive modifier | **John's** in "John's book" |
| nummod | Numeric modifier | **three** in "three books" |
| appos | Appositional modifier | **John** in "My friend John" |
| expl | Expletive | **There** in "There is a problem." |
| neg | Negation modifier | **not** in "I do not agree." |
| dep | Unspecified dependency | Used when a more specific relation is unavailable |

---

## 4. Important distinction: POS Tags vs Dependency Labels

These two concepts are different.

| POS Tagging | Dependency Parsing |
|---|---|
| Identifies the grammatical category of a word. | Identifies the grammatical relationship between words. |
| Examples: NOUN, VERB, ADJ, ADV | Examples: nsubj, dobj, aux, amod |
| Answers: "What type of word is this?" | Answers: "How is this word related to another word?" |

### Example

Sentence: "Harsh will learn NLP."

| Word | POS Tag | Dependency Label |
|---|---|---|
| Harsh | PROPN | nsubj |
| will | AUX | aux |
| learn | VERB | ROOT |
| NLP | PROPN | dobj |

<h3 style="color:#117864;">Step-by-Step Process</h3>

<div style="
background:#F8F9F9;
padding:16px 20px;
border:1px solid #D5DBDB;
border-radius:8px;
line-height:2;
">

<strong style="color:#2874A6;">1. Input Sentence</strong>  
The cat eats fish.

<br>

<strong style="color:#2874A6;">2. Tokenization</strong>  
The sentence is split into individual tokens:

`["The", "cat", "eats", "fish", "."]`

<br>

<strong style="color:#2874A6;">3. Contextual and Grammatical Analysis</strong>  
The tagger examines each word and its context to determine its grammatical role.

<br>

<strong style="color:#2874A6;">4. Assign POS Tags</strong>  
Each token receives an appropriate part-of-speech label.
</div>

<h3 style="color:#B9770E;">Final Tagged Output</h3>

| Token | POS Tag | Meaning |
|---|---|---|
| The | Determiner (DET) | Introduces the noun |
| cat | Noun (NOUN) | Names an animal |
| eats | Verb (VERB) | Describes an action |
| fish | Noun (NOUN) | Names something |
| . | Punctuation (PUNCT) | Marks the sentence ending |
