<strong>Custom Features</strong> are <strong>hand-crafted, domain-knowledge based features</strong> that we create ourselves using our understanding of the problem — instead of relying only on OHE / BoW / N-grams / TF-IDF (which are all purely frequency/statistics based).

</div>

<blockquote>
<strong>In short:</strong> We look at the text and the business problem, and manually engineer extra numeric columns that help the model — on top of the vectorized text.
</blockquote>

## <span style="color:#8E44AD; font-weight:700;">Why do we need Custom Features?</span>

<ul>
<li style="margin-bottom:8px;">OHE, BoW, N-grams, TF-IDF only capture <strong>word frequency / presence</strong> — they miss <strong>meta information</strong> about the text.</li>
<li style="margin-bottom:8px;">Domain knowledge often tells us something is important that pure word-counts can't capture — e.g. <strong>length of review</strong>, <strong>number of exclamation marks</strong>, <strong>positive/negative word count</strong>, etc.</li>
<li style="margin-bottom:8px;">These extra features, when combined with TF-IDF/BoW vectors, often <strong>boost model performance</strong>.</li>
</ul>

## <span style="color:#2C3E50; font-weight:800; border-bottom:2px solid #3498DB; padding-bottom:4px;">Common Custom Features (examples)</span>

| Feature | Meaning |
|---|---|
| `word_count` | Total number of words in the text |
| `char_count` | Total number of characters |
| `avg_word_length` | Average length of words |
| `stopword_count` | Number of stopwords (is, the, and, ...) |
| `unique_word_count` | Number of distinct words |
| `upper_case_word_count` | Number of ALL-CAPS words (often shows shouting/emphasis) |
| `numeric_count` | Count of numeric tokens |
| `special_char_count` | Count of special characters (!, ?, @, #, ...) |
| `sentence_count` | Number of sentences |

<blockquote>These are just examples — the actual features depend on <strong>your specific problem</strong> (e.g. for tweets: hashtag_count, mention_count; for reviews: exclamation_count, positive/negative word count).</blockquote>

````python
# code implementation of Custom Features

import pandas as pd
import string

documents = [
    "I enjoy learning Python",
    "Python is EASY and powerful!!",
    "I enjoy machine learning",
    "Machine learning is interesting, powerful and FUN 123"
]

df = pd.DataFrame({"text": documents})

# 1. word count
df["word_count"] = df["text"].apply(lambda x: len(x.split()))

# 2. character count
df["char_count"] = df["text"].apply(lambda x: len(x))

# 3. average word length
df["avg_word_length"] = df["char_count"] / df["word_count"]

# 4. unique word count
df["unique_word_count"] = df["text"].apply(lambda x: len(set(x.lower().split())))

# 5. upper case word count
df["upper_case_word_count"] = df["text"].apply(lambda x: len([w for w in x.split() if w.isupper()]))

# 6. numeric token count
df["numeric_count"] = df["text"].apply(lambda x: len([w for w in x.split() if w.isdigit()]))

# 7. special character count
df["special_char_count"] = df["text"].apply(lambda x: len([ch for ch in x if ch in string.punctuation]))

df
````

## <span style="color:#2C3E50; font-weight:800; border-bottom:2px solid #3498DB; padding-bottom:4px;">Custom Features — Pros & Cons</span>

### <span style="color:#27AE60; font-weight:800;">Pros</span>

1. <strong style="color:#2980B9;">Domain Knowledge Boost</strong> — captures signals that word-frequency methods miss.
2. <strong style="color:#2980B9;">Improves Model Performance</strong> — used alongside BoW/TF-IDF vectors as extra columns.

### <span style="color:#C0392B; font-weight:800;">Cons</span>

1. <strong style="color:#E67E22;">Manual Effort</strong> — needs human thought/experimentation per problem, not automatic.
2. <strong style="color:#8E44AD;">Not Generalizable</strong> — features good for one problem (e.g. tweets) may be useless for another (e.g. legal documents).
