<strong style="color:#2980B9;">Natural Language Processing (NLP)</strong> is a subfield of 
<strong style="color:#8E44AD;">linguistics, computer science, and artificial intelligence</strong> 
concerned with the interactions between computers and human language.

<br><br>

In particular, NLP focuses on how to program computers to 
<strong style="color:#27AE60;">process and analyze large amounts of natural language data.</strong>

</div>

## <span style="color:#2C3E50; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">What is Natural Language?</span>

<br>

<div style="
background-color:#EBF5FB;
border-left:5px solid #3498DB;
padding:15px 18px;
border-radius:5px;
">

<strong style="color:#2980B9;">Natural Language</strong> is the language that humans naturally develop and use to <strong>communicate with each other</strong> through everyday use and repetition.

</div>

### <span style="color:#8E44AD; font-weight:700;">Examples</span>

- <strong>English</strong>
- <strong>Hindi</strong>
- <strong>Tamil</strong>
- Other human languages

<br>

> <strong style="color:#27AE60;">Natural Language → Human Communication</strong>

Natural languages can take different forms, such as <strong>speech</strong> or <strong>signing</strong>.

They are different from <strong style="color:#C0392B;">formal or constructed languages</strong> such as programming languages used to communicate instructions to computers.

<br>

<div align="center">

<strong style="color:#2980B9;">In short:</strong>  
Natural Language = <strong>Language naturally used by humans to communicate.</strong>

</div>

## <span style="color:#2C3E50; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">Uses of NLP</span>

<br>

<div style="
background-color:#EBF5FB;
border-left:5px solid #3498DB;
padding:14px 18px;
border-radius:5px;
">

<strong style="color:#2980B9;">NLP</strong> is used to help computers <strong>understand, process, analyze, and generate human language.</strong>

</div>

### <span style="color:#8E44AD; font-weight:700;">Common Uses</span>

- <strong style="color:#2980B9;">Chatbots & Virtual Assistants</strong> → Understand user queries and respond naturally.
- <strong style="color:#27AE60;">Sentiment Analysis</strong> → Identify whether text is positive, negative, or neutral.
- <strong style="color:#D35400;">Machine Translation</strong> → Translate text from one language to another.
- <strong style="color:#8E44AD;">Text Classification</strong> → Categorize text into predefined classes.
- <strong style="color:#C0392B;">Speech Recognition</strong> → Convert human speech into text.
- <strong style="color:#16A085;">Text Summarization</strong> → Generate shorter summaries from large text.
- <strong style="color:#2C3E50;">Search Engines</strong> → Understand queries and retrieve relevant information.

<br>

> <strong style="color:#2980B9;">In short:</strong> NLP allows computers to work with <strong>human language</strong> in useful and meaningful ways.

Common Natural Language Processing (NLP) Tasks

<strong style="color:#2471A3;">NLP Tasks</strong> are different operations performed on human language to extract information, understand text, or generate language.

</div>

### <span style="color:#2980B9;">1. Text or Document Classification</span>

Categorizing text into predefined groups.

> **Example:** Spam / Not Spam, Payment Issue / Order Issue

---

### <span style="color:#27AE60;">2. Sentiment Analysis</span>

Determining the emotional tone (positive or negative) of text.

> **Example:** `"This product is amazing"` → Positive

---

### <span style="color:#8E44AD;">3. Information Retrieval</span>

Finding and retrieving relevant information from a large collection of text.

---

### <span style="color:#D35400;">4. Parts of Speech Tagging</span>

Assigning grammatical labels to words.

> **Example:** Noun, Verb, Adjective, Adverb

---

### <span style="color:#16A085;">5. Language Detection and Machine Translation</span>

Identifying languages and translating between them.

> **Example:** Hindi → English

---

### <span style="color:#C0392B;">6. Conversational Agents</span>

Building chatbots or virtual assistants.

---

### <span style="color:#2C3E50;">7. Knowledge Graphs and Question Answering</span>

Linking data to answer specific user queries.

---

### <span style="color:#3498DB;">8. Text Summarization</span>

Condensing long articles into shorter summaries.

---

### <span style="color:#8E44AD;">9. Topic Modeling</span>

Discovering themes within a large collection of text.

---

### <span style="color:#27AE60;">10. Text Generation</span>

Predicting upcoming words in a sequence.

---

### <span style="color:#D35400;">11. Spell Checking and Grammar Correction</span>

Identifying and fixing errors in text.

---

### <span style="color:#16A085;">12. Text Parsing</span>

Breaking down sentences to understand their grammatical structure.

---

### <span style="color:#C0392B;">13. Speech to Text</span>

Converting spoken audio into written text.

---

<h2 style="color:#2C3E50; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:6px;">
Approaches to NLP
</h2>

### <span style="color:#8E44AD;">1. Heuristic Methods</span>

- Uses **manually written rules and logic** to process language.
- Based on predefined patterns and conditions.

---

### <span style="color:#2980B9;">2. Machine Learning-Based Methods</span>

- Uses **Machine Learning algorithms** to learn patterns from text data.
- Text is converted into numerical features using techniques like **Bag of Words** and **TF-IDF**.

---

### <span style="color:#27AE60;">3. Deep Learning-Based Methods</span>

- Uses **Neural Networks** to learn complex patterns from text.
- Common architectures include:
  - **RNN**
  - **LSTM**
  - **GRU**
  - **Transformers**

```text
Heuristic Methods
        ↓
Machine Learning-Based Methods
        ↓
Deep Learning-Based Methods
```

1. Heuristic Approaches
</h2>

###  What are heuristic approaches?

Heuristic approaches use **rules, patterns, and predefined knowledge** to solve NLP problems.

### Examples

- **Regular Expressions**
- **WordNet**
- **Open Mind Common Sense**

### Advantages

- Simple to implement
- Easy to understand
- No training data required

### Current Scenario

Heuristic approaches are still useful for **simple and rule-based NLP tasks**, but modern NLP mainly uses **Machine Learning and Deep Learning** approaches.
