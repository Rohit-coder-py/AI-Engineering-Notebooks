# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">NLP Pipeline — Overview</span>

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<strong>Core Idea:</strong> An <b><u>NLP Pipeline</u></b> is a sequence of steps followed to convert raw data into a working NLP-based software/product.

</div>

---

## 1. What is an NLP Pipeline?

An **NLP Pipeline** is a series of steps required to build an end-to-end NLP software system.

### Main 5 Steps

```text
Data Acquisition
      ↓
Text Preparation
      ↓
Feature Engineering
      ↓
Modeling
      ↓
Deployment
```

The pipeline gives us an **end-to-end thinking framework** for solving real-world NLP problems.

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

⚠️ <b>Important:</b> This 5-step pipeline is <u>not universal</u>. Different NLP problems can require different pipelines and different preprocessing/feature-engineering steps.

</div>

Also, real-world pipelines are often **non-linear**. After deployment, you may return to feature engineering, modeling, or earlier stages to improve the product.

# <span style="background:linear-gradient(135deg,#3498DB,#8E44AD); color:white; padding:10px 20px; border-radius:8px; font-family:sans-serif; font-weight:800; display:inline-block; box-shadow:0 4px 10px rgba(0,0,0,0.2);">Data Acquisition :</span>

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">2. Data Acquisition 📥</span>

Data Acquisition is the **first and one of the most important steps**.

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

Without data, building an NLP/ML system is generally <b>impossible</b>.

</div>

Suppose we are building a **Sentiment Analysis System** for customer reviews.

There can be three situations:

### Case 1 — Data is Available Internally

The required data already exists inside the company.

Possible situations:

* Data is already available as a CSV/file.
* Data exists inside the company's database.
* Data exists, but the amount of data is insufficient.

### Case 2 — Data Exists Externally

The company does not have the required data, but it exists somewhere else.

Examples:

* Public datasets
* Websites
* APIs
* PDFs
* Images
* Audio

### Case 3 — Data Doesn't Exist

Nobody has the required dataset because the problem/use-case is new.

In this situation, we need to **generate/collect data ourselves**.

| Case | Situation | Typical Action |
|---|---|---|
| 1 | Data available internally | Use directly / request from data team |
| 2 | Data exists externally | Public datasets, scraping, APIs, PDFs, images, audio |
| 3 | Data doesn't exist | Collect / label data manually |


---

## 2.1 Data Already Available Internally

### Situation A — Data is already on your table

Example:

```text
reviews.csv
```

You can directly move to the next stage.

---

### Situation B — Data exists in the company's database

In a large organization, a Data Engineer/Data Engineering team may extract the required data from the running database and provide it to you.

<div style="background-color:#f8f9fa; border-left:5px solid #aeb6bf; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

The NLP engineer's responsibility may mainly involve <b>communicating the required data specification</b> to the data team.

</div>

---

### Situation C — Data is insufficient

Suppose:

```text
Required data = 50,000 reviews
Available data = 5,000 reviews
```


One technique discussed is:

## Data Augmentation

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Data augmentation</b> means creating additional training examples by modifying existing data while trying to preserve its meaning.

</div>

### Examples

#### 1. Synonym Replacement

```text
Original:
This movie is good.

Modified:
This movie is excellent.
```

The wording changes while the overall sentiment can remain similar.

#### 2. Bi-gram / Word Rearrangement

Words can be rearranged or modified to generate additional text examples.

#### 3. Back Translation

```text
English
  ↓
Another Language
  ↓
English
```

The translated-back sentence may have different wording while approximately preserving the original meaning.

#### 4. Adding Noise

Small modifications can be introduced to generate additional variations.

<div style="background-color:#eafaf1; border-left:5px solid #52be80; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

🎯 <b>Key Takeaway:</b> Data augmentation techniques (synonym replacement, rearrangement, back-translation, noise injection) help expand a small dataset without collecting new raw data.

</div>

---

When the required data exists outside the organization, several approaches can be used.

## Common Sources

### 1. Public Datasets

Data can be collected from:

* Public dataset platforms

* Open repositories

⚠️ <b>Limitation:</b> A public dataset may not perfectly match your specific business/domain.

### 2. Web Scraping

Data can be extracted from websites.

Example:

```text
Website
   ↓
Web Scraping
   ↓
Customer Reviews
   ↓
NLP Dataset
```

<div style="background-color:#f5eef8; border-left:5px solid #af7ac5; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

---

### 3. APIs

Instead of scraping a website, you can obtain data through an API.

```text
Application
     ↓
API Request
     ↓
API Response
     ↓
Data
```

<div style="background-color:#f5eef8; border-left:5px solid #af7ac5; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

### 4. PDF Data

Sometimes the required text exists inside PDF documents.

You need appropriate tools/libraries to extract the text.

### 5. Image → Text

If the text exists inside an image, use **OCR (Optical Character Recognition)**.

```text
Image
  ↓
OCR
  ↓
Text
  ↓
NLP Pipeline
```

### 6. Audio → Text

If customer reviews are recorded as audio:

```text
Audio
  ↓
Speech-to-Text
  ↓
Text
  ↓
NLP Pipeline
```

This converts speech into text so that NLP techniques can be applied.

<div style="background-color:#f8f9fa; border-left:5px solid #aeb6bf; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">4. When Data Doesn't Exist</span>

For example:

```text
Product
   ↓
Ask selected customers for reviews
   ↓
Collect reviews
   ↓
Label reviews
   ↓
Create dataset
```

For sentiment analysis, reviews may need labels such as:

```text
Positive
Negative
```

<div style="background-color:#f8f9fa; border-left:5px solid #aeb6bf; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

If there is no labelled data, some initial rules/manual labelling may be required. As the dataset grows, you can gradually move toward ML-based solutions.

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">6. Text Cleaning</span>

The purpose of cleaning is to remove unwanted elements from raw text.

## Common Cleaning Tasks

### 6.1 Remove HTML Tags

Example:

```html
<p>Hello</p>
```

After cleaning:

```text
Hello
```

Regular expressions can be used for simple HTML-tag removal.

---

### 6.2 Handle Emojis

Social-media text may contain emojis.

Example:

```text
I love this movie ❤️
```

Depending on the application, emojis may need to be converted into machine-readable representations.

---

### 6.3 Spelling Correction

Social-media text can contain:

* Fast typing
* Abbreviations
* Typing mistakes
* Nearby-key mistakes ("fat fingering")

Example:

```text
hello → hlo
```

Spelling correction can be applied when required.

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

⚠️ Cleaning operations are <b>problem-dependent</b>. Not every NLP project needs every cleaning technique.

</div>

---


# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">7. Basic Text Preprocessing</span>

This stage contains commonly used preprocessing operations.

## 7.1 Tokenization

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Tokenization</b> means breaking text into smaller units called <u>tokens</u>.

</div>

Two common forms:

### Sentence Tokenization

```text
Paragraph
   ↓
Sentences
```

Example:

```text
I like NLP. NLP is interesting.
```

↓

```text
[
  "I like NLP.",
  "NLP is interesting."
]
```

### Word Tokenization

```text
Sentence
   ↓
Words
```

Example:

```text
"I love NLP"
```

↓

```text
["I", "love", "NLP"]
```

Tokenization is generally an important part of the NLP pipeline.

---

## 7.2 Stop Word Removal

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Stop words</b> are common words that may contribute relatively little semantic information for certain NLP tasks.

</div>

Examples:

```text
the
is
a
an
and
of
to
```

Example:

```text
"The movie is very good"
```

Possible result:

```text
["movie", "very", "good"]
```

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

⚠️ Stop-word removal is <b>not always appropriate</b>. It depends on the application.

</div>

---

## 7.3 Stemming

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Stemming</b> attempts to reduce related words to a common root/stem.

</div>

Example:

```text
playing
played
plays
```

↓

```text
play
```

Stemming may produce a root that is not necessarily a proper dictionary word.

---

## 7.4 Lemmatization

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Lemmatization</b> converts words to their meaningful base/dictionary form.

</div>

Example:

```text
running → run
better  → good
```

### Stemming vs Lemmatization

| Stemming             | Lemmatization                  |
| --------------------- | ------------------------------- |
| Faster/simple         | More linguistically informed    |
| Produces stems        | Produces meaningful base forms  |
| Can create non-words  | Usually gives valid words       |
| Less accurate         | Generally more accurate         |

---

## 7.5 Remove Punctuation

Depending on the application, punctuation can be removed.

Example:

```text
Hello, World!
```

↓

```text
Hello World
```

---

## 7.6 Remove Digits

Numbers may sometimes be removed when they don't contribute useful information.

Example:

```text
I watched 3 movies.
```

Possible result:

```text
I watched movies.
```

Again, this depends on the problem.

---

## 7.7 Lowercasing

Convert text into a consistent case.

```text
Natural Language Processing
```

↓

```text
natural language processing
```

This helps maintain consistency.

---

## 7.8 Language Detection

Some applications need to identify the language of incoming text.

Example:

```text
Input Text
    ↓
Language Detection
    ↓
English / Hindi / Spanish / ...
```

<div style="background-color:#f8f9fa; border-left:5px solid #aeb6bf; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

This can be important for applications such as translation systems.

</div>

---


# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">8. Advanced Text Preprocessing 🧠</span>

Advanced preprocessing is generally used for more specific/complex NLP applications.

The lecture highlights:

### 8.1 POS Tagging

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Part-of-Speech (POS) Tagging</b> assigns a grammatical category to each word.

</div>

Example:

```text
I am living in Gurgaon.
```

Possible categories:

```text
I        → Pronoun
am       → Verb
living   → Verb
in       → Preposition
Gurgaon  → Proper Noun
```

POS information can be useful in complex NLP applications such as conversational systems.

---

### 8.2 Parsing

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Parsing</b> attempts to understand the <u>structure of a sentence</u>.

</div>

It can help identify relationships between different parts of a sentence.

```text
Sentence
   ↓
Parse Structure
   ↓
Understand grammatical relationships
```

Useful for applications such as:

* Chatbots
* Question-answering systems
* Language understanding

---

### 8.3 Coreference Resolution

<div style="background-color:#eafaf6; border-left:5px solid #48c9b0; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

<b>Coreference resolution</b> identifies when different words/phrases refer to the same entity.

</div>

Example:

```text
Rohit went to the store.
He bought a laptop.
```

Here:

```text
He → Rohit
```

A human understands this easily, but a machine may need explicit processing to establish the relationship.

---

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">11. ML vs Deep Learning Feature Engineering</span>

This is one of the most important concepts from the lecture.

## Traditional Machine Learning

```text
Raw Text
   ↓
Text Preparation
   ↓
Manual Feature Engineering
   ↓
ML Algorithm
   ↓
Prediction
```

The data scientist/engineer generally decides which features to create using **domain knowledge**.

### Advantages

* Features can be understandable.
* Model predictions can be easier to interpret.
* Domain knowledge can be directly incorporated.

### Disadvantages

* Requires domain knowledge.
* Requires additional effort.
* Poorly designed features can hurt model performance.

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

The transcript explicitly highlights the additional domain knowledge/effort and the possibility of harmful features.

</div>

---

## Deep Learning

```text
Raw Text
   ↓
Text Preparation
   ↓
Deep Learning Model
   ↓
Learned Representations
   ↓
Prediction
```

The model can automatically learn useful representations/features.

### Advantage

Less manual feature engineering.

### Disadvantage

The learned representations can be difficult to interpret.

<div style="background-color:#fdedf3; border-left:5px solid #f1948a; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

This creates an <b>interpretability problem</b>: it may be difficult to explain exactly why the model produced a particular prediction.

</div>

### ML vs DL Feature Engineering — Quick Comparison

| Aspect | Traditional ML | Deep Learning |
|---|---|---|
| Feature creation | Manual, domain-knowledge driven | Learned automatically |
| Interpretability | Easier | Harder |
| Effort | Higher upfront effort | Lower manual effort |
| Risk | Poor features hurt performance | Representations hard to explain |

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">12. Modeling 🤖</span>

Once the data is in a suitable numerical format, we can apply a model.

The lecture presents four broad approaches:

```text
1. Rule-based approach
2. Machine Learning
3. Deep Learning
4. Cloud APIs
```

---

## 12.1 Rule-Based Approach

Useful when:

* Very little data is available.
* The problem can be handled using clear rules.

Example:

```text
IF email comes from a known spam address
THEN classify as spam
```

Another rule:

```text
IF email contains many spam-related words
THEN classify as spam
```

---

## 12.2 Machine Learning

When enough data becomes available, ML algorithms can be used.

```text
Data
 ↓
Features
 ↓
ML Algorithm
 ↓
Prediction
```

---

## 12.3 Deep Learning

When a sufficiently large amount of data is available, deep-learning approaches can be considered.

```text
Large Dataset
      ↓
Deep Learning
      ↓
Learned Representations
      ↓
Prediction
```

<div style="background-color:#f8f9fa; border-left:5px solid #aeb6bf; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

The lecture emphasizes that deep learning generally benefits from larger amounts of training data.

</div>

---

## 12.4 Cloud APIs

Instead of building the complete model yourself, you may use an existing cloud/API-based solution.

Examples discussed conceptually include cloud platforms such as:

* Google Cloud
* AWS
* Azure

The choice depends on factors such as **amount of data** and **nature of the problem**.

---


# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">13. Choosing the Modeling Approach</span>

A major decision depends on:

### Factor 1 — Amount of Data

```text
Very Little Data
       ↓
Rule-based / heuristic approach

More Data
       ↓
Machine Learning

Large Amount of Data
       ↓
Deep Learning may become suitable
```

### Factor 2 — Nature of Problem

The type of NLP problem also affects the appropriate approach.

<div style="background-color:#fef9e7; border-left:5px solid #f4d03f; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

There is no single technique that should automatically be used for every NLP problem.

</div>

---


# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">14. Hybrid Approach 🔄</span>

Different approaches can also be combined.

For example:

```text
Rule-based knowledge
        +
ML features
        ↓
ML Model
```

A rule that was useful when data was scarce can sometimes be converted into a feature when more data becomes available.

Example:

```text
Feature 1 → Positive word count
Feature 2 → Negative word count
Feature 3 → Email came from known spam ID? (0/1)
```

<div style="background-color:#eafaf1; border-left:5px solid #52be80; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

🎯 The rule-based knowledge is therefore incorporated into the ML system as an additional feature.

Modeling contains two important parts:

Apply an algorithm to the prepared data.

### Evaluation

Measure how well the model performs and compare different models.

```text
Deployment
    ↓
Monitoring
    ↓
Model Updating
```

# <span style="color:#2C3E50; font-family:sans-serif; font-weight:800; border-bottom:3px solid #3498DB; padding-bottom:5px;">⭐ Most Important Takeaways</span>

<div style="background-color:#eafaf1; border-left:5px solid #52be80; padding:16px 18px; margin:14px 0 18px 0; border-radius:10px;">

1. **NLP Pipeline = end-to-end process for building an NLP system.**
2. The commonly discussed pipeline has **5 major stages**:
   * Data Acquisition
   * Text Preparation
   * Feature Engineering
   * Modeling
   * Deployment
3. **Data Acquisition is critical** because the entire pipeline depends on having useful data.
4. Data can come from:
   * Internal sources
   * Public datasets
   * Websites
   * APIs
   * PDFs
   * Images
   * Audio
5. **Text Preparation is problem-dependent.**
6. Tokenization converts text into smaller units such as sentences or words.
7. Stop-word removal, stemming, punctuation removal, digit removal and lowercasing are generally **optional/problem-dependent** operations.
8. POS tagging, parsing and coreference resolution are examples of **advanced NLP preprocessing**.
9. **Feature Engineering converts text into numerical representations.**
10. Traditional ML generally requires more **manual feature engineering**.
11. Deep learning can learn useful representations/features automatically.
12. Traditional ML can be more interpretable, while deep learning can be harder to interpret.
13. Modeling approach depends on **data availability + nature of the problem**.
14. Real-world NLP pipelines are often **non-linear**.
15. After deployment, you may return to earlier stages to improve the system.
