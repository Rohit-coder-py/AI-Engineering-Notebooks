What is Text Classification?
</h2>

**Text Classification** is the process of automatically assigning **predefined categories or labels** to a piece of text based on its content.

### <span style="color:#2980B9;">Example</span>

```text
"I loved this movie!"
        ↓
   Text Classification
        ↓
    Positive
```

```text
"This product is terrible."
        ↓
   Text Classification
        ↓
    Negative
```

### <span style="color:#8E44AD;">Common Examples</span>

- **Sentiment Analysis** → Positive / Negative / Neutral
- **Spam Detection** → Spam / Not Spam
- **Topic Classification** → Sports / Politics / Technology
- **Support Tickets** → Payment / Refund / Delivery

> **Text → Classification Model → Predicted Category**

# <span style="color:#1B4F72;"><u>Types of Text Classification</span>

### <span style="color:#2980B9;">1. Binary Classification</span>

Classification into **2 classes**.

```text
Text → Class A / Class B
```

**Example:**  
Spam / Not Spam

---

### <span style="color:#8E44AD;">2. Multiclass Classification</span>

Classification into **more than 2 classes**, where each text belongs to **one class**.

```text
News
 ├── Sports
 ├── Politics
 └── Science
```

**Example:**  
A news article → Sports / Politics / Science

---

### <span style="color:#16A085;">3. Multilabel Classification</span>

A single text can belong to **multiple classes at the same time**.

```text
News Article → Sports + Cricket
```

> **Binary → 2 classes**  
> **Multiclass → One class from many**  
> **Multilabel → Multiple classes at once**

# <span style="color:#1B4F72;">Applications of NLP</span>

### <span style="color:#2980B9;">1. Sentiment Analysis</span>

- Determines whether text expresses **Positive, Negative, or Neutral** sentiment.

### <span style="color:#8E44AD;">2. Spam Detection</span>

- Classifies messages or emails as **Spam or Not Spam**.

### <span style="color:#16A085;">3. News Classification</span>

- Categorizes news articles into topics such as **Sports, Politics, Business, or Science**.

### <span style="color:#D35400;">4. Customer Support</span>

- Classifies customer messages into categories such as **Payment, Refund, Delivery, or Account Issues**.

### <span style="color:#7D3C98;">5. Document Classification</span>

- Automatically categorizes documents based on their **content or topic**.

> **NLP → Understand Text → Classify / Analyze → Make Decisions**
