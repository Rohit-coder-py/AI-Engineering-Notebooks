Word Embeddings are a way of converting words into numerical vectors so that a machine-learning model can understand the meaning and relationships between words.

Simple definition:
Word Embedding = Representing a word as a vector of numbers that captures its meaning/context.

</div>

### <span style="color:#2980B9;">Key Idea</span>

**Word → Vector → Meaning & Relationships**

> Similar words → **Closer vectors**  
> Different words → **Farther vectors**

| One-Hot Encoding                    | Word Embeddings                  |
| ----------------------------------- | -------------------------------- |
| Sparse vector                       | Dense vector                     |
| Very high dimensional               | Usually lower dimensional        |
| Doesn't capture semantic similarity | Captures semantic relationships  |
| Mostly 0s                           | Many meaningful numerical values |
| Example: `[0,0,1,0,0]`              | Example: `[0.21,-0.45,0.78,...]` |

### <span style="color:#2980B9;">1. Frequency-Based</span>

These methods create word representations by looking at **how frequently words occur** and **how they co-occur with other words**.

**Examples:**

- **BoW (Bag of Words)** → Represents text using the **frequency/count of words**.
- **TF-IDF** → Gives higher weight to words that are **important in a document but less common across the corpus**.
- **GloVe** → Uses **global word co-occurrence statistics** to learn word vectors.
- **Matrix Factorization** → Factorizes a **word-context/co-occurrence matrix** to obtain lower-dimensional word representations.

### <span style="color:#8E44AD;">2. Prediction-Based</span>

These methods learn word representations by **predicting words from their context or context from words**.

**Example:**

- **Word2Vec** → Learns word vectors by using a neural network to predict **context words from a target word** or **a target word from its context**.

</div>

### <span style="color:#27AE60;">Basic Difference</span>

**Frequency-Based** → *How often do words occur/co-occur?*  
**Prediction-Based** → *Can we predict a word from its context?*

# <span style="color:#1B4F72;">Intuition Behind Word2Vec</span>

The main idea of Word2Vec is:

> **Words that appear in similar contexts tend to have similar meanings.**

### <span style="color:#2980B9;">Example</span>

Consider these sentences:

```text
The cat drinks milk.
The cat eats fish.
The dog drinks milk.
The dog eats meat.
```

The words around **cat** and **dog** are quite similar:

```text
cat → drinks, milk, eats, fish

dog → drinks, milk, eats, meat
```

Because they are used in **similar situations**, Word2Vec learns that **cat** and **dog** are related and gives them **similar vector representations**.

---

### <span style="color:#8E44AD;">Learning From Context</span>

Imagine you don't know what **king** means.

You see:

```text
The king ruled the kingdom.
The king spoke to the queen.
The king lived in the palace.
The king wore a crown.
```

By repeatedly seeing the words around **king**, you can learn something about its meaning.

Word2Vec works with a similar idea:

```text
Word
  ↓
Observe surrounding words
  ↓
Learn usage patterns
  ↓
Learn relationships
  ↓
Create a numerical vector
```

---

### <span style="color:#16A085;">What Does the Vector Represent?</span>

The vector is a **numerical representation learned from how a word is used in text**.

```text
cat → [0.21, -0.43, 0.67, 0.12, ...]

dog → [0.25, -0.40, 0.64, 0.15, ...]

car → [-0.72, 0.81, -0.14, 0.55, ...]
```

You do **not manually decide** what each number means.

The model **learns these numbers from the text and the contexts in which words appear**.

---

### <span style="color:#C0392B;">Core Intuition</span>

Traditional thinking:

> **"How do I represent this word?"**

Word2Vec thinks:

> **"What words usually appear around this word?"**

From these surrounding words, it learns **relationships between words**.

```text
Similar Context
      ↓
Similar Usage
      ↓
Similar Meaning
      ↓
Similar Vectors
```

> **In one line:**  
> **Word2Vec learns word meaning and relationships from the contexts in which words appear.**

# <span style="color:#1B4F72;">Types of Word2Vec :</span>

Word2Vec has **two main architectures**:

### <span style="color:#2980B9;">1. CBOW</span>

**CBOW (Continuous Bag of Words)**

> Uses the **surrounding/context words** to predict the **target word**.

```text
Context Words → Target Word
```

---

### <span style="color:#8E44AD;">2. Skip-Gram</span>

> Uses the **target word** to predict the **surrounding/context words**.

```text
Target Word → Context Words
```

---

### <span style="color:#16A085;">Important</span>

Both **CBOW** and **Skip-Gram** are **shallow neural network architectures** used by Word2Vec to learn **word embeddings**.

```text
                    Word2Vec
                       ↓
              ┌────────┴────────┐
              ↓                 ↓
            CBOW            Skip-Gram
              ↓                 ↓
     Context → Target     Target → Context
```

# <span style="color:#1B4F72;">CBOW — Continuous Bag of Words</span>

## <span style="color:#2980B9;">Basic Idea</span>

CBOW learns to **predict a target word using the surrounding context words**.

Consider the sentence:

> **"The cat is sitting on the mat"**

Assume:

```text
Window Size = 3
```

For **`sitting`** as the target:

```text
The   cat   is   [sitting]   on   the   mat
 ↑     ↑     ↑       ↑       ↑
 └─────┴─────┘       │       └─────┘
    Context          Target
```

### Target Word

```text
sitting
```

### Context Words

```text
cat
is
on
```

Therefore:

```text
X = [cat, is, on]

Y = sitting
```

---

## <span style="color:#8E44AD;">What CBOW Does</span>

```text
Context Words
     ↓
[ cat + is + on ]
     ↓
    CBOW
     ↓
Predict Target
     ↓
  "sitting"
```

> **"Given the surrounding words, what word is most likely to appear here?"**

---

## <span style="color:#16A085;">Multiple Training Examples</span>

From the same sentence, we can create **multiple training examples** by changing the target word.

### Example 1

```text
The [cat] is sitting on the mat

Context → [The, is]
Target  → cat
```

### Example 2

```text
The cat [is] sitting on the mat

Context → [cat, sitting]
Target  → is
```

### Example 3

```text
The cat is [sitting] on the mat

Context → [cat, is, on]
Target  → sitting
```

### Example 4

```text
The cat is sitting [on] the mat

Context → [sitting, the]
Target  → on
```

---

## <span style="color:#C0392B;">Core Intuition</span>

CBOW is essentially a **fill-in-the-blank learning task**.

```text
The cat is ______ on the mat
          ↑
        Target
```

The model looks at the surrounding context and learns to predict the missing word:

```text
cat + is + on
      ↓
    CBOW
      ↓
  "sitting"
```

During this process, the model learns useful **word embeddings**.

> **CBOW = Context Words → Predict the Target Word**

in Skip-gram we basically try to predict context word using target words

<b>Skip-gram vs CBOW :

| Feature         | **CBOW**                                            | **Skip-gram**                                 |
| --------------- | --------------------------------------------------- | --------------------------------------------- |
| Full form       | Continuous Bag of Words                             | Skip-gram                                     |
| Input           | **Context words**                                   | **Target word**                               |
| Output          | **Target word**                                     | **Context words**                             |
| Direction       | Context → Target                                    | Target → Context                              |
| Training speed  | ⚡ Faster                                            | 🐢 Slower                                     |
| Works well with | Frequent words                                      | Rare words                                    |
| Training data   | Needs less computation                              | Needs more computation                        |
| Main idea       | "Given surrounding words, predict the missing word" | "Given a word, predict its surrounding words" |


```text

      ↓

┌───────────────┬───────────────┐

                ↓
        ┌───────┴───────┐
        │ YES           │ NO
        ↓               ↓

# <span style="color:#1B4F72;">Tips to Improve Word2Vec Embedding Quality</span>

### <span style="color:#2980B9;">1. Increase Training Data</span>

- More training data provides the model with **more examples of word usage and context**.
- This generally helps in learning better word relationships.

### <span style="color:#8E44AD;">2. Increase Embedding Dimension</span>

- Increasing the **vector dimension** gives the model more space to capture word relationships.
- However, very high dimensions increase **memory usage and training cost**.

### <span style="color:#16A085;">3. Increase Window Size</span>

- A larger **window size** allows Word2Vec to consider more surrounding words.
- This can help capture **broader contextual relationships**.
- **Problem:** Larger windows increase the number of training examples, so **training takes more time**.

> **More Data + Higher Dimension + Larger Context → Potentially Better Embeddings**
>
> **But:** Higher settings also increase **computational cost and training time**.

````python
model = KeyedVectors.load_word2vec_format('GoogleNews-vectors-negative300.bin.gz',binary=True,limit=500000)
````

````python
model['cricket']

model.most_similar('man')

model.most_similar('cricket')

model.most_similar('facebook')

model.similarity('man','woman')

model.similarity('man','PHP')

model.doesnt_match(['PHP','java','monkey'])
````

````python
#we can also provide operations

vec = model['king'] - model['man'] + model['woman']
model.most_similar([vec])
````

````python
vec = model['INR'] - model ['India'] + model['England']
model.most_similar([vec])
````
