<h2 style="font-size:32px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Linked List</strong>
</h2>

<h3 style="font-size:24px; color:#374151;">
    <strong>Definition</strong>
</h3>

<p>

A <strong>Linked List</strong> is a <strong>linear data structure</strong> made up of
<strong>nodes</strong>. Unlike arrays, its elements are <strong>not stored in contiguous memory</strong>.
Each node contains <strong>data</strong> and a <strong>reference (pointer)</strong> to the
next node, forming a chain.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Example</strong>
</h3>

<div style="border:1px solid #D1D5DB; border-radius:10px; padding:16px; background:#FCFCFC; font-family:'Consolas', monospace; font-size:18px;">

10 → 20 → 30 → 40 → None

</div>

<h2 style="font-size:32px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Node</strong>
</h2>

<h3 style="font-size:24px; color:#374151;">
    <strong>Definition</strong>
</h3>

<p>

A <strong>Node</strong> is the <strong>basic building block</strong> of a
<strong>Linked List</strong>. Each node contains two parts:

</p>

<ul style="line-height:1.8;">

<li><strong>Data</strong> – The value stored in the node.</li>

<li><strong>Next</strong> – A pointer (reference) to the next node.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Structure</strong>
</h3>

<div style="border:1px solid #D1D5DB; border-radius:10px; padding:16px; background:#FCFCFC; font-family:'Consolas', monospace; font-size:17px;">

+----------------+<br>
| Data | Next &nbsp;&nbsp;&nbsp;|<br>
+----------------+

</div>

<h2 style="font-size:32px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Singly Linked List (SLL)</strong>
</h2>

<h3 style="font-size:24px; color:#374151;">
    <strong>Definition</strong>
</h3>

<p>

A <strong>Singly Linked List (SLL)</strong> is a type of <strong>Linked List</strong>
in which each node contains <strong>data</strong> and a single
<strong>pointer (<code>next</code>)</strong> to the next node.
Traversal is possible <strong>only in the forward direction</strong>.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Example</strong>
</h3>

<div style="border:1px solid #D1D5DB; border-radius:10px; padding:16px; background:#FCFCFC; font-family:'Consolas', monospace; font-size:18px;">

10 → 20 → 30 → 40 → None

</div>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Key Point</strong>
</h3>

<div style="border-left:4px solid #2563EB; padding:10px 14px;">

Each node stores the address of <strong>only the next node</strong>, so traversal is possible <strong>only in one direction (forward)</strong>.

</div>

<h2 style="font-size:32px; color:#7C3AED; text-align:center;">
    <strong>Singly Linked List vs Doubly Linked List</strong>
</h2>

| Singly Linked List | Doubly Linked List |
|--------------------|--------------------|
| Each node contains **Data** and **Next** pointer. | Each node contains **Previous**, **Data**, and **Next** pointers. |
| Traversal is possible only in the **forward** direction. | Traversal is possible in both **forward** and **backward** directions. |
| Uses **less memory** (one pointer per node). | Uses **more memory** (two pointers per node). |
| Simpler to implement. | More complex to implement. |
| Backward traversal is **not possible**. | Backward traversal is **possible**. |

### Structure

**Singly Linked List**
```text
10 → 20 → 30 → 40 → None
```

**Doubly Linked List**
```text
None ← 10 ⇄ 20 ⇄ 30 ⇄ 40 → None
```

**Key Point:**  
A **Singly Linked List** stores only the **next** node's reference, whereas a **Doubly Linked List** stores references to both the **previous** and **next** nodes.


<h3 style="font-size:24px; color:#374151;">
    <strong>How to create a Singly Linked List :</strong>
</h3>

````python
# Step 1: Create a Node class
# 
# Each node stores:
# 
# data
# next

class Node:
    def __init__(self, data):
        self.data = data      # Store the value
        self.next = None      # Initially points to nothing


#Step 2 :Create nodes(they have no conncetion between them )
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

#Step 3 : Connect the nodes
node1.next = node2
node2.next = node3
node3.next = node4

# Head of the linked list
head = node1

#Step 5: Traverse (Print the List)

current = head

while current is not None:
    print(current.data, end=" -> ")
    current = current.next

print("None")
````
