<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>Why do we find Time Complexity?</strong>
</h2>

<p>
Time Complexity is used to <strong>compare algorithms</strong> and determine
<strong>which one is more efficient</strong> as the input size <strong><code>n</code></strong> grows.
</p>

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>Order of Time Complexities</strong>
</h2>

<p>
As we move from left to right, the algorithm becomes <strong>less efficient</strong> because the number of operations grows faster.
</p>

<div style="background:#F8FAFC; border:1px solid #CBD5E1; border-radius:8px; padding:12px; font-size:18px; text-align:center;">

<strong>
1 &lt; log(log n) &lt; log(n) &lt; n &lt; n² &lt; n³ &lt; n⁴ &lt; ... &lt; 2ⁿ
</strong>

</div>

| Complexity     | Growth         | Meaning                 |
| -------------- | -------------- | ----------------------- |
| **1 (O(1))**   | Doesn't grow   | ⭐ Best                  |
| **log(log n)** | Extremely slow | ⭐ Excellent (rare)      |
| **log n**      | Very slow      | ⭐ Great (Binary Search) |
| **n**          | Linear         | ✅ Good                  |
| **n²**         | Quadratic      | ⚠️ Slow for large `n`   |
| **n³**         | Cubic          | ❌ Very slow             |
| **n⁴**         | Fourth power   | ❌ Worse                 |
| **2ⁿ**         | Exponential    | 🚫 Avoid for large `n`  |


<b>O is the notation. The expression inside the brackets tells us how the algorithm's work grows as the input size increases.

<span style="background-color:#DBEAFE; padding:4px 10px; border-radius:6px;">
<strong>What is the Dominating Factor?</strong>
</span><br>

<br><b>The dominating factor is the term that grows the fastest as the input size (n) becomes very large.

<b>In Big O, we keep only the dominating factor and ignore everything else.<b>



The dominating factor is the fastest-growing term in an equation. In Big O, we ignore constants and smaller terms, keeping only the dominating factor because it determines how the algorithm behaves for very large inputs.

| Mathematical Function | Dominating Factor | Big O                                                               |
| --------------------- | ----------------- | ------------------------------------------------------------------- |
| `y = 5`               | `5`               | **O(1)**                                                            |
| `y = 2x + 5`          | `x`               | **O(n)**                                                            |
| `y = 2x² + 3x - 5`    | `x²`              | **O(n²)**                                                           |
| `y = 3x³ - 2x² + 5`   | `x³`              | **O(n³)**                                                           |
| `y = log(x) + 2`      | `log(x)`          | **O(log n)**                                                        |
| `y = 3ˣ + 3`          | `3ˣ`              | **Exponential** *(commonly written as `O(2ⁿ)` in introductory DSA)* |


<span style="background-color:#DBEAFE; padding:4px 10px; border-radius:6px;">
<strong>Types of Analysis (Asymptotic Analysis)</strong>
</span><br>

<b>When we analyze an algorithm, we don't always assume the same input. Sometimes the input is favorable, sometimes average, and sometimes the worst possible.

So we analyze algorithms in three ways: 

| Analysis Type    | Notation      | Meaning                                  |
| ---------------- | ------------- | ---------------------------------------- |
| **Best Case**    | **Ω (Omega)** | Minimum time taken by the algorithm.     |
| **Average Case** | **Θ (Theta)** | Expected time taken on an average input. |
| **Worst Case**   | **O (Big O)** | Maximum time taken by the algorithm.     |

<b>Example 1 

```python
n = 5                      #1 unit
print(n)                   #1 unit 


for i in range(0,n):      # n unit (it depends on number of n maybe its 30 , 12 or something)
    print(i)              # n unit 
```


sum up all units of the code now to convert them into equations 

1 + 1 + n+n

= 2n + 2

now in (2n +2 ) the most dominating Factor is <b> <mark> (n)


and since it is a linear equation so : 

Time Complexity = O(n)

x = Total number of operations

1 + 1 + n + n

= 2n + 2

Dominating Factor = n

Time Complexity = O(n)

<b> Another example

```python
n = 5           # 1 operation
print(n)        # 1 operation

for i in range(n):      # n iterations
    print(i)            # n operations

for j in range(n):      # n iterations
    print(j * j)        # n operations
```

x = Total number of operations

1 + 1 + n + n + n + n

= 4n + 2

Dominating Factor = n

Time Complexity = O(n)
