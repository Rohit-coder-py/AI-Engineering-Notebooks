<h2 style="font-size:30px; color:#374151;">
    <strong>What is DSA?</strong>
</h2>

<p>
<span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>DSA = Data Structures + Algorithms</strong></span>
</p>

<p>
<strong>Data Structure</strong> = <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>How you organize data.</strong></span>
</p>

<p>
<strong>Algorithm</strong> = <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>How you solve a problem using that data.</strong></span>
</p>

</h2>

<p>

</h2>

<h2 style="font-size:34px; color:#0F172A; border-left:8px solid #3B82F6; padding-left:14px; font-family:Georgia, serif;">
    <strong>⏱️ Time Complexity</strong>
</h2>

<div style="background:#F8FAFC; border:2px solid #CBD5E1; border-radius:10px; padding:16px; margin-top:12px;">

<p style="font-size:17px; line-height:1.8;">

<span style="background-color:#DBEAFE; padding:4px 10px; border-radius:6px;">
<strong>Time Complexity</strong>
</span>

is a way to measure <strong>how the amount of work</strong> an algorithm performs as the
<span style="background-color:#DCFCE7; padding:4px 10px; border-radius:6px;">
<strong>input size (<code>n</code>)</strong>
</span>
increases.

</p>

</div>

Instead, it tells us

<span style="background:#FEF3C7; padding:3px 8px; border-radius:5px;">
<strong>how the number of operations grows</strong>
</span>

when the value of <strong><code>n</code></strong> becomes larger.

</p>

<strong style="color:#166534;">✅ Key Idea</strong>

<p style="margin-top:8px;">

As the input size
<span style="background:#DCFCE7; padding:3px 8px; border-radius:5px;"><strong><code>n</code></strong></span>
increases, <strong>Time Complexity describes how quickly the algorithm's work increases.</strong>

</p>

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>Why do we find Time Complexity?</strong>
</h2>

<p>
Time Complexity is used to <strong>compare algorithms</strong> and determine
<strong>which one is more efficient</strong> as the input size <strong><code>n</code></strong> grows.
</p>

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>Order of Time Complexities</strong>
</h2>

<p>
As we move from left to right, the algorithm becomes <strong>less efficient</strong> because the number of operations grows faster.
</p>

<div style="background:#F8FAFC; border:1px solid #CBD5E1; border-radius:8px; padding:12px; font-size:18px; text-align:center;">

<strong>
1 &lt; log(log n) &lt; log(n) &lt; n &lt; n² &lt; n³ &lt; n⁴ &lt; ... &lt; 2ⁿ
</strong>

</div>

| Complexity     | Growth         | Meaning                 |
| -------------- | -------------- | ----------------------- |
| **1 (O(1))**   | Doesn't grow   | ⭐ Best                  |
| **log(log n)** | Extremely slow | ⭐ Excellent (rare)      |
| **log n**      | Very slow      | ⭐ Great (Binary Search) |
| **n**          | Linear         | ✅ Good                  |
| **n²**         | Quadratic      | ⚠️ Slow for large `n`   |
| **n³**         | Cubic          | ❌ Very slow             |
| **n⁴**         | Fourth power   | ❌ Worse                 |
| **2ⁿ**         | Exponential    | 🚫 Avoid for large `n`  |


<b>O is the notation. The expression inside the brackets tells us how the algorithm's work grows as the input size increases.
