<h2 style="font-size:32px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Array</strong>
</h2>

<h3 style="font-size:24px; color:#374151;">
    <strong>Definition</strong>
</h3>

<p>
An <strong>array</strong> is a <strong>linear data structure</strong> that stores elements in
<strong>contiguous memory locations</strong>.
</p>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Properties</strong>
</h3>

<ul style="line-height:1.8;">
<li>Linear Data Structure</li>
<li>Contiguous Memory Allocation</li>
<li>Index-Based Access</li>
<li>Random Access</li>
<li>Same Data Type (Traditional Arrays)</li>
<li>Fixed Size (Traditional Arrays)</li>
<li>Easy Traversal</li>
</ul>

---

<h3 style="font-size:24px; color:#7C2D12;">
    <strong>Why do we use Arrays?</strong>
</h3>

<ul style="line-height:1.8;">
<li>Store multiple values using one variable.</li>
<li>Organize data efficiently.</li>
<li>Fast access using an index.</li>
</ul>

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">
<li>Fast index access (<strong>O(1)</strong>)</li>
<li>Easy to traverse</li>
<li>Memory efficient</li>
<li>Simple implementation</li>
</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Disadvantages</strong>
</h3>

<ul style="line-height:1.8;">
<li>Fixed size (Traditional Arrays)</li>
<li>Slow insertion (<strong>O(n)</strong>)</li>
<li>Slow deletion (<strong>O(n)</strong>)</li>
<li>Requires contiguous memory</li>
</ul>

## <mark>Creating a array using array module 

| Type Code | Data Type          | Byte Size | Description                |
| :-------: | ------------------ | :-------: | -------------------------- |
| `'b'`     | Signed Byte        | 1 Byte    | Signed integer             |
| `'B'`     | Unsigned Byte      | 1 Byte    | Unsigned integer           |
| `'h'`     | Signed Short       | 2 Bytes   | Signed short integer       |
| `'H'`     | Unsigned Short     | 2 Bytes   | Unsigned short integer     |
| `'i'`     | Signed Integer     | 2 or 4*   | Signed integer             |
| `'I'`     | Unsigned Integer   | 2 or 4*   | Unsigned integer           |
| `'l'`     | Signed Long        | 4 Bytes   | Signed long integer        |
| `'L'`     | Unsigned Long      | 4 Bytes   | Unsigned long integer      |
| `'q'`     | Signed Long Long   | 8 Bytes   | Signed long long integer   |
| `'Q'`     | Unsigned Long Long | 8 Bytes   | Unsigned long long integer |
| `'f'`     | Float              | 4 Bytes   | Single-precision float     |
| `'d'`     | Double             | 8 Bytes   | Double-precision float     |
| `'u'`     | Unicode Character* | 2 Bytes   | Unicode character          |

```python
from array import array

arr = array('i', [10, 20, 30, 90, 50])

print(arr)
```

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Array Module Limitation</strong>
</h2>

<p>

The <strong><code>array</code> module</strong> is designed to store only
<strong>fixed-size primitive data types</strong> (mainly numeric values).
It <strong>cannot store Python strings or mixed data types.</strong>

</p>

<div style="border-left:4px solid #2563EB; padding:10px 14px; margin-top:12px;">

<strong>For strings, use Python Lists.</strong>

</div>

```python
#we can also build one only for unicode character(single character)

chr = array('u',["H","P"])

chr
```
