<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Array Operations </strong>
</h2>


```python
#know typecode of an arr

print(arr.typecode)


#know length of array

print(f"Length : {len(arr)}")


#access Elements

print(f"First Element  : {arr[0]}")
print(f"Last Element   : {arr[-1]}")


#update Elements

print(f"Before : {arr}")

arr[3] = 980

print(f"After  : {arr}")
```

```python
#append Element

print(f"Before : {arr}")

arr.append(100)

print(f"After  : {arr}")


#insert Element

print(f"Before : {arr}")

arr.insert(2, 999)

print(f"After  : {arr}")


#remove Element

print(f"Before : {arr}")

arr.remove(20)

print(f"After  : {arr}")


#pop Element

print(f"Before : {arr}")

removed = arr.pop()

print(f"Removed : {removed}")
print(f"After   : {arr}")


#find Index

print(f"Index of 980 : {arr.index(980)}")
```

```python
#count Occurrences

arr.append(10)
arr.append(10)

print(f"Count of 10 : {arr.count(10)}")


#reverse Array

print(f"Before : {arr}")

arr.reverse()

print(f"After  : {arr}")


#traverse Array

print("Traversal :", end=" ")

for i in arr:
    print(i, end=" ")

print()


#convert Array to List

lst = arr.tolist()

print(lst)
print(type(lst))
```

```python
arr2 = array('i',[40,32,44,55,24])
print("\n Before:",arr2,"\n")
print("Traversal :", end=" ")

for i in arr2:
    print(i, end=" ")

print()
```

```python
arr3 = array('i',[40,32,44,55,24])

print(f"Before : {arr3}")

arr3.pop()

print(f"After  : {arr3}")
```

```python
#copy existing values in a new array 

copyArr = array(arr3.typecode, (i+1 for i in arr3))      #use list comprehension

copyArr
```

```python
# Reverse a arr 

rev = arr3[::-1]
print(arr3)
rev
```

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Array Slicing and Indexing</strong>
</h3>


<b>array</b>[start : stop : step]

```python
arr = array('i', [10, 20, 30, 40, 50, 60])

print("Original Array :", arr)

print("\nIndexing")

print("First Element       :", arr[0])

print("Third Element       :", arr[2])

print("Last Element        :", arr[-1])

print("Second Last Element :", arr[-2])


#Slicing

print("arr[:]    :", arr[:])
print("arr[1:4]  :", arr[1:4])
print("arr[:3]   :", arr[:3])

print("arr[3:]   :", arr[3:])

print("arr[::2]  :", arr[::2])
print("arr[::-1] :", arr[::-1])
arr[1::2]
```

```python
#take input from user to allocate it in array 


arr = array('i', [])

n = int(input("Enter a number"))

for i in range(n):
    arr.append(int(input("Enter next input")))

print(*arr)
```

```python
#know index of any element 

arr = array('i', [10, 20, 30, 40, 50, 60])


arr.index(60)

```

## <mark>Creating a array using numpy module 

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>Why is NumPy Better than the <code>array</code> Module?</strong>
</h2>

<ul style="line-height:1.8;">

<li>Supports <strong>1D, 2D, 3D, and higher-dimensional arrays.</strong></li>

<li>Faster and more <strong>memory-efficient</strong> for numerical computations.</li>

<li>Provides powerful <strong>vectorized mathematical operations.</strong></li>

<li>Offers built-in functions for <strong>statistics, linear algebra, and matrix operations.</strong></li>

<li>Integrates with <strong>Pandas, Matplotlib, Scikit-learn, TensorFlow, and PyTorch.</strong></li>

<li>Industry standard for <strong>AI, Machine Learning, Deep Learning, and Data Science.</strong></li>

<li>Better suited for handling <strong>large datasets.</strong></li>

</ul>

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])


for x in arr:
    print(x,end = ' ')
    
print('\n')

for i in range(len(arr)):
    print(arr[i],end = ' ')
    
```

```python
#we can also create hetrogenrous array using numpy module 


arr_hetro = np.array([10, 20, 30, 40, 50,9.0,9.383739,"Harsh"])

print(arr_hetro)
```

```python
#how to simply convert datatypes iun numpy array 

arr = np.array([10, 20, 30, 40, 50])


new_arr = np.array([10, 20, 30, 40, 50],float)

new_arr
```

```python
#create a array with arthemetic progresion using numpy 

#syntax = linspace(start :end: no.of partition)
arr = linspace(10,20,4)

for i in arr:
    print(i)
```

```python
#create an array containing a sequence of numbers.

#syntax ; np.arange(start, stop, step)

arr = np.arange(2, 11, 2)

print(arr)
```

```python
#create an array containing logarithmically spaced numbers (powers of a base).

#syntax : np.logspace(start, stop, num=50, base=10)

arr = np.logspace(0, 3, 4)

print(arr)
```


Explanation:

start = 0 → 10⁰ = 1

stop = 3 → 10³ = 1000

num = 4 → Generate 4 values

base = 10 → Use powers of 10
