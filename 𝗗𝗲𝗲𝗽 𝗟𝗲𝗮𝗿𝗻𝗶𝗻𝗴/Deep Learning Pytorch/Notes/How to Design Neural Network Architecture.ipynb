# Designing a Neural Network Architecture means deciding how your neural network will be built so it can solve a specific problem effectively.

##### *Neural Network Architecture Design Workflow :*

```text
Understand Problem
        ↓
Know Input Features
        ↓
Choose Input Layer
        ↓
Choose Number of Hidden Layers
        ↓
Choose Number of Neurons
        ↓
Choose Activation Functions
        ↓
Choose Output Layer
        ↓
Choose Loss Function
        ↓
Choose Optimizer
        ↓
Train the Model
        ↓
Evaluate Performance
        ↓
Tune the Architecture (if needed)
```

| Problem                    | Output                   |
| -------------------------- | ------------------------ |
| Regression                 | Predict continuous value |
| Binary Classification      | 2 classes                |
| Multi-Class Classification | More than 2 classes      |
| Image Classification       | CNN                      |
| Text Processing            | RNN / LSTM / Transformer |
