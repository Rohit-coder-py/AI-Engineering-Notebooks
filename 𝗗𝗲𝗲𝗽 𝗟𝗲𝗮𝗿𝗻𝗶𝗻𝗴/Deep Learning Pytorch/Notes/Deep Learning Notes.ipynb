# PyTorch vs TensorFlow

| **Aspect** | **PyTorch** | **TensorFlow** | **Verdict** |
|---|---|---|---|
| **Programming Language** | Primarily **Python** with a Pythonic interface | Supports **Python, C++, Java, JavaScript**, etc. | **PyTorch** for Python-centric development |
| **Ease of Use** | Intuitive syntax and generally easier for beginners | Improved significantly, but can still be complex | **PyTorch** |
| **Deployment & Production** | Strong deployment support with tools such as TorchServe and ONNX | Strong production ecosystem and deployment tools | **TensorFlow** |
| **Performance** | Competitive performance with dynamic/eager execution | Highly optimized for large-scale workloads | **Tie** |
| **Community & Ecosystem** | Strong research community and libraries such as TorchVision and Hugging Face integration | Large and established ecosystem with strong industry adoption | **Depends on use case** |
| **High-Level APIs** | `torch.nn` and libraries such as PyTorch Lightning | Keras provides a high-level API | **TensorFlow** |
| **Mobile & Embedded** | Supports mobile and edge deployment | Strong mobile and embedded support | **TensorFlow** |
| **Preferred Domains** | Strong in **research, prototyping, Computer Vision, and NLP** | Strong in **industry and production** | **Depends on use case** |
| **Learning Curve** | Easier due to intuitive design and dynamic execution | Generally steeper, although improved with Keras | **PyTorch** |
| **Interoperability** | Strong integration with the Python ecosystem and ONNX | Supports interoperability through TensorFlow tools and ONNX | **PyTorch** |
| **Customizability** | High level of customization and flexible model development | Custom operations are possible but can be more complex | **PyTorch** |
| **Deployment Tools** | Provides tools for model serving and deployment | Mature tools such as TensorFlow Serving and TFX | **TensorFlow** |
| **Distributed Training** | Supports distributed training through `torch.distributed` | Extensive support through `tf.distribute` | **TensorFlow** |
| **Pre-trained Models** | Strong access through TorchVision, Hugging Face, and community models | Large collection through TensorFlow Hub and ecosystem | **Tie** |

---

## Key Difference

**PyTorch** is generally preferred for **research, experimentation, rapid prototyping, and Python-centric deep learning**.

**TensorFlow** is generally strong for **production deployment, large-scale systems, and mature deployment tooling**.

---

## Simple Rule to Remember

> **PyTorch → Research + Flexibility + Easy Experimentation**
>
> **TensorFlow → Production + Deployment + Large-Scale Ecosystem**

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>1. Basic Structure of a Perceptron</strong>
</h2>

<pre style="font-family:Consolas, monospace; font-size:15px; line-height:1.6;">
x₁ ── w₁ ──┐
x₂ ── w₂ ──┼──→ Σ + b ──→ z ──→ Activation ──→ Output
x₃ ── w₃ ──┘
</pre>

<p><strong>Where:</strong></p>

<ul>
<li><strong>x₁, x₂, x₃</strong> → Input features</li>
<li><strong>w₁, w₂, w₃</strong> → Weights</li>
<li><strong>b</strong> → Bias</li>
<li><strong>z</strong> → Weighted sum / pre-activation</li>
<li><strong>Activation Function</strong> → Converts <code>z</code> into the neuron's output</li>
</ul>


<h2 style="font-size:30px; color:#166534; border-bottom:2px solid #DCFCE7; padding-bottom:7px; font-family:Arial, sans-serif;">
    <strong>2. Weighted Sum</strong>
</h2>

<p>
The neuron first calculates the <strong>weighted sum</strong>:
</p>

$$
z = x_1w_1 + x_2w_2 + x_3w_3 + b
$$

<p>In general:</p>

$$
z = \sum_{i=1}^{n} x_iw_i + b
$$

<p>
The <strong>weights</strong> determine how strongly each input contributes to the result.
</p>

<h3 style="font-size:23px; color:#374151;">
    <strong>Example</strong>
</h3>

<p>Suppose:</p>

$$
x_1=10,\quad x_2=2,\quad x_3=5
$$

<p>and:</p>

$$
w_1=0.8,\quad w_2=0.1,\quad w_3=0.5
$$

<p>Then:</p>

$$
z=(10)(0.8)+(2)(0.1)+(5)(0.5)+b
$$


<h2 style="font-size:30px; color:#7C2D12; border-bottom:2px solid #FED7AA; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>3. Bias</strong>
</h2>

<p>
<strong>Bias</strong> is an additional learnable value added to the weighted sum.
</p>

$$
z = \sum_{i=1}^{n}x_iw_i+b
$$

<p>
It allows the neuron to <strong>shift the output or decision boundary</strong>
rather than forcing it to pass through the origin.
</p>

<div style="border-left:4px solid #9CA3AF; padding:8px 14px; margin:15px 0;">

<strong>Weights</strong> control the influence of inputs;
<strong>bias</strong> provides an adjustable offset.

</div>


<h2 style="font-size:30px; color:#6B21A8; border-bottom:2px solid #E9D5FF; padding-bottom:7px; font-family:Arial, sans-serif;">
    <strong>4. Activation Function</strong>
</h2>

<p>
The weighted sum is generally passed through an <strong>activation function</strong>:
</p>

$$
a=f(z)
$$

<p>
Therefore, the complete neuron can be represented as:
</p>

$$
a=f(x_1w_1+x_2w_2+x_3w_3+b)
$$

<p><strong>Examples of activation functions:</strong></p>

<ul>
<li><strong>Sigmoid</strong></li>
<li><strong>ReLU</strong></li>
<li><strong>Tanh</strong></li>
<li><strong>Softmax</strong></li>
</ul>


<h2 style="font-size:30px; color:#0F766E; border-bottom:2px solid #CCFBF1; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>5. Alternative Notation</strong>
</h2>

<p>
The same equation can also be written as:
</p>

$$
z=\theta_1x_1+\theta_2x_2+\theta_3x_3+\theta_0
$$

<p><strong>Here:</strong></p>

<ul>
<li><strong>θ₁, θ₂, θ₃</strong> → Weights</li>
<li><strong>θ₀</strong> → Bias</li>
</ul>

<p>So:</p>

$$
w_i \equiv \theta_i
$$

<p>and:</p>

$$
b \equiv \theta_0
$$

<p>
They represent the <strong>same concept</strong> using different notation.
</p>


<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>6. Big Picture</strong>
</h2>

<pre style="font-family:Consolas, monospace; font-size:15px; line-height:1.7;">
Inputs
   ↓
Multiply by Weights
   ↓
Weighted Sum
   ↓
Add Bias
   ↓
z = Σ(xᵢwᵢ) + b
   ↓
Activation Function
   ↓
Output
</pre>

<h3 style="font-size:24px; color:#374151;">
    <strong>Fundamental Formula</strong>
</h3>

$$
\boxed{
\text{Output}=f\left(\sum_{i=1}^{n}x_iw_i+b\right)
}
$$

<p>
This equation is one of the <strong>fundamental building blocks of neural networks</strong>.
</p>

<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:7px; font-family:Georgia, serif;">
    <strong>Forward Propagation vs Backpropagation</strong>
</h2>

<p style="line-height:1.8;">

In <strong>forward propagation</strong>, the network makes a
<strong>prediction</strong> using the current <strong>weights</strong>,
<strong>biases</strong>, and <strong>activation functions</strong>.

In <strong>backpropagation</strong>, the prediction error is measured using a
<strong>loss function</strong>, <strong>gradients</strong> are computed, and the
<strong>optimizer</strong> updates the <strong>weights</strong> and
<strong>biases</strong> to minimize the <strong>loss</strong>.

</p>

$$

|:------:|---------|
| $L$ | Log Loss |
| $y$ | Actual (True) Label |
| $\hat{y}$ | Predicted Probability |
| $\log$ | Natural Logarithm (base $e$) |

<p><strong>In this step:</strong></p>

<ul style="line-height:1.8;">

<li>The prediction error is propagated <strong>backward</strong> through the network.</li>

<li>The <strong>gradient</strong> of the loss is computed for every <strong>weight</strong> and <strong>bias</strong>.</li>

<li>These gradients indicate <strong>how much each weight and bias contributed to the error.</strong></li>

<li>The computed gradients are then passed to the <strong>optimizer</strong>.</li>

<li>Finally, the optimizer updates the <strong>weights</strong> and <strong>biases</strong> to minimize the <strong>loss</strong>.</li>

</ul>



<span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;">
<strong>Backpropagation computes the gradients, and the optimizer uses those gradients to update the weights and biases.</strong>

## Weight Update Formula (Gradient Descent)

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial L}{\partial w}
$$

| Symbol                          | Meaning                                     |
| ------------------------------- | ------------------------------------------- |
| $w_{\text{new}}$                | Updated Weight                              |
| $w_{\text{old}}$                | Current (Old) Weight                        |
| $\eta$                          | Learning Rate                               |
| $L$                             | Loss Function                               |
| $\frac{\partial L}{\partial w}$ | Gradient of Loss with respect to the Weight |


## Backpropagation - Weight Update Using the Chain Rule

After computing the **Loss**, the next step is to update every **weight** in the network.

The optimizer updates each weight using the **Weight Update Formula**:

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial L}{\partial w}
$$

where:

- $w_{\text{new}}$ = Updated Weight
- $w_{\text{old}}$ = Current Weight
- $\eta$ = Learning Rate
- $\frac{\partial L}{\partial w}$ = Gradient computed by Backpropagation

---

## How does Backpropagation compute the Gradient?

Backpropagation uses the **Chain Rule** of calculus.

The gradient of every weight is computed by tracing the path from that weight to the **Loss Function**.

The farther a weight is from the output neuron, the more terms are included in the Chain Rule.

---

## Output Layer Weight (Example: $w_7$)

Since $w_7$ is directly connected to the output neuron, the gradient is:

$$
\frac{\partial L}{\partial w_7}
=
\frac{\partial L}{\partial O_3}
\times
\frac{\partial O_3}{\partial w_7}
$$

Only **two derivatives** are required because the weight directly affects the output neuron.

---

## Hidden Layer Weight (Example: $w_1$)

The weight $w_1$ is farther from the output.

Its effect reaches the loss through the hidden neuron and then the output neuron.

Therefore, the Chain Rule becomes:

$$
\frac{\partial L}{\partial w_1}
=
\frac{\partial L}{\partial O_3}
\times
\frac{\partial O_3}{\partial O_1}
\times
\frac{\partial O_1}{\partial w_1}
$$

Since the path is longer, more derivatives are multiplied together.

---

## Why does the formula become longer?

Every weight influences the **Loss** through one or more neurons.

The Chain Rule follows this complete path.

- Output layer weights have a **short path** to the Loss.
- Hidden layer weights have a **longer path** to the Loss.

Therefore, hidden-layer gradients contain **more terms**.

---

## Complete Training Process

```text
Forward Propagation
        ↓
Prediction (ŷ)
        ↓
Calculate Loss
        ↓
Backpropagation
        ↓
Compute Gradients using the Chain Rule
        ↓
Optimizer (Gradient Descent)
        ↓
Update Weights & Biases
        ↓
Forward Propagation Again
        ↓
Lower Loss
        ↓
Repeat until the Loss is minimized
```

### Key Points

- Backpropagation **does not update the weights**.
- Backpropagation **computes the gradients**.
- The optimizer uses these gradients to update the weights.
- The Chain Rule is used to compute the gradient for every weight.
- The farther a weight is from the output layer, the longer its Chain Rule expression becomes.

<h2 style="font-size:30px; color:#991B1B; border-bottom:2px solid #FECACA; padding-bottom:6px;">
    <strong>What is the Vanishing Gradient Problem?</strong>
</h2>

<p>

The <strong>Vanishing Gradient Problem</strong> occurs during
<strong>backpropagation</strong> when the gradients become so small that the
<strong>earlier layers</strong> of a deep neural network stop learning.

</p>

<p><strong>In simple words:</strong></p>

<ul style="line-height:1.8;">

<li>During training, gradients are multiplied <strong>layer by layer</strong>.</li>

<li>If the gradients are very small (<strong>&lt; 1</strong>), repeated multiplication makes them shrink toward <strong>zero</strong>.</li>

<li>As a result, the <strong>first few layers</strong> (closer to the input) receive almost no updates.</li>

<li>Therefore, the network fails to learn important <strong>low-level features</strong>.</li>

</ul>

so thats why we never use <b><span style="color:red;"><b>SIGMOID ACTIVATION FUNCTION</b></span></b> in hidden layer and always use sigmoid in outout layer

    <strong>1.Linear Activation Function</strong>
</h2>

<p>

A <strong>Linear Activation Function</strong> is the simplest activation function.
It returns the input <strong>without any transformation</strong>.

</p>

### Mathematical Formula

$$
f(x)=x
$$

<p>

The <strong>output is exactly the same as the input.</strong>

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Used in <strong>Regression Problems</strong>.</li>

<li>Usually used in the <strong>Output Layer</strong> of a neural network.</li>

<li>Suitable for predicting <strong>continuous values</strong> such as house price, salary, temperature, etc.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Example</strong>
</h3>

<p>

For <strong>house price prediction</strong>, the model should predict values like
<strong>₹50,00,000</strong> or <strong>₹80,00,000</strong>.

Using a <strong>Sigmoid</strong> or <strong>Tanh</strong> activation would restrict
the output, so a <strong>Linear Activation Function</strong> is preferred.

</p>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitation</strong>
</h3>

<ul style="line-height:1.8;">

<li>If every layer uses a <strong>Linear Activation Function</strong>, the entire neural network behaves like a <strong>single linear model</strong>.</li>

<li>It cannot learn <strong>complex non-linear relationships</strong>.</li>

<li>Therefore, hidden layers usually use <strong>ReLU</strong>, <strong>Sigmoid</strong>, or <strong>Tanh</strong>, while the output layer for regression can remain <strong>Linear</strong>.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>2.Sigmoid Activation Function</strong>
</h2>

<p>

The <strong>Sigmoid Function</strong> is an <strong>S-shaped activation function</strong> that
maps any real-valued input to a value between <strong>0</strong> and <strong>1</strong>.

</p>

### Mathematical Formula

$$
f(x)=\frac{1}{1+e^{-x}}
$$

<p>

Regardless of how large or small the input is, the output always lies in the range
<strong>0 to 1</strong>.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Used in <strong>Binary Classification</strong> problems.</li>

<li>Commonly used in the <strong>Output Layer</strong> to produce a <strong>probability</strong>.</li>

<li>Examples include <strong>Spam / Not Spam</strong>, <strong>Disease / No Disease</strong>, and <strong>Yes / No</strong> predictions.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Example</strong>
</h3>

<p>

If the sigmoid function outputs <strong>0.85</strong>, it can be interpreted as an
<strong>85% probability</strong> that the input belongs to the positive class.

</p>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Can suffer from the <strong>Vanishing Gradient Problem</strong> for very large or very small input values.</li>

<li>Learning becomes <strong>slow</strong> because the gradients approach <strong>0</strong>.</li>

<li>Rarely used in <strong>hidden layers</strong> today; <strong>ReLU</strong> is generally preferred.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>3. Tanh Activation Function</strong>
</h2>

<p>

The <strong>Tanh (Hyperbolic Tangent)</strong> activation function is similar to
<strong>Sigmoid</strong>, but it maps the input to the range
<strong>-1 to +1</strong> instead of <strong>0 to 1</strong>.

</p>

### Mathematical Formula

$$
f(x)=\tanh(x)
$$

<p style="text-align:center;"><strong>or</strong></p>

$$
f(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}
$$

<p>

No matter how large or small the input is, the output always lies between
<strong>-1</strong> and <strong>+1</strong>.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Commonly used in the <strong>Hidden Layers</strong> of neural networks.</li>

<li>Suitable when the data contains both <strong>positive and negative values</strong>.</li>

<li>Its output is <strong>zero-centered</strong>, which helps optimization.</li>

</ul>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li><strong>Zero-centered</strong> outputs.</li>

<li>Provides <strong>stronger gradients than Sigmoid</strong> in the range <strong>(-1, 1)</strong>.</li>

<li>Can lead to <strong>faster learning</strong> than Sigmoid.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Suffers from the <strong>Vanishing Gradient Problem</strong> for very large positive or negative inputs.</li>

<li>In modern deep learning, <strong>ReLU</strong> is generally preferred for hidden layers.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>4. ReLU Activation Function</strong>
</h2>

<p>

<strong>ReLU</strong> stands for <strong>Rectified Linear Unit</strong>. It is one of the
most widely used activation functions in modern deep learning.

</p>

### Mathematical Formula

$$
f(x)=\max(0,x)
$$

<p><strong>This means:</strong></p>

<ul style="line-height:1.8;">

<li>If <strong>x &lt; 0</strong>, then <strong>f(x)=0</strong>.</li>

<li>If <strong>x &ge; 0</strong>, then <strong>f(x)=x</strong>.</li>

</ul>

<p>

It passes all <strong>positive values</strong> unchanged and converts all
<strong>negative values</strong> to <strong>0</strong>.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Commonly used in the <strong>Hidden Layers</strong> of deep neural networks.</li>

<li>Widely used in <strong>CNNs</strong>, <strong>Computer Vision</strong>, <strong>NLP</strong>, and many other deep learning applications.</li>

</ul>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li>Very <strong>simple</strong> and <strong>computationally efficient</strong>.</li>

<li>Helps reduce the <strong>Vanishing Gradient Problem</strong>.</li>

<li>Enables <strong>faster training</strong> of deep neural networks.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Can suffer from the <strong>Dying ReLU Problem</strong>, where some neurons always output <strong>0</strong>.</li>

<li>Not differentiable at <strong>x = 0</strong>, although this is rarely a practical issue.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>5.Leaky ReLU Activation Function</strong>
</h2>

<p>

The <strong>Leaky ReLU</strong> activation function is an improved version of
<strong>ReLU</strong>. Instead of returning <strong>0</strong> for negative inputs,
it returns a <strong>small negative value</strong>.

</p>

### Mathematical Formula

$$
f(x)=
\begin{cases}
x, & \text{if } x>0\\
0.01x, & \text{if } x\le0
\end{cases}
$$

<p>

This small negative slope ensures that neurons continue to learn even when the
input is negative.

</p>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li>Helps solve the <strong>Dying ReLU (Dead Neuron)</strong> problem.</li>

<li>Simple and <strong>computationally efficient</strong>.</li>

<li>Provides <strong>better gradient flow</strong> for negative inputs.</li>

<li>Works well in <strong>deep neural networks</strong>.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>The small negative slope may introduce a slight bias.</li>

<li>The slope value (e.g., <strong>0.01</strong>) may require tuning for some applications.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>6.PReLU (Parametric Rectified Linear Unit) Activation Function</strong>
</h2>

<p>

<strong>PReLU (Parametric Rectified Linear Unit)</strong> is an improved version of
<strong>Leaky ReLU</strong>. Unlike Leaky ReLU, where the negative slope is fixed,
<strong>PReLU learns the negative slope automatically during training.</strong>

</p>

### Mathematical Formula

$$
f(x)=
\begin{cases}
x, & \text{if } x>0\\
ax, & \text{if } x\le0
\end{cases}
$$

<p>

where <strong><em>a</em></strong> is a <strong>learnable parameter</strong> representing the
negative slope.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Intuition</strong>
</h3>

<ul style="line-height:1.8;">

<li><strong>ReLU:</strong> Negative values become <strong>0</strong>.</li>

<li><strong>Leaky ReLU:</strong> Negative values use a fixed slope (e.g., <strong>0.01x</strong>).</li>

<li><strong>PReLU:</strong> The model automatically <strong>learns the best negative slope</strong> during training.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Used in the <strong>Hidden Layers</strong> of deep neural networks.</li>

<li>Commonly used in <strong>CNNs</strong> and other deep learning models.</li>

<li>Useful when a <strong>learnable activation function</strong> is preferred over a fixed one.</li>

</ul>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li>Helps solve the <strong>Dying ReLU</strong> problem.</li>

<li>The negative slope is <strong>learned automatically</strong>.</li>

<li>Can improve <strong>accuracy</strong> in deep neural networks.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Introduces <strong>additional trainable parameters</strong>.</li>

<li>May increase the risk of <strong>overfitting</strong> on small datasets.</li>

<li>Slightly more complex than <strong>ReLU</strong> and <strong>Leaky ReLU</strong>.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>7.Swish Activation Function</strong>
</h2>

<p>

The <strong>Swish</strong> activation function is a smooth, non-linear activation
function introduced by <strong>Google researchers</strong>. It combines the input
with the <strong>Sigmoid</strong> function.

</p>

### Mathematical Formula

$$
f(x)=x\cdot\sigma(x)
$$

<p style="text-align:center;"><strong>where</strong></p>

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

<p style="text-align:center;"><strong>or</strong></p>

$$
\boxed{\text{Swish}=x\times\text{Sigmoid}(x)}
$$

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Intuition</strong>
</h3>

<ul style="line-height:1.8;">

<li>For <strong>large positive inputs</strong>, the output is approximately equal to the input.</li>

<li>For <strong>large negative inputs</strong>, the output remains small but is <strong>not exactly zero</strong>.</li>

<li>Around <strong>0</strong>, the curve is <strong>smooth</strong>, making optimization easier.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Used in the <strong>Hidden Layers</strong> of deep neural networks.</li>

<li>Commonly used in <strong>CNNs</strong> and modern deep learning architectures.</li>

<li>Useful when smoother gradients are desired than <strong>ReLU</strong>.</li>

</ul>

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li>Provides a <strong>smooth gradient</strong>, improving gradient flow.</li>

<li>Can learn <strong>complex patterns</strong> due to its non-monotonic nature.</li>

<li>Often improves <strong>accuracy</strong> in deep neural networks.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Requires more computation because it uses the <strong>Sigmoid</strong> function.</li>

<li>May not always outperform <strong>ReLU</strong>.</li>

<li>Generally slower than <strong>ReLU</strong> during training.</li>

</ul>



<h2 style="font-size:30px; color:#1E3A8A; border-bottom:2px solid #DBEAFE; padding-bottom:6px;">
    <strong>8.Softmax Activation Function</strong>
</h2>

<p>

The <strong>Softmax</strong> activation function converts the outputs of a neural
network into a <strong>probability distribution</strong> over multiple classes.

Unlike <strong>Sigmoid</strong>, which is used for
<strong>binary classification</strong>, <strong>Softmax</strong> is used for
<strong>multi-class classification</strong>, where each input belongs to
<strong>exactly one class</strong>.

</p>

### Mathematical Formula

$$
\text{Softmax}(x_i)=\frac{e^{x_i}}{\sum_{j=1}^{n}e^{x_j}}
$$

<p>

where

</p>

<ul style="line-height:1.8;">

<li><strong>\(x_i\)</strong> = Score (logit) of the <strong>i<sup>th</sup></strong> class</li>

<li><strong>\(e\)</strong> = Euler's number (≈ 2.718)</li>

<li><strong>\(n\)</strong> = Total number of classes</li>

</ul>

<p>

The denominator is the sum of the exponentials of <strong>all class scores</strong>.

</p>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Where is it used?</strong>
</h3>

<ul style="line-height:1.8;">

<li>Used in the <strong>Output Layer</strong> for <strong>Multi-Class Classification</strong>.</li>

<li>Produces a probability distribution where the probabilities of all classes <strong>sum to 1</strong>.</li>

</ul>

---

<h3 style="font-size:24px; color:#374151;">
    <strong>Example</strong>
</h3>

<table style="width:320px; border-collapse:collapse;">

<tr style="background:#3B82F6; color:white;">
<th style="padding:8px; border:1px solid #D1D5DB;">Class</th>
<th style="padding:8px; border:1px solid #D1D5DB;">Probability</th>
</tr>

<tr>
<td style="padding:8px; border:1px solid #D1D5DB;">Cat</td>
<td style="padding:8px; border:1px solid #D1D5DB;">0.10</td>
</tr>

<tr style="background:#F9FAFB;">
<td style="padding:8px; border:1px solid #D1D5DB;">Dog</td>
<td style="padding:8px; border:1px solid #D1D5DB;">0.75</td>
</tr>

<tr>
<td style="padding:8px; border:1px solid #D1D5DB;">Horse</td>
<td style="padding:8px; border:1px solid #D1D5DB;">0.15</td>
</tr>

</table>

<p>

Since the highest probability is <strong>0.75</strong>, the model predicts
<strong>Dog</strong>.

</p>

$$
0.10+0.75+0.15=1.00
$$

---

<h3 style="font-size:24px; color:#166534;">
    <strong>Advantages</strong>
</h3>

<ul style="line-height:1.8;">

<li>Converts outputs into <strong>probabilities</strong>.</li>

<li>All probabilities <strong>sum to 1</strong>.</li>

<li>Ideal for <strong>Multi-Class Classification</strong>.</li>

<li>Works well with <strong>Cross-Entropy Loss</strong>.</li>

</ul>

---

<h3 style="font-size:24px; color:#991B1B;">
    <strong>Limitations</strong>
</h3>

<ul style="line-height:1.8;">

<li>Suitable only when each input belongs to <strong>one class</strong>.</li>

<li>More computationally expensive than <strong>Sigmoid</strong>.</li>

<li>Can become <strong>overconfident</strong> by assigning a very high probability to one class.</li>

</ul>

---

<h3 style="font-size:24px; color:#7C3AED;">
    <strong>Quick Tip</strong>
</h3>

<div style="border-left:4px solid #8B5CF6; padding:10px 14px;">

<strong>Sigmoid</strong> → Binary Classification (2 Classes)<br><br>

<strong>Softmax</strong> → Multi-Class Classification (3 or More Classes, One Correct Class)

</div>

<h2 style="margin-top:0; color:#4338CA;">Computation Graph</h2>

<p style="font-size:18px; line-height:1.8; color:#444;">

A <span style="color:#4338CA; font-weight:bold;">Computation Graph</span> is a <span style="color:#DC2626; font-weight:bold;">dynamic graph</span> that <span style="color:#2563EB; font-weight:bold;">PyTorch</span> automatically creates during the <span style="color:#16A34A; font-weight:bold;">forward pass</span> whenever tensors have <span style="color:#7C3AED; font-weight:bold;">requires_grad=True</span>.

</p>

<hr>

<p style="font-size:18px; line-height:1.8; color:#444;">

Instead of only computing the <span style="color:#EA580C; font-weight:bold;">final result</span> (e.g., <span style="color:#2563EB; font-weight:bold;">2 × 4 = 8</span>), <span style="color:#2563EB; font-weight:bold;">PyTorch</span> records every <span style="color:#DC2626; font-weight:bold;">mathematical operation</span> performed, including the <span style="color:#16A34A; font-weight:bold;">input tensors</span>, the <span style="color:#7C3AED; font-weight:bold;">operations</span> (such as multiplication, addition, or activation functions), and the <span style="color:#EA580C; font-weight:bold;">resulting outputs</span>.

</p>

<hr>

<p style="font-size:18px; line-height:1.8; color:#444;">

In the example <span style="color:#2563EB; font-weight:bold;">prediction = weight × x</span>, the graph connects <span style="color:#16A34A; font-weight:bold;">weight</span> and <span style="color:#16A34A; font-weight:bold;">x</span> to the <span style="color:#DC2626; font-weight:bold;">multiplication operation (MulBackward0)</span>, which then produces <span style="color:#EA580C; font-weight:bold;">prediction</span>.

</p>

<hr>

<p style="font-size:18px; line-height:1.8; color:#444;">

This recorded graph allows <span style="color:#0F766E; font-weight:bold;">Autograd</span> to trace the calculations <span style="color:#DC2626; font-weight:bold;">backward</span> during <span style="color:#2563EB; font-weight:bold;">loss.backward()</span>, compute <span style="color:#16A34A; font-weight:bold;">gradients</span> for each <span style="color:#7C3AED; font-weight:bold;">trainable parameter</span>, and enable the <span style="color:#EA580C; font-weight:bold;">optimizer</span> to update the model's <span style="color:#2563EB; font-weight:bold;">weights</span> automatically.

</p>

</div>

````python
#more deeper graph 


# Inputs
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# Forward Pass
a = w * x          # Multiply
y_pred = a + b     # Add bias
y_true = torch.tensor(10.0)

error = y_pred - y_true
loss = error ** 2

print("Loss:", loss)

# Backward Pass
loss.backward()

# Gradients
print("Gradient of w:", w.grad)
print("Gradient of b:", b.grad)
````

<h2 style="margin-top:0; color:#15803D;">Gradient</h2>

<p style="font-size:18px; line-height:1.8; color:#444;">

A <span style="color:#16A34A; font-weight:bold;">gradient</span> tells you:

</p>

<ol style="font-size:18px; line-height:1.9; color:#444;">

<li><span style="color:#2563EB; font-weight:bold;">How much</span> you should change a weight or bias.</li>

<li><span style="color:#DC2626; font-weight:bold;">In which direction</span> you should change it to reduce the <span style="color:#7C3AED; font-weight:bold;">loss</span>.</li>
.grad stores the gradient (how much and in which direction a trainable parameter should change to reduce the loss) after loss.backward() is called.
</ol>

</div>

````python
x = torch. tensor(2.0)
w = torch. tensor(3.0, requires_grad=True)
y_pred = w *x
y_true = torch. tensor(10.0)

loss = (y_pred - y_true) ** 2
print(loss)

loss.backward()


print("Prediction", y_pred. item())
print("Loss", loss.item())
print("w.grad", w.grad.item())
````

````python
import torch

# Inputs
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
y_true = torch.tensor(10.0)

# Forward Pass
y_pred = w * x

# Loss
loss = (y_pred - y_true) ** 2

# Backward Pass
loss.backward()

# Results
print("Prediction :", y_pred.item())
print("Actual     :", y_true.item())
print("Loss       :", loss.item())
print("Gradient   :", w.grad.item())

# Interpret the gradient
if w.grad.item() > 0:
    print("Gradient is POSITIVE (+)")
    print("➡️ Decrease the weight.")
elif w.grad.item() < 0:
    print("Gradient is NEGATIVE (-)")
    print("➡️ Increase the weight.")
else:
    print("Gradient is ZERO")
    print("➡️ Weight is already near the minimum.")
````

<h2 style="margin-top:0; color:#15803D;">Gradient Descent Update Formula</h2>

<span style="color:#2563EB; font-weight:bold;">Formula:</span>

<p style="font-size:22px; font-weight:bold; color:#15803D; margin:0; text-align:center;">

New Value = Old Value − (Learning Rate × Gradient)

</div>

<p style="font-size:18px; line-height:1.8; color:#444;">

Or equivalently,

</p>

<p style="margin:0; font-size:17px; color:#444;">

<span style="color:#15803D; font-weight:bold;">Note:</span> During training, the formula almost always written and used is <span style="color:#2563EB; font-weight:bold;">New Value = Old Value − (Learning Rate × Gradient)</span>.

<div style="background:#FEF2F2; border:2px solid #EF4444; border-radius:14px; padding:24px; margin:20px 0;">

<h2 style="margin-top:0; color:#B91C1C;">w.grad.zero_()</h2>

<p style="font-size:18px; line-height:1.8; color:#444;">

<span style="color:#2563EB; font-weight:bold;">w.grad.zero_()</span> resets the gradient stored in <span style="color:#16A34A; font-weight:bold;">w.grad</span> to <span style="color:#DC2626; font-weight:bold;">0</span>.

</p>

<div style="background:#FFFFFF; border-left:5px solid #EF4444; padding:14px; border-radius:8px; margin-top:12px;">

<code style="font-size:18px; color:#B91C1C;">w.grad.zero_()</code>

</div>

<hr>

<h3 style="color:#B91C1C;">Why is it needed?</h3>

<p style="font-size:18px; line-height:1.8; color:#444;">

<span style="color:#2563EB; font-weight:bold;">PyTorch</span> <span style="color:#DC2626; font-weight:bold;">accumulates (adds)</span> gradients by default.

</p>

<p style="font-size:18px; line-height:1.8; color:#444;">

If we do not reset the gradient after each <span style="color:#16A34A; font-weight:bold;">training step</span>, the <span style="color:#7C3AED; font-weight:bold;">new gradient</span> will be added to the <span style="color:#EA580C; font-weight:bold;">previous gradient</span>, resulting in <span style="color:#DC2626; font-weight:bold;">incorrect weight updates</span>.

</p>

</div>

````python
import torch
#input fixed
x = torch.tensor(2.0)

#learnable number
w = torch. tensor(3.0, requires_grad=True)

#correct answer
y_true = torch. tensor(10.0)

#learning rate
lr = 0.1



y_pred = w*x

loss = (y_pred - y_true)**2

loss.backward()

print("Before update")
print("w:", w.item())
print("loss", loss.item())
print("gradient", w.grad.item())


#now lets setup a grad 

# update weight
with torch.no_grad():
    w -= lr * w.grad

# IMPORTANT: reset gradient
w.grad.zero_()
print("\n -------------------------------------------------\n")
print("After update")
print("w:", w.item())
print("loss", loss.item())
print("gradient", w.grad.item())
print(y_true)
````

<div style="background:#EFF6FF; border:2px solid #3B82F6; border-radius:14px; padding:24px; margin:20px 0;">

<h2 style="margin-top:0; color:#1D4ED8;">Training Loop</h2>

<h3 style="color:#1D4ED8;">Definition</h3>

<p style="font-size:18px; line-height:1.8; color:#444;">

A <span style="color:#2563EB; font-weight:bold;">Training Loop</span> is the process of <span style="color:#DC2626; font-weight:bold;">repeatedly training a model</span> by making <span style="color:#16A34A; font-weight:bold;">predictions</span>, calculating the <span style="color:#7C3AED; font-weight:bold;">loss</span>, computing <span style="color:#EA580C; font-weight:bold;">gradients</span>, updating the <span style="color:#2563EB; font-weight:bold;">weights</span>, and repeating these steps for many <span style="color:#16A34A; font-weight:bold;">iterations</span> or <span style="color:#DC2626; font-weight:bold;">epochs</span> until the model <span style="color:#7C3AED; font-weight:bold;">learns</span>.

</p>

</div>

                START

                  │
                  ▼
        Make Prediction
        (Forward Pass)

                  │
                  ▼
        Calculate Loss

                  │
                  ▼
      Compute Gradients
      (loss.backward())

                  │
                  ▼
        Update Weights

                  │
                  ▼
       Clear Gradients

                  │
                  ▼
        More Epochs Left?

          Yes ───────────► Repeat 🔁
           │
           │ No
           ▼
      Training Finished

````python
x = torch. tensor([1., 2., 3., 4.])
y_true = torch. tensor([2., 4., 6., 8.]) #correct answers 

w = torch.tensor(0.0, requires_grad=True)

lr = 0.1
epochs = 10


for epoch in range(epochs):
    #step1 prediction
    y_pred = w *x
    
    #step -2 loss
    loss = ((y_pred - y_true) ** 2).mean()

    loss.backward()

    #now 

    with torch.no_grad():
        w -= lr * w.grad
    print(f"Epoch {epoch+1}: w={w.item():.4f}, loss={loss.item():.4f}")

    

        
````

## *Optimiser*


## <span style="color:#7C3AED;">Definition</span>

An **<span style="color:#2563EB;">Optimizer</span>** is an **<span style="color:#DC2626;">algorithm</span>** that updates the model's **<span style="color:#16A34A;">weights</span>** and **<span style="color:#EA580C;">biases</span>** to **<span style="color:#059669;">minimize the loss (error)</span>** during training.

---

## <span style="color:#F59E0B;">Simple Explanation</span>

In simple words, an **<span style="color:#2563EB;">optimizer</span>** tells the model:

- ➜ **How much** to adjust its parameters.
- ➜ **In which direction** to adjust them.
- ➜ **After each training step** so that predictions become **<span style="color:#16A34A;">more accurate</span>**.

---

## <span style="color:#7C3AED;">Key Idea</span>

> **Optimizer** = **Decision Maker** for updating the model's **<span style="color:#2563EB;">parameters</span>**.

It uses the **<span style="color:#DC2626;">gradients</span>** calculated by **<span style="color:#16A34A;">Autograd</span>** to update the **<span style="color:#2563EB;">weights</span>** and **<span style="color:#EA580C;">biases</span>**.

---

## <span style="color:#059669;">Workflow</span>

**Prediction**
&nbsp;&nbsp;&nbsp;&nbsp;↓  
**Calculate Loss**
&nbsp;&nbsp;&nbsp;&nbsp;↓  
**Compute Gradients**
&nbsp;&nbsp;&nbsp;&nbsp;↓  
**Optimizer Updates Weights & Biases**
&nbsp;&nbsp;&nbsp;&nbsp;↓  
**Better Predictions**

---

### <span style="color:#DC2626;">Remember</span>

- **Autograd** → Computes **<span style="color:#DC2626;">gradients</span>**.
- **Optimizer** → Uses those gradients to **<span style="color:#059669;">update parameters</span>**.
- **Goal** → **<span style="color:#2563EB;">Reduce the loss</span>** and improve the model after every training step.

# <span style="color:#2563EB;">Role of an Optimizer</span>

## <span style="color:#7C3AED;">What Does an Optimizer Do?</span>

The **<span style="color:#2563EB;">optimizer</span>** uses the **<span style="color:#DC2626;">gradients</span>** computed during **<span style="color:#16A34A;">backpropagation</span>** to update the model's **<span style="color:#EA580C;">parameters</span>** *(weights and biases)*.

---

## <span style="color:#F59E0B;">General Update Rule</span>

```text
New Weight = Old Weight − (Learning Rate × Gradient)
```

---

## <span style="color:#059669;">Where</span>

- **<span style="color:#2563EB;">Weight</span>** → Model parameter.
- **<span style="color:#DC2626;">Gradient</span>** → Direction and magnitude of change.
- **<span style="color:#7C3AED;">Learning Rate</span>** → Step size for updating the parameter.

---

## <span style="color:#0F766E;">Step-by-Step Process</span>

**Backpropagation**
&nbsp;&nbsp;↓  
Computes **<span style="color:#DC2626;">Gradients</span>**
&nbsp;&nbsp;↓  
**Optimizer** reads the **<span style="color:#DC2626;">Gradients</span>**
&nbsp;&nbsp;↓  
Applies the **<span style="color:#7C3AED;">Update Rule</span>**
&nbsp;&nbsp;↓  
Updates the **<span style="color:#2563EB;">Weights & Biases</span>**
&nbsp;&nbsp;↓  
Produces **<span style="color:#16A34A;">Better Predictions</span>**

---

## <span style="color:#B91C1C;">Key Insight</span>

> **Backpropagation** computes the **<span style="color:#DC2626;">gradients</span>**.  
> **Optimizer** uses those **<span style="color:#DC2626;">gradients</span>** to update the model's **<span style="color:#2563EB;">parameters</span>** and **<span style="color:#059669;">reduce the loss</span>**.

````python
import torch
import torch.nn as nn
import torch.optim as optim


class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.Linear(x)


x = torch.tensor([[1.], [2.], [3.], [4.]])
y = torch.tensor([[2.], [4.], [6.], [8.]])

model = MyModel()

loss_fn = nn.MSELoss()

optimiser = optim.Adam(model.parameters(), lr=0.1)

for epoch in range(10):
    optimiser.zero_grad()

    y_pred = model(x)

    loss = loss_fn(y_pred, y)

    loss.backward()

    optimiser.step()

    print(f"Epoch {epoch + 1}, Loss: {loss.item():.4f}")
````

````python
#lets start 


import torch 
import torch.nn as nn
import torch.optim as optim 
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
````

````python
#loading dataset, dataloaders and tranformers

from torchvision.datasets import MNIST
from torchvision import transforms
from torch.utils.data import DataLoader
````

````python
# Device
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# Transform
transform = transforms.ToTensor()

# Dataset
train_data = MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform
)

test_data = MNIST(
    root="data",
    train=False,
    download=True,
    transform=transform
)

# DataLoaders
train_loader = DataLoader(
    train_data,
    batch_size=64,
    shuffle=True
)

test_loader = DataLoader(
    test_data,
    batch_size=64,
    shuffle=False
)






# ==============================================================================
# Data Preparation Completed now model training and evaluation 
# =============================================================================





# Model
class DigitModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(28 * 28, 128),
            nn.ReLU(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        return self.net(x)


model = DigitModel().to(device)

# Loss Function & Optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training Loop
epochs = 5

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for images, labels in train_loader:

        images = images.view(images.size(0), -1).to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = loss_fn(outputs, labels)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch + 1}, Loss = {total_loss:.4f}")

# Evaluation
model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        images = images.view(images.size(0), -1).to(device)
        labels = labels.to(device)

        outputs = model(images)

        predictions = outputs.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

accuracy = 100 * correct / total

print(f"Total Accuracy: {accuracy:.2f}%")

# Save Model
torch.save(model.state_dict(), "mnist_model.pth")
print("Model saved as mnist_model.pth")

# Test One Image
index = 90

image, true_label = test_data[index]

plt.imshow(image.squeeze(), cmap="gray")
plt.title(f"Actual Label: {true_label}")
plt.axis("off")
plt.show()

image_flat = image.view(1, -1).to(device)

with torch.no_grad():

    output = model(image_flat)

    predicted_label = output.argmax(dim=1).item()

print("User picked image index:", index)
print("Actual Label:", true_label)
print("Model Prediction:", predicted_label)
````

# <span style="color:#2563EB; font-family:Georgia;">Convolutional Neural Networks (CNNs)</span>

---

## <span style="color:#7C3AED; font-family:Trebuchet MS;">What are CNNs?</span>

A **<span style="color:#2563EB;">Convolutional Neural Network (CNN)</span>** is a type of ***<span style="color:#DC2626;">Deep Learning</span>*** **<span style="color:#16A34A;">neural network</span>** that is **<span style="color:#EA580C;">specially designed</span>** to understand **<span style="color:#EC4899;">images</span>**.

---

## <span style="color:#059669; font-family:Georgia;">How does a CNN work?</span>

Instead of looking at the **<span style="color:#DC2626;">entire image</span>** at once, a **<span style="color:#2563EB;">CNN</span>** looks at **<span style="color:#16A34A;">small parts of the image</span>**, learns **<span style="color:#EA580C;">important patterns</span>** *(like* **<span style="color:#7C3AED;">edges</span>**, **<span style="color:#0F766E;">shapes</span>**, **<span style="color:#EC4899;">textures</span>***)*, and **<span style="color:#2563EB;">gradually</span>** understands what the **<span style="color:#DC2626;">image contains</span>**.

---

## <span style="color:#B45309; font-family:Georgia;">Learning Process</span>

**<span style="color:#EC4899;">Image</span>**
&nbsp;&nbsp;→&nbsp;&nbsp;
**<span style="color:#16A34A;">Small Parts (also known as <span style="color:#DC2626;">Filters</span>)</span>**
&nbsp;&nbsp;→&nbsp;&nbsp;
**<span style="color:#EA580C;">Patterns</span>**
&nbsp;&nbsp;→&nbsp;&nbsp;
**<span style="color:#7C3AED;">Features</span>**
&nbsp;&nbsp;→&nbsp;&nbsp;
**<span style="color:#2563EB;">Prediction</span>**

---

### <span style="color:#B91C1C;">Key Idea</span>

> A **<span style="color:#2563EB;">CNN</span>** does **<span style="color:#DC2626;">not</span>** analyze the **<span style="color:#DC2626;">whole image at once</span>**.  
> It first learns **<span style="color:#16A34A;">small features</span>** such as **<span style="color:#7C3AED;">edges</span>**, **<span style="color:#0F766E;">shapes</span>**, and **<span style="color:#EC4899;">textures</span>**, then combines them to understand the **<span style="color:#EA580C;">complete image</span>**.

# <span style="color:#2563EB; font-family:Georgia;">Convolutional Neural Networks (CNNs)</span>

---

# <span style="color:#7C3AED; font-family:Georgia;">`nn.Conv2d(in_channels, out_channels, kernel_size, stride, padding)`</span>

---

## <span style="color:#059669;">Syntax</span>

```python
nn.Conv2d(in_channels, out_channels, kernel_size, stride, padding)
```

---

## <span style="color:#2563EB;">`in_channels`</span>

**Input Channels**

- **<span style="color:#16A34A;">Black & White</span>** → **1**
- **<span style="color:#DC2626;">Color (RGB)</span>** → **3**

**Input Shape**

```text
[batch, channels, height, width]

[1, 1, 28, 28]
```

---

## <span style="color:#7C3AED;">`out_channels`</span>

Each **<span style="color:#EA580C;">output channel</span>** = **<span style="color:#DC2626;">1 pattern detector</span>**

Example:

```text
8
```

```text
Input
[1, 1, 28, 28]

↓

Output
[1, 8, ?, ?]
```

---

## <span style="color:#059669;">`kernel_size`</span>

- **3** → **<span style="color:#2563EB;">3 × 3</span>** pixels
- **5** → **<span style="color:#7C3AED;">5 × 5</span>** pixels

---

## <span style="color:#EA580C;">`stride`</span>

Filter **<span style="color:#DC2626;">moves</span>** by **<span style="color:#2563EB;">1</span>** or **<span style="color:#7C3AED;">2</span>** steps.

```text
Stride = 1  →  Moves 1 step

Stride = 2  →  Moves 2 steps
```

---

## <span style="color:#DC2626;">`padding`</span>

Padding decides whether the filter can **<span style="color:#2563EB;">cover the edges</span>** of the image.

  - ➜ Filter **<span style="color:#DC2626;">cannot fully sit on edges</span>**
  - ➜ **<span style="color:#EA580C;">Image shrinks</span>**

- **`padding = 1`**
  - ➜ Adds a **<span style="color:#16A34A;">1-pixel border</span>**
  - ➜ Helps preserve **<span style="color:#7C3AED;">edge information</span>**

# <span style="color:#2563EB; font-family:Georgia;">Pooling</span>

---

## <span style="color:#7C3AED; font-family:Georgia;">What is Pooling?</span>

Pooling is a **<span style="color:#2563EB;">downsampling operation</span>** that **<span style="color:#16A34A;">reduces the spatial size</span>** *(height & width)* of a **<span style="color:#EA580C;">feature map</span>** while keeping the **<span style="color:#DC2626;">most important information</span>**.

---

## <span style="color:#059669;">Why do we use Pooling?</span>

- ➜ **<span style="color:#2563EB;">Reduce</span>** image size
- ➜ **<span style="color:#16A34A;">Reduce</span>** computation
- ➜ **<span style="color:#EA580C;">Reduce</span>** overfitting
- ➜ Keep **<span style="color:#7C3AED;">important features</span>**

---

## <span style="color:#B45309;">How it works</span>

```text
Feature Map
      ↓
Pooling Layer
      ↓
Smaller Feature Map
```

---

## <span style="color:#DC2626;">Types of Pooling</span>

### <span style="color:#2563EB;">1. Max Pooling</span>

Chooses the **<span style="color:#16A34A;">largest value</span>** from each pooling window.

### <span style="color:#7C3AED;">2. Average Pooling</span>

Calculates the **<span style="color:#EA580C;">average value</span>** from each pooling window.

---

### <span style="color:#B91C1C;">Key Idea</span>

> **<span style="color:#2563EB;">Pooling</span>** makes the **<span style="color:#16A34A;">feature map smaller</span>** while preserving the **<span style="color:#DC2626;">most useful information</span>** for the CNN.

</h2>

<p>

<strong>Dropout</strong> is a <strong>regularization technique</strong> used in neural
networks to reduce <strong>overfitting</strong>.

During training, Dropout <strong>randomly sets a percentage of neuron activations
to zero</strong> during each forward pass.

</p>

<div style="background:#F8FAFC; border-left:5px solid #7C3AED; padding:12px 16px; border-radius:6px;">

<strong>Example:</strong><br><br>

If Dropout = <strong>20%</strong>, approximately 20% of the neuron activations are
randomly turned off during a training forward pass.

</div>

<h3 style="font-size:23px; color:#374151; margin-top:20px;">
    <strong>How Dropout Works</strong>
</h3>

<ol style="line-height:1.9;">

<li>Usually applied to the <strong>hidden layers</strong>.</li>

<li>Often placed <strong>after an activation function such as ReLU</strong>.</li>

<li>Randomly turns off a percentage of neuron <strong>activations</strong> during each training forward pass.</li>

<li>This prevents neurons from becoming <strong>too dependent on each other</strong>.</li>

<li>As a result, Dropout provides a <strong>regularization effect</strong> and helps reduce overfitting.</li>

</ol>

<h3 style="font-size:23px; color:#166534; margin-top:20px;">
    <strong>During Evaluation</strong>
</h3>

<div style="border:1px solid #D1D5DB; padding:12px 16px; border-radius:7px;">

Dropout is <strong>not used during evaluation or inference</strong>. All neurons
remain active.

</div>

<h3 style="font-size:23px; color:#991B1B; margin-top:20px;">
    <strong>Why Do We Use Dropout?</strong>
</h3>

<p>

Dropout helps the network learn <strong>more robust features</strong> instead of
memorizing the training data, thereby reducing <strong>overfitting</strong>.

</p>

### 2nd Neural Network Optimisation Technique : Batch Norm 




<h2 style="font-size:30px; color:#0F766E; border-bottom:3px solid #CCFBF1; padding-bottom:7px;">
    <strong>Batch Normalization</strong>
</h2>

<p>

<strong>Batch Normalization (Batch Norm)</strong> is a technique used to
<strong>normalize the activations</strong> of a neural network during training.
It helps make training more <strong>stable and faster</strong>.

</p>

<div style="background:#F0FDFA; border-left:5px solid #0D9488; padding:12px 16px; border-radius:6px;">

<strong>Basic idea:</strong><br><br>

Normalize the output of a layer before passing it to the
<strong>activation function</strong>.

</div>

<h3 style="font-size:23px; color:#374151; margin-top:20px;">
    <strong>How Batch Normalization Works</strong>
</h3>

<ol style="line-height:1.9;">

<li>
<strong>Applied to Hidden Layers</strong> – Typically applied to hidden layers,
not the output layer.
</li>

<li>
<strong>Position</strong> – Usually applied after a <code>Linear</code> or
convolutional layer and before the activation function.
</li>

<li>
<strong>Normalizes Activations</strong> – Computes the <strong>mean</strong> and
<strong>variance</strong> of activations within a mini-batch and uses them to
normalize the activations.
</li>

<li>
<strong>Learnable Parameters</strong> – Introduces two learnable parameters:
<strong>γ (gamma)</strong> for scaling and <strong>β (beta)</strong> for shifting.
</li>

<li>
<strong>Improves Training Stability</strong> – Helps stabilize the training
process and can allow the use of <strong>higher learning rates</strong>.
</li>

<li>
<strong>Regularization Effect</strong> – The use of mini-batch statistics can
provide a small <strong>regularization effect</strong>.
</li>

</ol>

<h3 style="font-size:23px; color:#7C2D12; margin-top:20px;">
    <strong>Typical Order</strong>
</h3>

<div style="text-align:center; background:#FFF7ED; border:1px solid #FED7AA; padding:14px; border-radius:7px; font-size:17px;">

<strong>Linear Layer → Batch Normalization → Activation Function</strong>

</div>

````python
#code example 

class MyNN(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        self.model = nn.Sequential(
            nn.Linear(num_features, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(p=0.3),

            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(p=0.3),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        return self.model(x)
````

#### 3rd Neural Network Optimisation Technique : Regularization 




<h2 style="font-size:30px; color:#B45309; border-bottom:3px solid #FEF3C7; padding-bottom:7px;">
    <strong>L2 Regularization</strong>
</h2>

<p>

<strong>L2 Regularization</strong> is a regularization technique that penalizes
<strong>large model weights</strong> and encourages smaller, more
<strong>generalizable weights</strong>.

</p>

<h3 style="font-size:23px; color:#374151;">
    <strong>Applied to Model Weights</strong>
</h3>

<p>
Regularization is applied to the <strong>weights</strong> of the model to
discourage excessively large values.
</p>

<h3 style="font-size:23px; color:#374151;">
    <strong>Penalty Term</strong>
</h3>

<p>
In L2 regularization, a penalty term is added to the original loss:
</p>

$$
\mathrm{Loss}_{reg}
=
\mathrm{Loss}_{original}
+
\lambda\sum_i w_i^2
$$

<p>

Here, <strong>λ (lambda)</strong> controls the strength of the regularization.

</p>

<h3 style="font-size:23px; color:#374151;">
    <strong>Weight Decay</strong>
</h3>

<p>

In weight decay, the gradient update includes an additional term that
<strong>shrinks the weights during training</strong>.

</p>

$$
w \leftarrow w-\eta(\nabla\mathrm{Loss}+\lambda w)
$$

<h3 style="font-size:23px; color:#374151;">
    <strong>Why Use L2 Regularization?</strong>
</h3>

<ul style="line-height:1.9;">

<li>Penalizes <strong>large weights</strong>.</li>

<li>Encourages learning to be distributed across <strong>multiple parameters</strong>.</li>

<li>Reduces reliance on a few very large weights.</li>

<li>Helps reduce <strong>overfitting</strong>.</li>

<li>Improves the model's ability to <strong>generalize to unseen data</strong>.</li>

</ul>

````python
#l2 regularization applied on optimisers

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.1, weight_decay=1e-4)
````
