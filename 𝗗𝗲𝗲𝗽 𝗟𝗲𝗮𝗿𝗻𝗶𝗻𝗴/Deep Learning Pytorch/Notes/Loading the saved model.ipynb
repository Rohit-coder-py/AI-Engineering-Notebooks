```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.Linear(x)



# No need to copy the optimizer, loss function, training data, or training loop. The model has already been trained.
```

```python
model = MyModel()

model.load_state_dict(
    torch.load(
        r"C:\Users\shobh\Downloads\AI and DATA SCIENCE\𝗗𝗲𝗲𝗽 𝗟𝗲𝗮𝗿𝗻𝗶𝗻𝗴\dl_models\first_dl_model.pth"
    )
)

model.eval()
```

```python
#now lets predict it on new unseen data 

x = torch.tensor([[5.0]])

with torch.no_grad():
    prediction = model(x)

print(prediction.item())
```
