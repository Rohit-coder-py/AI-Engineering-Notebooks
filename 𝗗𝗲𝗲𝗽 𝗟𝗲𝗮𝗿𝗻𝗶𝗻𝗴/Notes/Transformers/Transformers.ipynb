<strong style="color:#1F618D;">Transformers</strong> are a type of deep learning neural network architecture designed to process sequential data, such as text, by using an <strong style="color:#884EA0;">attention mechanism</strong> to understand relationships between different elements of the input.
</div>

They were introduced in the 2017 research paper *Attention Is All You Need*.

**Transformers are widely used in modern AI applications, including:**

- Chatbots and conversational AI
- Machine translation
- Text summarization
- Sentiment analysis
- Text generation
- Large Language Models (LLMs)

<h2 style="color:#B03A2E; border-bottom:2px solid #F1948A; padding-bottom:7px;">
2. Why Were Transformers Introduced?
</h2>

Before Transformers, recurrent neural networks (**RNNs**), including **LSTMs** and **GRUs**, were commonly used for sequence-processing tasks.

<blockquote style="
border-left: 4px solid #E67E22;
padding: 10px 16px;
background: #FDF2E9;
color: #784212;
font-size: 16px;
">
"The animal didn't cross the street because it was tired."
</blockquote>

To understand this sentence, a model needs to determine what the word **"it"** refers to.

- **RNN:** Processes words sequentially, maintaining information in its hidden state as it moves through the sentence.
- **Transformer:** Uses attention to let the model directly examine relationships between words, helping it determine which words are relevant to one another.

<b>Researchers didn't invent Transformers because RNNs, LSTMs, or GRUs were useless. They invented them because existing architectures had limitations that became increasingly difficult to overcome, especially when processing long sequences and training on large amounts of data.

<b>To understand why Transformers were invented, let's look at how neural network architectures evolved.

| Architecture | What it was designed to do                                              | Main limitation                                                                                                      |
| ------------ | ----------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------- |
| ANN          | Learn patterns from data using interconnected neurons.                  | Doesn't inherently model sequential relationships.                                                                   |
| CNN          | Learn spatial patterns, particularly in images.                         | Ordinary CNNs don't naturally provide unrestricted, direct interactions between distant sequence positions.          |
| RNN          | Process sequential data while carrying information from previous steps. | Sequential computation makes training difficult to parallelize; long-range information can be difficult to preserve. |
| LSTM         | Improve RNN memory using gates and a cell state.                        | Still processes sequences sequentially, and long-range dependencies can remain challenging.                          |
| GRU          | Provide a simpler gated alternative to LSTM.                            | Also retains sequential computation.                                                                                 |
| Transformer  | Model relationships between sequence elements using attention.          | Attention can be computationally and memory intensive for long sequences.                                            |

# Why Were Transformers Invented?
### The real evolution from ANN, CNN, RNN, LSTM and GRU to Transformers

**Deep Learning · NLP · Architecture Evolution**

</div>

---

> **Core idea:** Researchers did not invent Transformers because earlier neural networks were useless. They introduced them because sequence models—especially recurrent models—had limitations that became increasingly important for large-scale language processing and training.

## 1. The journey before Transformers

| Architecture | What it was designed to do | Important limitation |
|---|---|---|
| **ANN** | Learn patterns through interconnected neurons. | Does not inherently model sequential relationships. |
| **CNN** | Learn spatial patterns, especially in images. | Ordinary CNNs do not naturally provide unrestricted direct interactions between distant sequence positions. |
| **RNN** | Process sequences while carrying information from earlier steps. | Sequential computation limits parallelization; preserving long-range information can be difficult. |
| **LSTM** | Improve recurrent memory using gates and a cell state. | Still processes sequences recurrently; long-range dependencies can remain challenging. |
| **GRU** | Offer a simpler gated recurrent alternative to LSTM. | Still retains sequential computation. |
| **Transformer** | Learn relationships between sequence elements using attention. | Standard full self-attention can be memory- and computation-intensive for long sequences. |

<div style="border-left: 5px solid #16a34a; padding: 10px 14px; background: #f0fdf4;">

**Green highlight — Important:** These architectures were not simply replacing one another. They address different kinds of problems, and many remain useful today. CNNs, RNNs, LSTMs and GRUs have not become universally obsolete.

</div>

## 2. The language problem that motivated better sequence models

Imagine translating:

> The animal didn't cross the street because **it** was too tired.

To translate the sentence correctly, a model needs to infer that *it* likely refers to the animal.

Now consider:

> The animal didn't cross the street because **it** was too wide.

Here, *it* refers to the street.

A language model must learn relationships between words. The meaning of a word can depend on other words elsewhere in the sentence. This is one form of a **long-range dependency**.

### How an RNN processes a sentence

```text
The → animal → didn't → cross → the → street → because → it → was → too → wide
```

At each time step, an RNN updates its hidden state:

\[
h_t = f(x_t, h_{t-1})
\]

Where:

- \(x_t\): representation of the current token.
- \(h_{t-1}\): hidden state from the previous time step.
- \(h_t\): updated hidden state.

The hidden state carries information forward through the sequence.

<div style="border-left: 5px solid #dc2626; padding: 10px 14px; background: #fef2f2;">

**Red highlight — Limitation:** Information from earlier tokens must pass through recurrent steps. As sequences become long, learning and preserving distant relationships can be difficult.

</div>

## 3. Why LSTM and GRU were not the final answer

LSTM and GRU improved recurrent networks by using gates to regulate information flow.

- **LSTM:** Uses input, forget and output gates, along with a cell state.
- **GRU:** Uses update and reset gates in a more compact recurrent design.

These improvements helped with learning longer-term dependencies. But both architectures still rely on recurrence.

```text
RNN / LSTM / GRU:

Time step 1 → Time step 2 → Time step 3 → Time step 4
```

The next recurrent state depends on the previous one. That limits parallel computation across time steps during training.

<div style="border-left: 5px solid #dc2626; padding: 10px 14px; background: #fef2f2;">

**Red highlight — Bottleneck:** GPUs are good at parallel computation, but the dependency between recurrent steps restricts how much of a sequence can be processed simultaneously.

</div>

Researchers wanted a way to model relationships across tokens without requiring the same recurrent chain.

## 4. Attention: a crucial step toward Transformers

**Attention existed before the Transformer.** Earlier sequence-to-sequence systems used attention to help a decoder focus on relevant parts of an input, rather than relying only on a single compressed representation.

For example, while generating a translated word, a model could assign more attention to the source words relevant to that word.

Researchers then explored a larger idea:

> What if attention became the central mechanism of the architecture, rather than an additional component attached to a recurrent network?

That idea helped lead to the Transformer.

## 5. 2017 — The Transformer arrives

The paper **“Attention Is All You Need”** was published in 2017. Its eight authors were Ashish Vaswani, Noam Shazeer, Niki Parmar, Jakob Uszkoreit, Llion Jones, Aidan N. Gomez, Łukasz Kaiser and Illia Polosukhin.

The paper introduced an attention-based encoder–decoder architecture for sequence transduction, including machine translation. Its main sequence-processing structure dispensed with recurrence and convolution.

### The central change

**RNN-style processing:**

```text
Token 1 → Token 2 → Token 3 → Token 4 → Token 5
```

**Self-attention (simplified illustration):**

```text
Token 1 ↔ Token 2 ↔ Token 3 ↔ Token 4 ↔ Token 5
```

In a standard full self-attention layer, each token can incorporate information from other permitted positions. The model learns which relationships are useful during training.

For example, the representation of **“it”** can attend to **“street”** in the sentence about the street being too wide.

<div style="border-left: 5px solid #16a34a; padding: 10px 14px; background: #f0fdf4;">

**Green highlight — Breakthrough:** Self-attention provides direct connections between sequence positions, while the Transformer can process positions in parallel within a layer.

</div>

## 6. Why self-attention mattered

| Property | RNN | Transformer with self-attention |
|---|---|---|
| Sequence computation | Recurrent dependency between time steps | Token positions can be processed in parallel within a layer |
| Distant relationships | Information passes through recurrent steps | Direct attention connections can link distant positions |
| Long-range dependencies | May be difficult to learn | Shorter information paths can help |
| Long-sequence cost | Depends on recurrent computation and model design | Standard full self-attention has quadratic sequence-length cost |

**Important qualification:** Transformers are not automatically faster or cheaper for every task. Standard full self-attention has computation and memory costs that grow approximately quadratically with sequence length. RNNs can still be useful for some streaming or sequential applications.

## 7. What happened after the Transformer?

The original Transformer targeted sequence-to-sequence tasks. Researchers subsequently adapted Transformer components and training objectives to many other tasks.

| Year | Development | Significance |
|---|---|---|
| **2017** | Original Transformer | Introduced an attention-centered sequence-transduction architecture. |
| **2018** | BERT | Demonstrated the effectiveness of Transformer encoders for language representations. |
| **2018 onward** | GPT models | Developed decoder-only Transformer language models for autoregressive text generation. |
| **2020** | Vision Transformer (ViT) | Applied Transformer processing to image patches for image recognition. |
| **2020s** | Large-scale generative AI systems | Transformer-based models became central to many widely used language and multimodal systems. |

These models do not all share an identical architecture or training procedure. They adapt Transformer designs to different goals.

## 8. Why Transformers became important for Generative AI

A language model learns patterns in text, including grammar, context and relationships between tokens. Transformer architectures provide mechanisms for learning these relationships, and their parallelizable training structure makes them suitable for large-scale training.

### Autoregressive generation in a decoder-only model

```text
Context: Artificial intelligence is
                         ↓
                 Predict next token
                         ↓
                       changing
                         ↓
          Append token and predict again
```

During training, an autoregressive model learns to predict the next token from preceding context. At generation time, it repeatedly predicts and appends tokens.

<div style="border-left: 5px solid #16a34a; padding: 10px 14px; background: #f0fdf4;">

**Green highlight — Connection to AI Engineering:** Transformers are an important foundation of modern generative language models, but Generative AI is broader than Transformers alone. Not every generative model uses a Transformer.

</div>

## 9. Results reported in the original paper

The 2017 paper evaluated the Transformer on machine-translation benchmarks and reported:

| Task / measurement | Reported result |
|---|---|
| English → German | **28.4 BLEU** for the Transformer big model |
| English → French | **41.8 BLEU** for the Transformer big model |
| English → French training | The paper reported training the model for **3.5 days on eight GPUs**. |

The authors reported improved translation results compared with the systems they evaluated, along with greater parallelizability and reduced training time in their comparisons. These are results reported in that paper, not a guarantee that Transformers outperform every other architecture on every task.

## 10. The evolution in one diagram

```text
ANN
 │  Learn patterns from data
 ▼
CNN
 │  Learn spatial patterns, especially in images
 ▼
RNN
 │  Process sequential information using recurrent state
 ▼
LSTM / GRU
 │  Improve information flow and recurrent memory
 ▼
Attention
 │  Focus on relevant parts of the input
 ▼
Transformer (2017)
 │  Make attention the core sequence-processing mechanism
 ▼
BERT / GPT / ViT and other Transformer models
    Adapt the architecture to language understanding,
    text generation, vision and other tasks
```

> **Takeaway:** The Transformer was not simply “the next neural network after GRU.” It was a different architectural approach that made attention central and reduced the need for sequential recurrence. That combination proved particularly useful for many language tasks and large-scale training.

## 11. Key distinctions to remember

- **ANN, CNN, RNN, LSTM, GRU and Transformer:** neural-network architectures or architectural families.
- **BERT and GPT:** model families built using Transformer components, with different designs and objectives.
- **Generative AI:** a broad field that includes multiple model families and techniques; Transformers are one major foundation.
- **Architecture choice:** depends on the task, data, latency, compute budget and deployment requirements.

|---|---|

- Input and output sequences can have **different lengths**.
- It can learn to transform one sequence into another.
- It is useful for tasks such as **machine translation, text summarization, and conversational systems**.

```text
Input Sequence
      ↓
   Encoder
      ↓
Reads the sequence
      ↓
Understands the context
      ↓
Context Vector
      ↓
   Decoder
      ↓
Output Sequence

<b> <mark>2. Decoders 

<h2 style="font-family:Georgia, serif; color:#16a34a; border-bottom:3px solid #86efac; padding-bottom:6px;">
Decoder
</h2>

<p style="font-size:16px; line-height:1.7;">
The <b style="color:#2563eb;">Decoder</b> generates the output sequence <b>one element at a time</b>.
</p>

---

### Training vs Prediction

The decoder works differently during the <b style="color:#dc2626;">training</b> and <b style="color:#7c3aed;">prediction</b> phases.

#### 1. Training Phase — <mark>Teacher Forcing</mark>

> During training, the decoder already knows the <b>correct output</b>.

At every time step, the <b style="color:#16a34a;">correct previous output</b> is given to the decoder to help it predict the next word.

```text
Context Vector
      ↓
   Decoder
      ↓
Correct Output → Next Prediction
      ↓
Correct Output → Next Prediction
      ↓
Correct Output → Next Prediction

<h3 style="font-family:Georgia, serif; color:#dc2626; border-left:4px solid #dc2626; padding-left:10px;">
Prediction Phase
</h3>

> During <b style="color:#dc2626;">prediction</b>, the decoder does <b>not know the correct output</b>.

- It uses its <b style="color:#2563eb;">own previous prediction</b> as the input for generating the next output.

```text
Previous Prediction
        ↓
     Decoder
        ↓
   Next Prediction
        ↓
     Decoder
        ↓
   Next Prediction
        ↓
       ...

<mark><b>How Decoders Work?

<b>We already have <mark>Context Vector </mark> we generated it from Encoders 

<div align="center">

# <b><u>TRANSFORMER DECODER</u></b>
### <mark>How Decoders Work</mark>

</div>

---

> **Core Idea:** A **Decoder** is responsible for **generating the output sequence**. It processes previously generated tokens and predicts the **next token**, one step at a time.

---

## <b><u>1. What is a Decoder?</u></b>

A **Transformer Decoder** is the part of a Transformer architecture that is responsible for **generating output**.

For example, in machine translation:

```text
Input Sentence
      ↓
   Encoder
      ↓
Contextual Representation
      ↓
   Decoder
      ↓
Output Sentence
```

Example:

```text
Input:
"मैं एक छात्र हूँ"

        ↓ Encoder

Encoded Representation

        ↓ Decoder

"I am a student"
```

The Decoder generates the output progressively:

```text
<BOS> → I → am → a → student → <EOS>
```

Where:

- **BOS** = Beginning Of Sequence
- **EOS** = End Of Sequence

---

## <b><u>2. Basic Decoder Architecture</u></b>

A simplified Transformer Decoder can be represented as:

```text
             Previous Output Tokens
                       ↓
                Token Embedding
                       ↓
              Positional Encoding
                       ↓
        ┌──────────────────────────┐
        │ Masked Self-Attention    │
        └────────────┬─────────────┘
                     ↓
        ┌──────────────────────────┐
        │ Cross-Attention          │ ← Encoder Output
        └────────────┬─────────────┘
                     ↓
        ┌──────────────────────────┐
        │ Feed Forward Network     │
        └────────────┬─────────────┘
                     ↓
                 Linear Layer
                     ↓
                  Softmax
                     ↓
               Next Token
```

> <mark><b>Important:</b></mark> In the **original Encoder–Decoder Transformer**, the Decoder contains both **Masked Self-Attention** and **Cross-Attention**.

---

## <b><u>3. Step 1 — Input to the Decoder</u></b>

The Decoder receives the tokens that have already been generated.

Suppose the desired output is:

```text
I am a student
```

The generation starts with:

```text
<BOS>
```

The Decoder predicts:

```text
I
```

Then the input becomes:

```text
<BOS> I
```

It predicts:

```text
am
```

Then:

```text
<BOS> I am
```

It predicts:

```text
a
```

And so on.

### Generation Flow

```text
<BOS>
  ↓
<BOS> I
  ↓
<BOS> I am
  ↓
<BOS> I am a
  ↓
<BOS> I am a student
  ↓
<EOS>
```

> <mark><b>Key Point:</b></mark> The Decoder generates the output **step by step**, rather than simply producing a sentence as a single sequential operation.

---

## <b><u>4. Step 2 — Token Embedding</u></b>

The input tokens are converted into numerical vectors using an **Embedding Layer**.

Example:

```text
"I"       → [0.21, 0.54, 0.12, ...]
"am"      → [0.72, 0.11, 0.83, ...]
"student" → [0.45, 0.92, 0.31, ...]
```

These vectors are called **token embeddings**.

```text
Tokens
  ↓
Embedding Layer
  ↓
Numerical Vectors
```

The embedding allows the neural network to work with numerical representations instead of raw words.

---

## <b><u>5. Step 3 — Positional Information</u></b>

Transformers do not naturally understand the **order of tokens**.

For example:

```text
I am a student
```

contains positional information:

```text
I        → Position 1
am       → Position 2
a        → Position 3
student  → Position 4
```

Therefore, the model adds **positional information** to token embeddings.

```text
Token Embedding
       +
Positional Information
       ↓
Decoder Input
```

> <mark><b>Important:</b></mark> Position information helps the Transformer understand **where each token occurs in the sequence**.

---

# <b><u>6. Step 4 — Masked Self-Attention</u></b>

This is one of the **most important components** of a Transformer Decoder.

The Decoder must not look at **future tokens** while predicting the current token.

Suppose the target sentence is:

```text
I am a student
```

When predicting `am`, the model can see:

```text
I
```

but it cannot see:

```text
a student
```

### Causal Mask

```text
          I    am    a    student

I         ✓    ✗     ✗      ✗
am        ✓    ✓     ✗      ✗
a         ✓    ✓     ✓      ✗
student   ✓    ✓     ✓      ✓
```

The `✗` positions represent **masked future tokens**.

---

## <b><u>Why is Masking Necessary?</u></b>

Without masking, the model could **cheat** during training.

For example, if the model is supposed to predict:

```text
student
```

it should not already have access to the actual word:

```text
student
```

Therefore:

> <mark><b>Masked Self-Attention prevents the Decoder from using future tokens.</b></mark>

This is also called:

- **Causal Attention**
- **Masked Attention**
- **Autoregressive Attention**

---

# <b><u>7. Self-Attention vs Cross-Attention</u></b>

These two mechanisms are very important to understand.

### <b>Self-Attention</b>

The Decoder attends to its **own previous tokens**.

```text
Decoder Tokens
      ↕
Decoder Tokens
```

It answers:

> **"What previously generated information is important for predicting the next token?"**

---

### <b>Cross-Attention</b>

The Decoder attends to the **Encoder's output**.

```text
Decoder
   │
   │ Query
   ↓
Attention
   ↑
   │ Key + Value
   │
Encoder
```

It answers:

> **"Which information from the input is important for generating the current output?"**

---

# <b><u>8. Step 5 — Cross-Attention</u></b>

In the original Transformer, the Decoder receives information from the Encoder through **Cross-Attention**.

```text
Input
  ↓
Encoder
  ↓
Encoder Output
  ↓
Cross-Attention
  ↑
Decoder
```

The Decoder uses the Encoder's representation to understand the **input context**.

For example, in translation:

```text
English:
"I am a student"

       ↓ Encoder

Contextual Representation

       ↓ Cross-Attention

Decoder

       ↓

Another Language
```

---

## <b><u>Query, Key and Value in Cross-Attention</u></b>

In cross-attention:

```text
Query → Decoder
Key   → Encoder
Value → Encoder
```

Conceptually:

```text
                Decoder
                   │
                   │ Query
                   ↓
             Cross-Attention
                   ↑
                   │ Key + Value
                   │
                Encoder
```

The Decoder asks:

> **"Which parts of the Encoder's representation are relevant to what I am currently generating?"**

---

# <b><u>9. Step 6 — Feed Forward Network</u></b>

After the attention operations, the representation passes through a **Feed Forward Network (FFN)**.

Simplified:

```text
Attention Output
       ↓
   Linear Layer
       ↓
   Activation
       ↓
   Linear Layer
       ↓
      Output
```

Mathematically:

```text
FFN(x) = Linear₂(Activation(Linear₁(x)))
```

The FFN adds additional **non-linear transformation** to the token representations.

> <mark><b>Important:</b></mark> Attention helps tokens interact with each other, while the **Feed Forward Network transforms the resulting representations**.

---

# <b><u>10. Step 7 — Linear Layer</u></b>

After passing through the Decoder layers, the final hidden representation is converted into scores for every token in the vocabulary.

Suppose the vocabulary contains:

```text
["I", "am", "a", "student", "teacher", "cat", ...]
```

The Linear layer may produce:

```text
I        → 2.1
am       → 4.7
a        → 1.8
student  → 6.2
teacher  → 2.3
cat      → 0.4
```

These values are called **logits**.

> <mark><b>Logits are raw prediction scores before Softmax.</b></mark>

---

# <b><u>11. Step 8 — Softmax</u></b>

The logits are converted into **probabilities** using Softmax.

Example:

```text
I        → 0.05
am       → 0.15
a        → 0.03
student  → 0.70
teacher  → 0.06
cat      → 0.01
```

The probabilities represent how likely each token is to be selected.

The model can then select a token according to the chosen **decoding strategy**.

For a simple example:

```text
Highest Probability
        ↓
    "student"
```

---

# <b><u>13. Complete Decoder Flow</u></b>

```text
              Previous Tokens
                    ↓
              Token Embedding
                    ↓
            Positional Information
                    ↓
        ┌──────────────────────────┐
        │  Masked Self-Attention   │
        └────────────┬─────────────┘
                     ↓
        ┌──────────────────────────┐
        │     Cross-Attention      │
        │                          │
        │ ← Encoder Output        │
        └────────────┬─────────────┘
                     ↓
        ┌──────────────────────────┐
        │ Feed Forward Network     │
        └────────────┬─────────────┘
                     ↓
                 Linear
                     ↓
                  Logits
                     ↓
                 Softmax
                     ↓
              Next Token
                     ↓
              Add Token
                     ↓
                  Repeat
```

# <b><u>14. Decoder Layer — Conceptual Structure</u></b>

A Transformer Decoder layer can be viewed as:

```text
Input
  ↓
Masked Multi-Head Self-Attention
  ↓
Add & Layer Normalization
  ↓
Cross-Attention
  ↓
Add & Layer Normalization
  ↓
Feed Forward Network
  ↓
Add & Layer Normalization
  ↓
Output
```

A Transformer usually stacks **multiple Decoder layers**.

```text
Decoder Layer 1
      ↓
Decoder Layer 2
      ↓
Decoder Layer 3
      ↓
      ...
      ↓
Decoder Layer N
```

> <mark><b>More layers allow the model to learn increasingly complex representations.</b></mark>

# <b><u>15. Decoder vs Encoder</u></b>

| Feature | <b>Encoder</b> | <b>Decoder</b> |
|---|---|---|
| Main Purpose | Understand input | Generate output |
| Self-Attention | Yes | Yes |
| Causal Mask | Usually No | Yes |
| Cross-Attention | No | Yes, in Encoder–Decoder models |
| Future Tokens | Can access input sequence | Cannot access future output tokens |
| Generation | No | Yes |
| Example | BERT | GPT-style Decoder |

# <b><u>16. Original Transformer vs GPT</u></b>

This distinction is **very important**.

## Original Transformer

The original Transformer architecture uses:

```text
Encoder + Decoder
```

```text
Input
  ↓
Encoder
  ↓
Encoder Output
  ↓
Decoder
  ↓
Output
```

This architecture is useful for tasks such as **machine translation**.

---

## GPT

GPT uses a **Decoder-only Transformer architecture**.

```text
Input Tokens
      ↓
Masked Self-Attention
      ↓
Feed Forward Network
      ↓
Next Token Prediction
```

There is **no separate Encoder** in GPT-style decoder-only models.

> <mark><b>Important:</b></mark> When we say **"GPT uses a Transformer Decoder"**, we mean a **Decoder-only Transformer architecture**.

---

# <b><u>17. How GPT Generates Text</u></b>

Suppose the prompt is:

```text
The capital of France is
```

The model calculates probabilities for the next token:

```text
Paris     → 0.92
London    → 0.02
Berlin    → 0.01
Madrid    → 0.01
...
```

The model selects a token:

```text
Paris
```

Now the sequence becomes:

```text
The capital of France is Paris
```

The model then predicts the **next token**.

This process continues:

```text
Prompt
  ↓
Predict Next Token
  ↓
Add Token
  ↓
Predict Next Token
  ↓
Add Token
  ↓
Repeat
```

This is called **Autoregressive Generation**.

---

# <b><u>18. Autoregressive Generation</u></b>

The Decoder predicts:

$$
P(x_t \mid x_1, x_2, ..., x_{t-1})
$$

This means:

> **The probability of the current token depends on the previously generated tokens.**

Example:

```text
"The weather is"
       ↓
Predict next token
       ↓
"beautiful"
       ↓
"The weather is beautiful"
       ↓
Predict next token
       ↓
"today"
```

So:

```text
Previous Tokens
      ↓
Next Token
      ↓
Previous + New Token
      ↓
Next Token
      ↓
Repeat
```

---

# <b><u>19. Training vs Inference</u></b>

There is an important difference between **training** and **generation/inference**.

## During Training

The model can process the complete target sequence at once because the **causal mask** prevents each position from seeing future tokens.

Example:

```text
Input:

I     am     a     student
↓     ↓      ↓       ↓
am    a    student   <EOS>
```

The model learns multiple next-token predictions in parallel.

---

## During Inference

The model generates tokens **step by step**:

```text
<BOS>
 ↓
I
 ↓
am
 ↓
a
 ↓
student
 ↓
<EOS>
```

> <mark><b>Training can be parallelized, but autoregressive inference generates the output progressively.</b></mark>

---

# <b><u>20. Final Mental Model</u></b>

Think of a Decoder as a system repeatedly asking:

```text
┌─────────────────────────────────────┐
│ What have I generated so far?       │
└──────────────────┬──────────────────┘
                   ↓
          Masked Self-Attention
                   ↓
┌─────────────────────────────────────┐
│ What input information is relevant? │
└──────────────────┬──────────────────┘
                   ↓
             Cross-Attention
                   ↓
┌─────────────────────────────────────┐
│ Process the representation          │
└──────────────────┬──────────────────┘
                   ↓
                  FFN
                   ↓
┌─────────────────────────────────────┐
│ What token should come next?        │
└──────────────────┬──────────────────┘
                   ↓
                Linear
                   ↓
               Softmax
                   ↓
              Next Token
                   ↓
                 Repeat
```

---

# <b><u>21. One-Line Definition</u></b>

> <mark><b>A Transformer Decoder uses masked self-attention to understand previously generated tokens and, in Encoder–Decoder architectures, cross-attention to access encoder information before predicting the next token.</b></mark>

---

# <b><u>22. Quick Revision</u></b>

```text
Decoder
   ↓
Token Embedding
   ↓
Positional Information
   ↓
Masked Self-Attention
   ↓
Cross-Attention
   ↓
Feed Forward Network
   ↓
Linear Layer
   ↓
Logits
   ↓
Softmax
   ↓
Next Token
   ↓
Repeat
```

### <mark>Remember these 5 concepts</mark>

1. **Masked Self-Attention** → Looks at previous tokens, not future tokens.
2. **Cross-Attention** → Connects the Decoder with Encoder output.
3. **Feed Forward Network** → Further transforms token representations.
4. **Softmax** → Converts logits into probabilities.
5. **Autoregressive Generation** → Predicts one token at a time.

<h2 style="font-family:Georgia, serif; color:#dc2626; border-bottom:3px solid #fca5a5; padding-bottom:6px;">
Problems with Encoders and Decoders
</h2>

<ol style="font-size:16px; line-height:2;">

<li>
<b style="color:#dc2626;">Information Bottleneck</b>
<br>
The entire input sequence is compressed into a <b>fixed-size context vector</b>, which can cause important information to be lost.
</li>

<li>
<b style="color:#2563eb;">No Word-to-Word Alignment</b>
<br>
The decoder receives only the <b>context vector</b> instead of directly focusing on the relevant input words.
</li>

<li>
<b style="color:#7c3aed;">Exposure Bias</b>
<br>
During training, <b>teacher forcing</b> provides the correct previous word, but during prediction the decoder uses its <b>own previous prediction</b>.
</li>

<li>
<b style="color:#ea580c;">Long Sentence Failure</b>
<br>
As the input sequence becomes longer, the fixed-size context vector struggles to retain <b>all the important information</b>.
</li>

</ol>

These limitations motivated the development of <b style="color:#16a34a;">Attention Mechanism</b>, which allows the decoder to focus on different parts of the input sequence.
</blockquote>

# 🧠 Attention Mechanism in Transformers

---

## 1. What is the Attention Mechanism?

> **Attention Mechanism** is a technique that allows a neural network to **focus on the most relevant parts of the input** when producing an output.

In simple words:

<mark>**Attention answers the question: "Which parts of the input are important for understanding the current word/token?"**</mark>

### Example

Consider the sentence:

> **"The cat sat on the mat because it was tired."**

When the model processes the word **"it"**, it needs to understand what **"it"** refers to.

The model should give more importance to:

**"cat"**

rather than words such as **"on"**, **"the"**, or **"because"**.

Conceptually:

| Word | Attention |
|---|---:|
| The | 0.02 |
| **cat** | **0.45** |
| sat | 0.08 |
| on | 0.03 |
| the | 0.02 |
| mat | 0.10 |
| because | 0.05 |
| it | 0.10 |
| was | 0.05 |
| tired | 0.10 |

<mark>**Higher attention weight = more influence on the current token's representation.**</mark>

> **Note:** These numbers are only for intuition. Real attention weights are calculated from the model's learned parameters.

---

# 2. Why Do We Need Attention?

Before Transformers, models such as **RNNs, LSTMs and GRUs** were commonly used for sequence data.

For example:

```text
"The cat sat on the mat"

The → cat → sat → on → the → mat
```

The model processes information **sequentially**.

This can make it difficult to maintain relationships between words that are far apart.

### Example

> **"The boy who was playing football near the school forgot his bag."**

The model needs to understand relationships such as:

```text
boy → forgot → bag
```

even though these words are separated by several other tokens.

<mark>**Attention provides a direct way for a token to gather information from other relevant tokens.**</mark>

---

# 3. Core Idea of Attention

The fundamental idea is:

```text
Input tokens
     ↓
Find relevant tokens
     ↓
Calculate their importance
     ↓
Give more weight to important information
     ↓
Combine the information
     ↓
Create a better representation
```

Therefore:

<mark>**Attention = Selective Information Gathering**</mark>

Instead of treating every token equally, the model assigns different weights:

```text
Token A → low importance
Token B → HIGH importance
Token C → medium importance
Token D → low importance
Token E → HIGH importance
```

The weighted information is then combined to create the output.

---

# 4. How Does Attention Know What is Important?

This is where **Query (Q), Key (K), and Value (V)** come into the mechanism.

Each input representation is transformed into three vectors:

```text
Q = Query
K = Key
V = Value
```

They are created using learned weight matrices:

```text
Q = XWQ

K = XWK

V = XWV
```

Where:

- **X** = input representation
- **WQ** = Query weight matrix
- **WK** = Key weight matrix
- **WV** = Value weight matrix

<mark>**Q, K and V are learned projections of the input representation.**</mark>

---

# 5. What Do Q, K and V Mean?

A useful intuition is:

### 🔎 Query — "What am I looking for?"

The **Query** represents what the current token is looking for.

### 🏷️ Key — "What information do I contain?"

The **Key** represents what kind of information a token contains for matching purposes.

### 📦 Value — "What information should I actually provide?"

The **Value** contains the actual information that will be used to construct the output.

### Simple Analogy

Imagine a library:

```text
Query → What book am I looking for?

Key   → Information written on the book's label.

Value → The actual contents of the book.
```

The Query is compared with Keys to determine:

> **"Which information is relevant to me?"**

Then the corresponding Values are collected.

---

# 6. Step 1 — Calculate Attention Scores

Suppose we have:

```text
"The cat sat on the mat"
```

For a particular token, its Query is compared with the Keys of the tokens.

This is generally done using a **dot product**:

```text
Score = Q · K
```

For the entire sequence:

```text
QKᵀ
```

This produces an **attention score matrix**.

Conceptually:

```text
              The     cat     sat     on     mat

The           0.8     1.2     0.3     0.1    0.2
cat           0.4     1.5     0.7     0.2    0.4
sat           0.2     0.8     1.3     0.5    0.3
on            0.1     0.3     0.6     1.0    0.7
mat           0.2     0.5     0.3     0.8    1.4
```

These numbers represent **how strongly tokens relate to one another**.

<mark>**Higher score → stronger relevance.**</mark>

---

# 7. Step 2 — Scaling

The raw scores are divided by:

```text
√dk
```

where:

```text
dk = dimension of the Key vector
```

Therefore:

```text
Scaled Scores = QKᵀ / √dk
```

### Why scale?

When the vector dimension is large, dot products can become large.

Large values can make the Softmax function produce extremely sharp probabilities.

Scaling helps keep the values in a more stable range.

---

# 8. Step 3 — Softmax

The scaled scores are passed through **Softmax**:

```text
Attention Weights =
Softmax(QKᵀ / √dk)
```

Softmax converts the scores into normalized weights.

For example:

```text
Before Softmax:

The  → 1.2
cat  → 3.8
sat  → 0.7
on   → 0.2
mat  → 1.1
```

After Softmax:

```text
The  → 0.06
cat  → 0.62
sat  → 0.04
on   → 0.02
mat  → 0.26
```

The weights sum to approximately 1:

```text
0.06 + 0.62 + 0.04 + 0.02 + 0.26 = 1.00
```

<mark>**The Softmax output gives the attention weights.**</mark>

---

# 9. Step 4 — Multiply Attention Weights by Values

Now the attention weights are multiplied by the corresponding **Value vectors**.

```text
Output = Attention Weights × V
```

More precisely:

```text
Output = Softmax(QKᵀ / √dk)V
```

This produces the final **context-aware representation**.

---

# 10. Complete Attention Formula

## <mark>Scaled Dot-Product Attention</mark>

```text
                 QKᵀ
Attention(Q,K,V) = Softmax(────)V
                    √dk
```

Or:

```text
<mark>Attention(Q,K,V) = Softmax(QKᵀ / √dk)V</mark>
```

### Remember the flow:

```text
Q + K
  ↓
Compare
  ↓
Attention Scores
  ↓
Scale
  ↓
Softmax
  ↓
Attention Weights
  ↓
Multiply by V
  ↓
Output
```

---

# 11. What Does the Final Output Represent?

The final output is a **context-aware representation**.

Instead of representing a token using only its own information, the token now contains information gathered from other relevant tokens.

For example:

```text
Before Attention:

"it"
  ↓
Own representation
```

After attention:

```text
"it"
  ↓
Looks at other tokens
  ↓
Calculates relevance
  ↓
Gives higher weight to relevant information
  ↓
Combines information
  ↓
Context-aware representation
```

<mark>**Attention allows the meaning of a token to depend on its surrounding context.**</mark>

---

# 12. Self-Attention

When the **Query, Key and Value come from the same input sequence**, the mechanism is called:

# <mark>Self-Attention</mark>

Example:

```text
"The cat sat on the mat"
```

Every token can interact with other tokens:

```text
The ─────┐
cat ─────┤
sat ─────┤
on ──────┼──→ Self-Attention
the ─────┤
mat ─────┘
```

For example:

```text
"cat" → can attend to "The", "sat", "mat", etc.

"mat" → can attend to "cat", "sat", "the", etc.
```

<mark>**Every token can consider information from the other tokens in the sequence.**</mark>

---

# 13. Attention vs Self-Attention

These terms are related but not exactly identical.

### Attention

A general mechanism:

```text
Query
  ↓
Compare with Keys
  ↓
Calculate relevance
  ↓
Use corresponding Values
  ↓
Produce output
```

### Self-Attention

A specific type of attention where:

```text
Q ← same sequence
K ← same sequence
V ← same sequence
```

Therefore:

> **Self-Attention is Attention applied within the same sequence.**

---

# 14. Why is Attention Important?

Attention allows the model to directly capture relationships between tokens.

For example:

```text
The ────────────────┐
cat ────────────────┤
sat ────────────────┤
on ─────────────────┼──→ Attention
the ────────────────┤
mat ────────────────┘
```

This helps the model capture:

- **Long-range relationships**
- **Word dependencies**
- **Context**
- **Semantic relationships**
- **Relationships between distant tokens**

<mark>**A token does not have to rely only on information passed through neighboring tokens.**</mark>

---

# 15. Attention in a Simple Example

Consider:

> **"The animal didn't cross the road because it was tired."**

When processing **"it"**, the model calculates how relevant the other tokens are.

Conceptually:

```text
animal   → HIGH relevance
road     → lower relevance
because  → lower relevance
was      → lower relevance
tired    → some relevance
```

The model then uses these relevance weights to combine information from the corresponding Value vectors.

<mark>**The model does not simply choose one word. It calculates weights for the tokens and combines their information.**</mark>

---

# 16. Attention Is NOT Just "Looking at Important Words"

A common oversimplification is:

> "Attention finds the most important word."

That is **not completely correct**.

Attention usually produces a **distribution of weights**:

```text
Token 1 → 0.05
Token 2 → 0.15
Token 3 → 0.50
Token 4 → 0.20
Token 5 → 0.10
```

So the model can use information from **multiple tokens simultaneously**.

<mark>**Attention is weighted information aggregation, not simply selecting one word.**</mark>

---

# 17. Complete Attention Mechanism

```text
                    INPUT
                      │
                      ▼
               Token Embeddings
                      │
                      ▼
                     X
                      │
            ┌─────────┼─────────┐
            ▼         ▼         ▼
            Q         K         V
            │         │         │
            └────┬────┘         │
                 ▼              │
                QKᵀ             │
                 │              │
                 ▼              │
               / √dk            │
                 │              │
                 ▼              │
              Softmax           │
                 │              │
                 ▼              │
        Attention Weights       │
                 │              │
                 └──────┬───────┘
                        ▼
                  Weighted Sum
                        │
                        ▼
                     OUTPUT
```

---

# 18. The Four Main Operations

## 1. QKᵀ

```text
QKᵀ
```

<mark>Calculates similarity/relevance between Queries and Keys.</mark>

---

## 2. Scaling

```text
QKᵀ / √dk
```

<mark>Controls the magnitude of the scores.</mark>

---

## 3. Softmax

```text
Softmax(QKᵀ / √dk)
```

<mark>Converts scores into normalized attention weights.</mark>

---

## 4. Multiply by V

```text
Softmax(QKᵀ / √dk)V
```

<mark>Uses the weights to combine information from the Value vectors.</mark>

---

# 19. The Most Important Formula

> ## <mark>Scaled Dot-Product Attention</mark>

```text
┌──────────────────────────────────────────────┐
│                                              │
│                 QKᵀ                         │
│ Attention = Softmax(──────) V               │
│                  √dk                        │
│                                              │
└──────────────────────────────────────────────┘
```

Remember:

```text
QKᵀ
 ↓
Similarity

/ √dk
 ↓
Scaling

Softmax
 ↓
Attention Weights

× V
 ↓
Weighted Information

= Output
```

---

# 20. One-Line Definition

<mark>**Attention Mechanism is a mechanism that computes relevance between a Query and Keys, converts those relevance scores into weights, and uses those weights to combine the corresponding Values into a context-aware representation.**</mark>

---

# 21. Final Mental Model

Think of Attention like this:

```text
              "What am I looking for?"
                       │
                       ▼
                     QUERY
                       │
                       ▼
              Compare with KEYS
                       │
                       ▼
                "How relevant?"
                       │
                       ▼
               Attention Scores
                       │
                       ▼
                    Softmax
                       │
                       ▼
               Attention Weights
                       │
                       ▼
          Collect information from
                  the VALUES
                       │
                       ▼
             Contextual OUTPUT
```

### The easiest way to remember it:

> **Q asks → K matches → V provides → Attention combines**

# 🔑 Key Takeaways

- **Attention** allows a model to focus on relevant information.
- **Query (Q)** represents what the current token is looking for.
- **Key (K)** represents what each token can be matched on.
- **Value (V)** contains the information that can actually be passed forward.
- **QKᵀ** calculates similarity/relevance scores.
- **√dk** scales the scores.
- **Softmax** converts scores into attention weights.
- The weights are used to create a **weighted combination of Values**.
- **Self-Attention** occurs when Q, K and V come from the same sequence.
- The result is a **context-aware representation**.
- <mark>**Attention does not simply select the most important word — it calculates how much information should be gathered from each token.**</mark>

```text
QUERY  → What am I looking for?
KEY    → What do you contain?
VALUE  → What information can you give me?

Q × Kᵀ → How relevant?
Softmax → How much attention?
× V    → What information should I take?

             ↓

       CONTEXT-AWARE OUTPUT
```

> <mark>**ATTENTION = Relevance-based Information Aggregation**</mark>


# Attention Mechanism — Context Vector

$$
\boxed{
C_i = \sum_{j=1}^{n} \alpha_{ij}h_j
}
$$

### Meaning of Each Term

- **\(C_i\)** → **Context vector for the current token \(i\)**  
  Represents the combined information that token \(i\) receives from the sequence.

- **\(\alpha_{ij}\)** → **Attention weight**  
  Represents **how much attention token \(i\) gives to token \(j\)**.

- **\(h_j\)** → **Hidden representation/vector of token \(j\)**  
  Represents the **information contained in token \(j\)**.

- **\(j\)** → **Index of the token being attended to**  
  It tells us **which token's information is being considered**.

- **\(n\)** → **Total number of tokens**  
  The summation considers all tokens from **\(j=1\) to \(j=n\)**.

### Formula in Words

$$
\boxed{
\text{Context Vector}
=
\sum
(\text{Attention Weight} \times \text{Hidden Representation})
}
$$

> **In simple words:** For the current token \(i\), take every token's hidden representation \(h_j\), multiply it by its attention weight \(\alpha_{ij}\), and sum them to get the context vector \(C_i\).
