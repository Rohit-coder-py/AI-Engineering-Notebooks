">
🤔 Why Do We Need RNN?
</h2>

Traditional ANN treats inputs mostly as independent.

For example:

```text
Input 1 ──┐
Input 2 ──┼──> ANN ──> Output
Input 3 ──┘
```

It doesn't naturally maintain information about **what came before**.

RNN introduces a **memory**.

```text
Previous information
        ↓
Input → RNN → Hidden State
        ↓
      Output
```

The hidden state carries information from previous time steps.

---

<h2 style="
color: #943126;
border-bottom: 4px solid #EC7063;
padding-bottom: 8px;
font-weight: 800;
">
🧠 Core Idea of RNN
</h2>

At every time step, RNN takes:

1. Current input `xₜ`
2. Previous hidden state `hₜ₋₁`

and produces:

1. New hidden state `hₜ`
2. Output `yₜ`

```text
          hₜ₋₁
             ↓
xₜ ──────> [ RNN ] ──────> hₜ
             │
             ↓
            yₜ
```

The important idea:

> **RNN remembers previous information through its hidden state.**

---

<h2 style="
color: #154360;
border-bottom: 4px solid #5DADE2;
padding-bottom: 8px;
font-weight: 800;
">
🔄 RNN Unrolled Through Time
</h2>

Suppose we have:

```text
I love machine learning
```

RNN processes one word at a time:

```text
x₁        x₂          x₃          x₄
I    →    love   →    machine →   learning
↓         ↓            ↓            ↓
RNN       RNN          RNN          RNN
↓         ↓            ↓            ↓
h₁   →    h₂     →     h₃     →     h₄
```

Each step receives information from the previous step.

```text
h₁ → h₂ → h₃ → h₄
```

So:

```text
h₄
```

contains information influenced by:

```text
I + love + machine + learning
```

<h2 style="
color: #6E2C00;
border-bottom: 4px solid #EB984E;
padding-bottom: 8px;
font-weight: 800;
">
🧮 RNN Mathematical Intuition
</h2>

The basic equation is:

hₜ = f(Wₓₕxₜ + Wₕₕhₜ₋₁ + b)

Where:

* `xₜ` → current input
* `hₜ₋₁` → previous hidden state
* `hₜ` → current hidden state
* `Wₓₕ` → input → hidden weights
* `Wₕₕ` → hidden → hidden weights
* `b` → bias
* `f` → activation function, commonly `tanh`

Output:

yₜ = Wₕᵧhₜ + bᵧ

<h2 style="

">
🔁 RNN Weight Sharing
</h2>

One important property:

> **The same weights are reused at every time step.**

```text
       Same RNN parameters
       ↓      ↓      ↓

x₁ → [RNN] → [RNN] → [RNN] → [RNN]
       ↓       ↓       ↓       ↓
      h₁      h₂      h₃      h₄
```

This makes RNN much more parameter-efficient than creating a separate neural network for every time step.

">
🚨 Main Problem with Basic RNN
</h2>

Basic RNN has difficulty remembering information from **very long sequences**.

Example:

```text
The cat ................. was hungry.
```

If the relevant information is very far away, the basic RNN may lose it.

This happens mainly because of:

### ❌ Vanishing Gradient

During Backpropagation Through Time (BPTT), gradients can become extremely small.

```text
Large sequence
     ↓
Gradient × Gradient × Gradient × ...
     ↓
Very small gradient
     ↓
Early information is poorly learned
```

Result:

> RNN struggles with **long-term dependencies**.

---

color: #0B5345;
border-bottom: 4px solid #48C9B0;
padding-bottom: 8px;
font-weight: 800;
">
💥 Exploding Gradient
</h2>

The opposite problem can also occur.

Gradients become extremely large:

```text
Gradient × Gradient × Gradient
              ↓
       Very large value
              ↓
    Exploding Gradient
```

Common solutions include:

* Gradient clipping
* Better architectures such as LSTM/GRU

⚙️ RNN Training
</h2>

RNN is trained using:

### Backpropagation Through Time (BPTT)

RNN is essentially **unrolled through time** and backpropagation is performed across those time steps.

```text
x₁ → RNN → h₁ → RNN → h₂ → RNN → h₃
                       ↓
                    Loss

              ← Backpropagation
```

The gradients flow backward through the sequence.

This is why long sequences can cause:

* Vanishing gradients
* Exploding gradients

# <span style="color:#1B4F72;">What is RNN?</span>

> **Recurrent Neural Network (RNN)** is a type of neural network designed for processing **sequential data**.

Unlike traditional **feedforward neural networks**, which process inputs independently, RNNs maintain a **memory of previous inputs** using **loops in their architecture**.

### <span style="color:#2980B9;">Why RNN?</span>

This makes RNNs well-suited for tasks where **context and order matter**, such as:

- **Time Series Forecasting**
- **Speech Recognition**
- **Text Generation**

> **RNN = Sequential Data + Memory of Previous Inputs**

````python
import pandas as pd
import numpy as np 
````

````python
df = pd.read_csv(r"C:\Users\rohit\Downloads\100_Unique_QA_Dataset.csv")

df.head()
````

````python
#now we have to convert this dataset into numbers 


#tokenize

def tokenize(txt):
    txt = txt.lower()
    txt = txt.replace('?','')
    txt = txt.replace(',','')
    return txt.split()


tokenize("what is the boiliNg point of water?")
````

````python
#now vocabulary building

vocab  = {'<UNK>' : 0}

def build_vocab(row):
    tokenized_ques = tokenize(row['question'])
    tokenized_ans =  tokenize(row['answer'])
    
    merged_list = tokenized_ques + tokenized_ans

    for token in merged_list:
        if token not in vocab:
            vocab[token] = len(vocab)
    
````

````python
df.apply(build_vocab,axis = 1)
````

````python
vocab
````

````python
#next step is to convert vocab into numerical indices 
def text_to_indices(text, vocab):

  indexed_text = []

  for token in tokenize(text):

    if token in vocab:
      indexed_text.append(vocab[token])
    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text
    
````

````python
text_to_indices("What is chatgpt", vocab)
````

````python
import torch
from torch.utils.data import Dataset, DataLoader
````

````python
class QADataset(Dataset):

  def __init__(self, df, vocab):
    self.df = df
    self.vocab = vocab

  def __len__(self):
    return self.df.shape[0]

  def __getitem__(self, index):

    numerical_question = text_to_indices(self.df.iloc[index]['question'], self.vocab)
    numerical_answer = text_to_indices(self.df.iloc[index]['answer'], self.vocab)

    return torch.tensor(numerical_question), torch.tensor(numerical_answer)
````

````python
dataset = QADataset(df, vocab)
````

````python
dataloader = DataLoader(dataset, batch_size=1, shuffle=True)
````

````python
for question, answer in dataloader:
  print(question, answer[0])
````

````python
import torch.nn as nn
````

````python
class SimpleRNN(nn.Module):

  def __init__(self, vocab_size):
    super().__init__()
    self.embedding = nn.Embedding(vocab_size, embedding_dim=50)
    self.rnn = nn.RNN(50, 64, batch_first=True)
    self.fc = nn.Linear(64, vocab_size)

  def forward(self, question):
    embedded_question = self.embedding(question)
    hidden, final = self.rnn(embedded_question)
    output = self.fc(final.squeeze(0))

    return output

x = nn.Embedding(324, embedding_dim=50)
y = nn.RNN(50, 64, batch_first=True)
z = nn.Linear(64, 324)

a = dataset[0][0].reshape(1,6)
print("shape of a:", a.shape)
b = x(a)
print("shape of b:", b.shape)
c, d = y(b)
print("shape of c:", c.shape)
print("shape of d:", d.shape)

e = z(d.squeeze(0))

print("shape of e:", e.shape)

learning_rate = 0.001
epochs = 20

model = SimpleRNN(len(vocab))

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

# training loop

for epoch in range(epochs):

  total_loss = 0

  for question, answer in dataloader:

    optimizer.zero_grad()

    # forward pass
    output = model(question)

    # loss -> output shape (1,324) - (1)
    loss = criterion(output, answer[0])

    # gradients
    loss.backward()

    # update
    optimizer.step()

    total_loss = total_loss + loss.item()

  print(f"Epoch: {epoch+1}, Loss: {total_loss:4f}")

def predict(model, question, threshold=0.5):

  # convert question to numbers
  numerical_question = text_to_indices(question, vocab)

  # tensor
  question_tensor = torch.tensor(numerical_question).unsqueeze(0)

  # send to model
  output = model(question_tensor)

  # convert logits to probs
  probs = torch.nn.functional.softmax(output, dim=1)

  # find index of max prob
  value, index = torch.max(probs, dim=1)

  if value < threshold:
    print("I don't know")

  print(list(vocab.keys())[index])

predict(model, "What is the largest planet in our solar system?")

list(vocab.keys())[7]
````
