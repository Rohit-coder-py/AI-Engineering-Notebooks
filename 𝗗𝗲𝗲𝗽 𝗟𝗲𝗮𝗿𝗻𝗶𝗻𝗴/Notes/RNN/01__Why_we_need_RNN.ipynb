## 1. What is RNN?

**RNN** stands for **Recurrent Neural Network**.

It is a type of neural network specifically designed to work with **sequential data**.

Unlike a traditional ANN, an RNN is designed to consider the **order and previous information** present in a sequence.

### Common Areas Where RNNs Are Used

* Natural Language Processing (NLP)
* Time Series
* Speech
* Text
* DNA sequences
* Sequence prediction

">What is Sequential Data?</h2>
</div>

Sequential data is data where **the order of elements matters**.

In other words:

> The current element can depend on what appeared before it.

---

## Example 1 — Text

Consider:

```text
My name is Nitish
```

The words occur in a particular sequence:

```text
My → name → is → Nitish
```

Changing the order can change the meaning.

Therefore, **text is sequential data**.

When humans read a sentence, we process words sequentially and retain information about what came before.

---

## Example 2 — Time Series

Consider a graph showing values over different years:

```text
2001 → 2002 → 2003 → 2004 → ...
```

The previous values can provide information about future values.

Therefore:

> **Time-series data is sequential data.**

---

## Example 3 — Audio

Audio can be represented as a waveform:

```text
Amplitude
   ↑
   │    /\      /\
   │   /  \    /  \
   │__/    \__/    \__
   └──────────────────→ Time
```

Since the waveform changes over time, audio is also sequential data.

---

## Example 4 — DNA

A DNA sequence is another example:

```text
A → T → G → C → A → ...
```

The sequence/order of elements matters.

">Problem 1 — Text Cannot Directly Be Given to a Neural Network</h2>
</div>

A neural network works with numerical values.

But text consists of words:

```text
My name is Nitish
```

Therefore, we first need to convert words into numerical representations.

One simple method is:

### One-Hot Encoding

Suppose our vocabulary contains:

```text
My
Mike
Name
Is
Nitish
...
```

Each word can be represented as a vector.

For example:

```text
My     → [1, 0, 0, 0, 0, ...]
Mike   → [0, 1, 0, 0, 0, ...]
Name   → [0, 0, 1, 0, 0, ...]
```

Thus:

> **Words → Numerical Vectors**

">Problem 2 — Variable Input Size</h2>
</div>

Consider two sentences:

```text
Sentence 1:
My name is Nitish
```

```text
Sentence 2:
My name is Nitish and I am learning Deep Learning
```

The number of words is different.

Therefore, their input dimensions will also be different if we simply concatenate all word vectors.

For example:

```text
Sentence 1
5 words × vocabulary size

Sentence 2
8 words × vocabulary size
```

A standard ANN expects a **fixed input size**.

So variable-length sequences create a major problem.

">Problem 3 — Zero Padding</h2>
</div>

One possible solution is **zero padding**.

Suppose the maximum sequence length is:

```text
5 words
```

But our sentence contains only:

```text
3 words
```

We can add two zero vectors:

```text
Word 1
Word 2
Word 3
[0]
[0]
```

Now every input has the same size.

### Problem with Zero Padding

This is somewhat of a workaround rather than an ideal solution.

If the vocabulary is very large and sequences are long, the number of unnecessary values can become huge.

This results in:

* More computation
* More parameters
* More memory usage
* Unnecessary processing

**📘 What is Zero Padding — In Simple Words**

**Zero padding** simply means adding extra `0`s to the *shorter* sequences so that every sequence in a batch ends up with the **same length**.

```text
Longest sentence in the batch = 5 words   →  this becomes the target length

"I love ML"           (3 words)
        ↓  pad with 2 zero-vectors
[I] [love] [ML] [0] [0]      ← now length 5 ✅
```

It's called *zero* padding because the filler values added are literally zero vectors — they carry **no real information**, they only exist to make the shapes match so the data can be stacked into a fixed-size tensor and fed to the network.

">Most Important Problem — Loss of Sequence Information</h2>
</div>

This is the **most important reason** for using RNNs.

Suppose we have:

```text
I → love → this → movie
```

The order matters.

Compare:

```text
I love this movie
```

with:

```text
Movie this love I
```

The same words are present, but the meaning is completely different.

---

## What Happens in ANN?

If all words are given to an ANN at the same time:

```text
Word 1 ─┐
Word 2 ─┤
Word 3 ─┤ → ANN → Output
Word 4 ─┘
```

The ANN does not naturally maintain information about:

```text
What came first?
What came next?
What came previously?
```

Therefore, it loses important **sequence information**.

">RNN's Main Idea</h2>
</div>

RNNs were introduced to specifically address the challenges associated with sequential data.

The key idea is:

> **An RNN processes sequential information while maintaining information from previous time steps.**

Conceptually:

```text
Input₁ → RNN → Hidden State₁
                  ↓
Input₂ → RNN → Hidden State₂
                  ↓
Input₃ → RNN → Hidden State₃
                  ↓
Input₄ → RNN → Output
```

The hidden state carries information from previous steps.

This gives RNNs a form of **memory**.

">RNN and Memory</h2>
</div>

The important concept is:

### Hidden State

The RNN maintains a hidden state that contains information from previous time steps.

Conceptually:

```text
Previous Information
        ↓
   Hidden State
        ↓
Current Input
        ↓
      RNN
        ↓
New Hidden State
```

So the model can use:

```text
Previous Information + Current Input
```

to produce the next hidden state/output.

## 15.1 Sentiment Analysis

### Input

A text/review:

```text
The movie was amazing!
```

### Output

```text
Positive
```

Or:

```text
The movie was terrible.
```

### Output

```text
Negative
```

RNNs can process the sequence of words and predict the sentiment.

---

## 15.2 Sentence Completion

Example:

```text
I hope ...
```

The model predicts possible next words.

Conceptually:

```text
I → hope → ...
             ↓
          prediction
```

This idea is used in applications such as:

* Email suggestions
* Keyboard next-word prediction
* Text completion

---

## 15.3 Image Caption Generation

The system receives an image and generates a textual description.

Conceptually:

```text
Image
  ↓
Image Model
  ↓
Features
  ↓
RNN
  ↓
Text Caption
```

Example:

```text
Input: Image of a person riding a bike

Output:
"A person is riding a bicycle."
```

This combines computer vision with sequence generation.

---

## 15.4 Machine Translation

Machine translation converts one language into another.

Example:

```text
Hindi
  ↓
RNN-based Model
  ↓
English
```

The model needs to understand the sequence of words in the source language and generate a sequence in the target language.

---

## 15.5 Question Answering

A model can receive a paragraph and answer questions related to that paragraph.

Conceptually:

```text
Paragraph
    +
Question
    ↓
Sequence Model
    ↓
Answer
```

This demonstrates the importance of understanding context.

---

## 15.6 Time-Series Forecasting

RNNs can also be used for time-series problems.

Examples:

* Stock-related forecasting
* Weather-related sequences
* Sales forecasting
* Sensor data
* Temperature prediction

The basic idea is:

```text
Past Values
     ↓
   RNN
     ↓
Future Prediction
```

---

## 15.7 Speech Classification

Audio/speech is sequential because it changes over time.

RNNs can therefore be used for tasks involving:

* Speech
* Audio sequences
* Speech classification

color:#7E5109;
font-weight:800;
font-size:21px;
">Backpropagation Through Time (BPTT)</h2>
</div>

In normal neural networks, backpropagation is used to update weights.

For RNNs, the network operates across multiple time steps.

Therefore, the backpropagation process is performed through the sequence of time steps.

This is called:

> **Backpropagation Through Time (BPTT)**

It is an important concept for understanding how RNNs learn.

---

<div style="
background:#EBF5FB;
border:1px solid #AED6F1;
border-left: 5px solid #1A5276;
border-radius:8px;
padding:14px 18px;
margin:14px 0;
">

**💡 BPTT Explained Simply**

Think of BPTT as normal backpropagation, but **unrolled across every time step** of the sequence:

```text
x₁ → RNN → h₁ → RNN → h₂ → RNN → h₃ → Loss
                                    ↓
        gradients flow backward through h₃ → h₂ → h₁ → x₁
```

Because the *same* weights were reused at every step, the gradient for those weights is the **sum of the gradients contributed at every time step**. The longer the sequence, the more steps the gradient has to travel backward through — which is exactly why long sequences are where vanishing/exploding gradients start to bite.

</div>

font-size:21px;
">Problems with Basic RNN</h2>
</div>

Although RNNs solve important problems, basic RNNs also have limitations.

Two major problems are:

### 1. Vanishing Gradient

Gradients can become extremely small during training.

This makes it difficult for the model to learn long-term dependencies.

### 2. Exploding Gradient

Gradients can become extremely large.

This can make training unstable.

---

<div style="
background:#FDEDEC;
border:1px solid #F5B7B1;
border-left: 5px solid #943126;
border-radius:8px;
padding:14px 18px;
margin:14px 0;
">

**💡 Why Do Gradients Vanish or Explode?**

During BPTT, the same weight matrix gets multiplied into the gradient **once per time step** (chain rule). Multiplying the same number again and again has only two possible fates:

```text
value < 1  →  0.9 × 0.9 × 0.9 × ... × 0.9   →  shrinks toward 0   (Vanishing)
value > 1  →  1.5 × 1.5 × 1.5 × ... × 1.5   →  blows up           (Exploding)
```

A **vanishing** gradient means early time steps barely get updated → the model "forgets" long-range context.
An **exploding** gradient means the updates become huge and unstable → training diverges (often seen as `NaN` losses).

</div>

">LSTM</h2>
</div>

**LSTM** stands for:

> **Long Short-Term Memory**

LSTM is an improved recurrent architecture designed to better handle long-term dependencies.

It was introduced as a solution to important limitations of traditional RNNs.

---

<div style="
background:#EAFAF1;
border:1px solid #A9DFBF;
border-left: 5px solid #186A3B;
border-radius:8px;
padding:14px 18px;
margin:14px 0;
">

**💡 LSTM — A Slightly Deeper Look**

LSTM adds a separate **cell state** (think of it as a conveyor belt running through all time steps) alongside the hidden state. Three gates decide what happens to it at every step:

- **Forget Gate** → what old information should be thrown away?
- **Input Gate** → what new information should be added?
- **Output Gate** → what should be exposed as the current output?

Because the cell state can carry information across many steps with only small, gated modifications, gradients survive much longer — which is exactly what basic RNNs struggle to do.

</div>

">GRU</h2>
</div>

**GRU** stands for:

> **Gated Recurrent Unit**

GRU is another improved recurrent architecture.

It provides a mechanism for controlling what information should be retained and what information should be discarded.

---

<div style="

**💡 GRU — A Slightly Deeper Look**

GRU reaches a similar goal as LSTM but with a **simpler design** — no separate cell state, only the hidden state, controlled by two gates:

- **Update Gate** → how much of the past hidden state to keep vs. replace
- **Reset Gate** → how much of the past hidden state to forget when computing new candidate content

Fewer gates → fewer parameters → often faster to train, while still handling long-term dependencies noticeably better than a basic RNN.

font-size:21px;
">Bidirectional RNN</h2>
</div>

A **Bidirectional RNN** processes information in two directions:

```text
Past → Future
Future → Past
```

This allows the model to use information from both previous and future positions in a sequence when appropriate.

---

<div style="
background:#F4F6F6;
border:1px solid #D5D8DC;
border-left: 5px solid #1B2631;
border-radius:8px;
padding:14px 18px;
margin:14px 0;
">

**💡 Why Go Bidirectional?**

A normal (forward) RNN only knows what came **before** the current word. But sometimes the meaning only becomes clear from what comes **after** it:

```text
"The bank was closed for the holiday."      → riverbank? financial bank? (ambiguous alone)
"The bank raised interest rates today."     → future word "interest rates" resolves it
```

A Bidirectional RNN runs **two RNNs** — one left→right, one right→left — and combines both hidden states, so every position has context from *both directions*.

⚠️ Trade-off: since it needs the **entire sequence** before it can process anything, it can't be used for real-time / streaming predictions (e.g. live speech-to-text).

">Key Takeaways ⭐</h2>
</div>

### Remember these points:

1. **RNN = Recurrent Neural Network**
2. RNNs are designed for **sequential data**.
3. **Text, time series, audio, speech, and DNA** are examples of sequential data.
4. The **order of data matters** in sequential problems.
5. Traditional ANNs do not naturally maintain sequence information.
6. Variable-length sequences are difficult for fixed-input ANN architectures.
7. Zero padding can solve input-size problems but may introduce unnecessary computation.
8. RNNs maintain information through their **hidden state**.
9. RNNs have historically been important in **NLP and sequence-related tasks**.
10. Important applications include:

* Sentiment Analysis
* Sentence Completion
* Image Captioning
* Machine Translation
* Question Answering
* Time-Series Forecasting
* Speech Classification

11. **BPTT** is used to train RNNs through time.
12. Basic RNNs suffer from **vanishing and exploding gradients**.
13. **LSTM and GRU** were developed to address important limitations of basic RNNs.
14. Advanced topics include **Deep RNNs and Bidirectional RNNs**.

">🔥 The Core Idea</h2>

```text
Traditional ANN

Input₁ ─┐
Input₂ ─┤
Input₃ ─┤ → ANN → Output
Input₄ ─┘

❌ Sequence information is not naturally maintained.


RNN

Input₁ → RNN → Hidden₁
                 ↓
Input₂ → RNN → Hidden₂
                 ↓
Input₃ → RNN → Hidden₃
                 ↓
Input₄ → RNN → Output

✅ Previous information is carried forward.
```

> **The main reason we need RNNs: sequence matters, and RNNs are designed to retain information across the sequence.**
