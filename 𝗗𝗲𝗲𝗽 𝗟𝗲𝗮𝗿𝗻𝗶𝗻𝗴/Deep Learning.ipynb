<h2 style="font-size:30px; color:#374151;">
    <strong>What is Deep Learning?</strong>
</h2>

<p>
<strong>Deep Learning</strong> is a <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>subset of Machine Learning</strong></span> that teaches <strong>computers</strong> to <strong>learn patterns from data</strong> using <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>Artificial Neural Networks</strong></span>, which are inspired by the way the <strong>human brain</strong> works.
</p>

<p>
Instead of <strong>manually telling</strong> the computer what <strong>features</strong> to look for, <strong>Deep Learning</strong> <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>automatically learns those features from the data</strong></span>.
</p>

<h1 style="margin-top:0; color:#1E3A8A; font-size:34px;">
    <span style="color:#2563EB;">Perceptron</span>
</h1>

<p style="font-size:18px; line-height:1.9; color:#374151;">

The <span style="color:#DC2626; font-weight:bold;">Perceptron</span> is the <span style="color:#2563EB; font-weight:bold;">simplest type of artificial neuron</span>, introduced by <span style="color:#7C3AED; font-weight:bold;">Frank Rosenblatt</span> in <span style="color:#EA580C; font-weight:bold;">1958</span>.

It is the <span style="color:#059669; font-weight:bold;">foundation of neural network and Deep learning.</span>

It mimics how a <span style="color:#DB2777; font-weight:bold;">biological neuron</span> works taking <span style="color:#2563EB; font-weight:bold;">inputs</span>, processing them, and giving an <span style="color:#DC2626; font-weight:bold;">output</span>.
