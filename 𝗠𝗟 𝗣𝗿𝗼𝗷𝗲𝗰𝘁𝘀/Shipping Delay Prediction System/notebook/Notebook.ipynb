```python
# boxplots for outliers detection

import os
import matplotlib.pyplot as plt
import seaborn as sns

save_path = r"C:\Users\shobh\Downloads\AI and DATA SCIENCE\𝗠𝗟 𝗣𝗿𝗼𝗷𝗲𝗰𝘁𝘀\Shipping Delay Prediction System\graphs"

# Boxplots for Outlier Detection

numerical_cols = [
    "Customer_care_calls",
    "Customer_rating",
    "Cost_of_the_Product",
    "Prior_purchases",
    "Discount_offered",
    "Weight_in_gms"
]

for col in numerical_cols:

    plt.figure(figsize=(8, 4))

    sns.boxplot(
        x=df[col],
        color="skyblue"
    )

    plt.title(f"Boxplot of {col}", fontsize=16, fontweight="bold", pad=15)

    plt.xlabel(col, fontsize=12)

    plt.grid(axis="x", linestyle="--", alpha=0.5)

    plt.tight_layout()

    plt.savefig(
        os.path.join(save_path, f"{col}_Boxplot.png"),
        dpi=300, bbox_inches="tight")

    plt.show()
```
