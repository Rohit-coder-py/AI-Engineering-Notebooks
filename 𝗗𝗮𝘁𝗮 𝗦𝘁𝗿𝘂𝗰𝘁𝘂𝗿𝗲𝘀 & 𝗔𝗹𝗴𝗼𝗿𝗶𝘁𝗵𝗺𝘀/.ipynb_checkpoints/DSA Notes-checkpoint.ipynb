<h2 style="font-size:30px; color:#374151;">
    <strong>What is DSA?</strong>
</h2>

<p>
<span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>DSA = Data Structures + Algorithms</strong></span>
</p>

<p>
<strong>Data Structure</strong> = <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>How you organize data.</strong></span>
</p>

<p>
<strong>Algorithm</strong> = <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>How you solve a problem using that data.</strong></span>
</p>

<h2 style="font-size:30px; color:#374151;">
    <strong>What are Data Structures?</strong>
</h2>

<p>
A <strong>data structure</strong> is simply a <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>way to store information efficiently</strong></span>.
</p>

| Category         | Data Structure            |
| ---------------- | ------------------------- |
| **Linear**       | Array                     |
|                  | Linked List               |
|                  | Stack                     |
|                  | Queue                     |
|                  | Deque                     |
| **Hash-based**   | Hash Map (Dictionary)     |
|                  | Hash Set                  |
| **Hierarchical** | Tree                      |
|                  | Binary Tree               |
|                  | Binary Search Tree (BST)  |
|                  | Heap                      |
|                  | Trie                      |
| **Network**      | Graph                     |
| **Specialized**  | Priority Queue            |
|                  | Disjoint Set (Union-Find) |
|                  | Segment Tree              |
|                  | Fenwick Tree (BIT)        |


<h2 style="font-size:30px; color:#374151;">
    <strong>What is an Algorithm?</strong>
</h2>

<p>
An <strong>algorithm</strong> is a <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>step-by-step method to solve a problem</strong></span>.
</p>
