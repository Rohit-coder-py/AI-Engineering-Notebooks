<h2 style="font-size:32px; color:#1E3A8A; border-bottom:3px solid #DBEAFE; padding-bottom:8px; margin-top:20px; margin-bottom:16px;">
    <strong>What is Big O Notation?</strong>
</h2>

<p>
    <span style="background-color:#E0F2FE; padding:3px 8px; border-radius:6px;">
        <strong>Big O Notation</strong>
    </span>
    is a way to measure how efficient an algorithm is as the input size grows.
</p>

<p><strong>It mainly tells us:</strong></p>

<div style="margin-left:20px; line-height:1.9;">
<span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;">
    <strong>How the execution time grows (Time Complexity)</strong>
</span>

<br><br>
<span style="background-color:#FCE7F3; padding:2px 6px; border-radius:4px;">
    <strong>How the memory usage grows (Space Complexity)</strong>
</span>

</div>

```python
syntax :

O(n) = order of n

where O = Order of Growth


example : 

O(1) → Order of Constant Time
O(log n) → Order of Logarithmic Time
O(n) → Order of Linear Time
O(n log n) → Order of Linearithmic Time
O(n²) → Order of Quadratic Time
O(2ⁿ) → Order of Exponential Time
```

<h2 style="font-size:30px; color:#374151;">
    <strong>2. O(log n) — Logarithmic Time</strong>
</h2>

<p>
<strong>O(log n)</strong> means that as the <strong>input size</strong> grows, the <span style="background-color:#E0F2FE; padding:2px 6px; border-radius:4px;"><strong>number of operations grows very slowly</strong></span>.
</p>

<p>
Instead of checking <strong>every element</strong>, the algorithm <span style="background-color:#DCFCE7; padding:2px 6px; border-radius:4px;"><strong>removes a large portion of the remaining data</strong></span> after each step.
</p>

<p>
Usually, it <span style="background-color:#FEF3C7; padding:2px 6px; border-radius:4px;"><strong>cuts the search space in half</strong></span> every iteration.
</p>

| Linear Search                    | Binary Search                     |
| -------------------------------- | --------------------------------- |
| Works on any list                | Requires a sorted list            |
| Checks every element one by one  | Checks the middle element first   |
| Time Complexity: **O(n)**        | Time Complexity: **O(log n)**     |
| Simple but slower for large data | Much faster for large sorted data |
